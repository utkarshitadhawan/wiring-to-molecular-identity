```python
# ============================================================
# CHECKPOINT 3
# Inspect repository data structure
# ============================================================

inst_api_url = (
    f"https://api.github.com/repos/"
    f"{NT_REPO_OWNER}/{NT_REPO_NAME}/contents/inst"
)

response = requests.get(inst_api_url, timeout=30)
response.raise_for_status()

inst_contents = response.json()

print("Contents of repository /inst:")
print("=" * 70)

for item in inst_contents:
    print(f"{item['type']:<10} {item['name']}")
```

```python
# ============================================================
# CHECKPOINT 3
# Locate cross-connectome mapping file
# ============================================================

crossmatch_candidates = []

def search_github_tree(path=""):
    """
    Recursively search the repository for files containing
    'cross' and 'matching' in the filename.
    """

    api_url = (
        f"https://api.github.com/repos/"
        f"{NT_REPO_OWNER}/{NT_REPO_NAME}/contents/{path}"
    )

    r = requests.get(api_url, timeout=30)
    r.raise_for_status()

    items = r.json()

    for item in items:
        name_lower = item["name"].lower()

        if item["type"] == "file":
            if (
                "cross" in name_lower
                and "match" in name_lower
            ):
                crossmatch_candidates.append({
                    "name": item["name"],
                    "path": item["path"],
                    "download_url": item.get("download_url")
                })

        elif item["type"] == "dir":
            search_github_tree(item["path"])


search_github_tree()

print("Cross-matching candidates found:")
print("=" * 70)

for item in crossmatch_candidates:
    print("Name :", item["name"])
    print("Path :", item["path"])
    print("URL  :", item["download_url"])
    print("-" * 70)
```

```python
# ============================================================
# CHECKPOINT 3
# Download cross-connectome mapping
# ============================================================

if len(crossmatch_candidates) == 0:
    raise FileNotFoundError(
        "No cross-connectome mapping file was found."
    )

if len(crossmatch_candidates) > 1:
    print("Multiple candidates found:")
    for item in crossmatch_candidates:
        print(item["path"])

    raise RuntimeError(
        "More than one cross-matching candidate was found. "
        "Inspect the list above before continuing."
    )

crossmatch_info = crossmatch_candidates[0]

crossmatch_url = crossmatch_info["download_url"]

response = requests.get(
    crossmatch_url,
    timeout=30
)

response.raise_for_status()

crossmatch_df = pd.read_csv(
    StringIO(response.text)
)

print("✓ Cross-connectome mapping downloaded.")
print()
print("Source path:")
print(crossmatch_info["path"])

print()
print("Shape:")
print(crossmatch_df.shape)

print()
print("Columns:")
print(crossmatch_df.columns.tolist())
```

```python
# ============================================================
# CHECKPOINT 3
# Inspect cross-connectome mapping schema
# ============================================================

print("CROSS-CONNECTOME MAPPING")
print("=" * 70)

print("Shape:", crossmatch_df.shape)

print("\nColumn names:")
for i, column in enumerate(
    crossmatch_df.columns,
    start=1
):
    print(f"{i:>2}. {column}")

print("\nFirst 15 rows:")
display(crossmatch_df.head(15))
```

```python
# ============================================================
# CHECKPOINT 3
# Identify HemiBrain-related columns
# ============================================================

hemibrain_columns = [
    column
    for column in crossmatch_df.columns
    if "hemi" in column.lower()
]

print("HemiBrain-related columns:")
print("=" * 60)

for column in hemibrain_columns:
    print("•", column)
```

```python
# ============================================================
# CHECKPOINT 3
# Inspect HemiBrain mapping coverage
# ============================================================

for column in hemibrain_columns:

    print("\n" + "=" * 70)
    print(column)
    print("=" * 70)

    values = crossmatch_df[column].dropna()

    print("Non-missing:", len(values))
    print("Unique:", values.nunique())

    print("\nFirst 30 unique values:")
    display(
        values.astype(str)
        .drop_duplicates()
        .head(30)
    )
```

```python
# ============================================================
# CHECKPOINT 3
# Compare existing HemiBrain olfactory types
# ============================================================

print("Existing HemiBrain olfactory dataset")
print("=" * 70)

olfactory_types = (
    upn_df["type"]
    .dropna()
    .drop_duplicates()
    .sort_values()
    .tolist()
)

print("uPN neurons:", upn_df["bodyid"].nunique())
print("Unique olfactory cell types:", len(olfactory_types))

print("\nOlfactory HemiBrain types:")
for cell_type in olfactory_types:
    print(cell_type)
```

```python
# ============================================================
# CHECKPOINT 3
# Find which mapping column contains our HemiBrain types
# ============================================================

olfactory_type_set = set(olfactory_types)

overlap_results = []

for column in crossmatch_df.columns:

    values = set(
        crossmatch_df[column]
        .dropna()
        .astype(str)
    )

    overlap_count = len(
        olfactory_type_set.intersection(values)
    )

    overlap_results.append({
        "column": column,
        "overlap_with_olfactory_types": overlap_count,
        "unique_values": len(values)
    })

overlap_df = (
    pd.DataFrame(overlap_results)
    .sort_values(
        "overlap_with_olfactory_types",
        ascending=False
    )
)

display(overlap_df)
```

```python
# ============================================================
# CHECKPOINT 3
# Identify exact olfactory types missing from canonical mapping
# ============================================================

canonical_types = set(
    crossmatch_df["cell_type"]
    .dropna()
    .astype(str)
)

hemibrain_types_in_mapping = set(
    crossmatch_df["hemibrain_type"]
    .dropna()
    .astype(str)
)

missing_from_cell_type = sorted(
    olfactory_type_set - canonical_types
)

missing_from_hemibrain_type = sorted(
    olfactory_type_set - hemibrain_types_in_mapping
)

print("OLFACTORY TYPES NOT FOUND IN canonical 'cell_type'")
print("=" * 70)

print("Count:", len(missing_from_cell_type))

for x in missing_from_cell_type:
    print("•", x)

print("\n")
print("OLFACTORY TYPES NOT FOUND IN 'hemibrain_type'")
print("=" * 70)

print("Count:", len(missing_from_hemibrain_type))

for x in missing_from_hemibrain_type:
    print("•", x)
```

```python
# ============================================================
# CHECKPOINT 3
# Search mapping for the missing olfactory types
# ============================================================

for target in missing_from_cell_type:

    print("\n" + "=" * 80)
    print("SEARCHING FOR:", target)
    print("=" * 80)

    matches = crossmatch_df[
        crossmatch_df.astype(str)
        .apply(
            lambda row: row.str.contains(
                target,
                case=False,
                regex=False
            ).any(),
            axis=1
        )
    ]

    if len(matches) == 0:
        print("No direct string match found.")
    else:
        display(matches)
```

```python
# ============================================================
# CHECKPOINT 3
# Extract canonical cell types matching olfactory types
# ============================================================

olfactory_canonical_matches = crossmatch_df[
    crossmatch_df["cell_type"]
    .astype(str)
    .isin(olfactory_type_set)
].copy()

print(
    "Matching rows:",
    len(olfactory_canonical_matches)
)

print(
    "Unique canonical cell types:",
    olfactory_canonical_matches["cell_type"].nunique()
)

display(
    olfactory_canonical_matches[
        [
            "cell_type",
            "hemibrain_type",
            "morphology_group",
            "hemilineage",
            "region",
            "in_hemibrain"
        ]
    ]
)
```

```python
# ============================================================
# CHECKPOINT 3
# Validate HemiBrain status of matched olfactory types
# ============================================================

olfactory_canonical_matches[
    "in_hemibrain"
] = olfactory_canonical_matches[
    "in_hemibrain"
].astype(str).str.lower().eq("true")

print("HemiBrain membership:")
print(
    olfactory_canonical_matches[
        "in_hemibrain"
    ].value_counts(dropna=False)
)

print("\nTypes marked as HemiBrain:")
display(
    olfactory_canonical_matches[
        olfactory_canonical_matches["in_hemibrain"]
    ][
        [
            "cell_type",
            "hemibrain_type",
            "region",
            "in_hemibrain"
        ]
    ]
)
```

```python
# ============================================================
# CHECKPOINT 3
# Connect olfactory cell types to neurotransmitter evidence
# ============================================================

# Use the already-loaded neurotransmitter dataframe
nt_df = nt_data["gt_data.csv"].copy()

NEUROTRANSMITTERS = [
    "acetylcholine",
    "glutamate",
    "gaba",
    "glycine",
    "dopamine",
    "serotonin",
    "octopamine",
    "tyramine",
    "histamine",
    "nitric_oxide"
]

# Keep the molecular database at its original study-level
# structure for now. We are NOT collapsing evidence yet.

olfactory_nt_matches = olfactory_canonical_matches.merge(
    nt_df,
    on="cell_type",
    how="left",
    suffixes=("_mapping", "_nt")
)

print("Merged rows:", len(olfactory_nt_matches))

print(
    "Olfactory canonical types:",
    olfactory_canonical_matches["cell_type"].nunique()
)

print(
    "Types represented in neurotransmitter dataset:",
    olfactory_nt_matches[
        "neurotransmitter_verified_source"
    ].notna().groupby(
        olfactory_nt_matches["cell_type"]
    ).any().sum()
)

display(
    olfactory_nt_matches.head(20)
)
```

```python
# ============================================================
# CHECKPOINT 3
# Count molecular evidence records per olfactory cell type
# ============================================================

evidence_counts = (
    olfactory_nt_matches
    .groupby("cell_type")
    .size()
    .sort_values(ascending=False)
)

print("Molecular evidence records per olfactory cell type:")
print("=" * 70)

display(evidence_counts)
```

```python
# ============================================================
# CHECKPOINT 3
# Inspect neurotransmitter evidence for olfactory types
# ============================================================

evidence_columns = [
    "cell_type",
    "neurotransmitter_verified_source",
    "neurotransmitter_verified_evidence",
    "neurotransmitter_verified_confidence"
] + NEUROTRANSMITTERS

display(
    olfactory_nt_matches[
        evidence_columns
    ].sort_values(
        ["cell_type", "neurotransmitter_verified_confidence"],
        ascending=[True, False]
    )
)
```

```python
# ============================================================
# AUDIT MOLECULAR LABEL CONSISTENCY
# ============================================================

# Neurotransmitter columns in the ground-truth dataset
nt_cols = [
    "acetylcholine",
    "glutamate",
    "gaba",
    "glycine",
    "dopamine",
    "serotonin",
    "octopamine",
    "tyramine",
    "histamine",
    "nitric_oxide"
]

# Working only with the 65 olfactory cell types represented
olf_nt = nt_df[
    nt_df["cell_type"].isin(olfactory_types)
].copy()

print("=" * 70)
print("MOLECULAR LABEL CONSISTENCY AUDIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. Number of evidence records per cell type
# ------------------------------------------------------------

record_counts = (
    olf_nt.groupby("cell_type")
    .size()
    .sort_values(ascending=False)
)

print("\nEvidence records per cell type:")
print(record_counts.to_string())

# ------------------------------------------------------------
# 2. For each cell type, inspect positive neurotransmitter calls
# ------------------------------------------------------------

positive_summary = []

for cell_type, group in olf_nt.groupby("cell_type"):

    positives = {}

    for nt in nt_cols:
        n_positive = (group[nt] == 1).sum()

        if n_positive > 0:
            positives[nt] = int(n_positive)

    positive_summary.append({
        "cell_type": cell_type,
        "n_records": len(group),
        "positive_neurotransmitters": positives
    })

positive_df = pd.DataFrame(positive_summary)

print("\n" + "=" * 70)
print("POSITIVE NEUROTRANSMITTER EVIDENCE")
print("=" * 70)

display(
    positive_df.sort_values(
        ["n_records", "cell_type"],
        ascending=[False, True]
    )
)

# ------------------------------------------------------------
# 3. Detect cell types with multiple positive neurotransmitters
# ------------------------------------------------------------

multi_label = positive_df[
    positive_df["positive_neurotransmitters"].apply(len) > 1
].copy()

print("\n" + "=" * 70)
print("CELL TYPES WITH >1 POSITIVE NEUROTRANSMITTER")
print("=" * 70)

print(f"Count: {len(multi_label)}")

if len(multi_label) > 0:
    display(multi_label)
else:
    print("None.")

# ------------------------------------------------------------
# 4. Detect cell types with conflicting evidence across records
# ------------------------------------------------------------

conflicts = []

for cell_type, group in olf_nt.groupby("cell_type"):

    for nt in nt_cols:

        values = set(group[nt].dropna().unique())

        # A conflict exists if both positive and negative evidence
        # occur for the same transmitter.
        if 1 in values and -1 in values:

            conflicts.append({
                "cell_type": cell_type,
                "neurotransmitter": nt,
                "values_present": sorted(values),
                "n_records": len(group)
            })

conflict_df = pd.DataFrame(conflicts)

print("\n" + "=" * 70)
print("CONFLICTING MOLECULAR EVIDENCE")
print("=" * 70)

print(f"Number of conflicts: {len(conflict_df)}")

if len(conflict_df) > 0:
    display(conflict_df)
else:
    print("No positive-vs-negative conflicts detected.")

# ------------------------------------------------------------
# 5. Save audit
# ------------------------------------------------------------

from pathlib import Path

audit_dir = Path("data/processed")
audit_dir.mkdir(parents=True, exist_ok=True)

audit_path = audit_dir / "neurotransmitter_label_consistency_audit.csv"

conflict_df.to_csv(audit_path, index=False)

print("Saved molecular label audit:")
print(audit_path)
```

```python
# ============================================================
# CHECKPOINT 3
# VALIDATE OLFACTORY HEMIBRAIN MAPPING EXCEPTIONS
# ============================================================

print("=" * 70)
print("OLFACTORY HEMIBRAIN MAPPING EXCEPTIONS")
print("=" * 70)

exceptions = [
    "VC3_adPN",
    "VM6_adPN",
    "VM6_lvPN"
]

exception_rows = crossmatch_df[
    crossmatch_df["cell_type"].isin(exceptions)
].copy()

print("\nExact matches in canonical 'cell_type':")
print("Rows:", len(exception_rows))

if len(exception_rows) > 0:
    display(
        exception_rows[
            [
                "cell_type",
                "hemibrain_type",
                "morphology_group",
                "hemilineage",
                "region",
                "in_hemibrain"
            ]
        ]
    )
else:
    print("No exact canonical matches found.")

# ------------------------------------------------------------
# Checking whether the exception names occur anywhere
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SEARCHING ALL MAPPING COLUMNS")
print("=" * 70)

for target in exceptions:

    print(f"\n{target}")

    matches = crossmatch_df[
        crossmatch_df.astype(str)
        .apply(
            lambda row: row.str.contains(
                target,
                case=False,
                regex=False
            ).any(),
            axis=1
        )
    ]

    print("Matching rows:", len(matches))

    if len(matches) > 0:
        display(
            matches[
                [
                    "cell_type",
                    "hemibrain_type",
                    "morphology_group",
                    "hemilineage",
                    "region",
                    "in_hemibrain"
                ]
            ].head(20)
        )
```

```python
# ============================================================
# CHECKPOINT 3
# BUILD TYPE-LEVEL MOLECULAR TARGET TABLE
# ============================================================

# ------------------------------------------------------------
# Re-establish neurotransmitter dataframe from notebook state
# ------------------------------------------------------------

nt_df = nt_data["gt_data.csv"].copy()

NEUROTRANSMITTERS = [
    "acetylcholine",
    "glutamate",
    "gaba",
    "glycine",
    "dopamine",
    "serotonin",
    "octopamine",
    "tyramine",
    "histamine",
    "nitric_oxide"
]

# ------------------------------------------------------------
# Keep only canonical olfactory cell types with molecular data
# ------------------------------------------------------------

olf_nt = nt_df[
    nt_df["cell_type"].isin(olfactory_type_set)
].copy()

# ------------------------------------------------------------
# Build one row per cell type
# ------------------------------------------------------------

target_rows = []

for cell_type, group in olf_nt.groupby("cell_type"):

    positive_nts = []

    for nt in NEUROTRANSMITTERS:

        if (group[nt] == 1).any():
            positive_nts.append(nt)

    target_rows.append({
        "cell_type": cell_type,
        "n_molecular_records": len(group),
        "positive_neurotransmitters": positive_nts
    })

target_df = pd.DataFrame(target_rows)

# ------------------------------------------------------------
# Assign single neurotransmitter target
# ------------------------------------------------------------

target_df["neurotransmitter"] = target_df[
    "positive_neurotransmitters"
].apply(
    lambda x: x[0] if len(x) == 1 else np.nan
)

# ------------------------------------------------------------
# Add HemiBrain mapping information
# ------------------------------------------------------------

mapping_for_targets = crossmatch_df[
    crossmatch_df["cell_type"].isin(
        target_df["cell_type"]
    )
].copy()

target_table = mapping_for_targets.merge(
    target_df[
        [
            "cell_type",
            "n_molecular_records",
            "neurotransmitter"
        ]
    ],
    on="cell_type",
    how="inner"
)

print("=" * 70)
print("TYPE-LEVEL MOLECULAR TARGET TABLE")
print("=" * 70)

print("Molecular cell types:", target_df["cell_type"].nunique())
print("Mapping rows:", len(mapping_for_targets))
print("Final target rows:", len(target_table))
print(
    "Unique target cell types:",
    target_table["cell_type"].nunique()
)

print(
    "Missing neurotransmitter labels:",
    target_table["neurotransmitter"].isna().sum()
)

display(
    target_table[
        [
            "cell_type",
            "hemibrain_type",
            "in_hemibrain",
            "n_molecular_records",
            "neurotransmitter"
        ]
    ]
)
```

```python
# ============================================================
# CHECKPOINT 3
# CHECK MAPPING UNIQUENESS
# ============================================================

print("=" * 70)
print("MAPPING UNIQUENESS AUDIT")
print("=" * 70)

mapping_counts = (
    target_table
    .groupby("cell_type")
    .size()
    .sort_values(ascending=False)
)

print("\nRows per target cell type:")
display(
    mapping_counts
    .rename("n_mapping_rows")
    .reset_index()
)

duplicate_mapping_types = mapping_counts[
    mapping_counts > 1
]

print("\nCell types with >1 mapping row:",
      len(duplicate_mapping_types))

if len(duplicate_mapping_types) > 0:

    print("\nPotential duplicate mappings:")

    display(
        target_table[
            target_table["cell_type"].isin(
                duplicate_mapping_types.index
            )
        ][
            [
                "cell_type",
                "hemibrain_type",
                "morphology_group",
                "hemilineage",
                "region",
                "in_hemibrain"
            ]
        ]
    )

else:

    print("✓ Every target cell type has one mapping row.")
```

```python
# ============================================================
# CHECKPOINT 3
# SEPARATE TYPE-LEVEL TARGETS FROM HEMIBRAIN MAPPINGS
# ============================================================

print("=" * 70)
print("TYPE-LEVEL TARGET vs HEMIBRAIN MAPPING")
print("=" * 70)

# ------------------------------------------------------------
# Type-level molecular target
# Exactly ONE row per canonical molecular cell type
# ------------------------------------------------------------

type_targets = (
    target_table[
        [
            "cell_type",
            "n_molecular_records",
            "neurotransmitter"
        ]
    ]
    .drop_duplicates(subset="cell_type")
    .copy()
)

# ------------------------------------------------------------
# HemiBrain mapping
# Potentially MULTIPLE rows per canonical cell type
# ------------------------------------------------------------

hemibrain_mapping = (
    crossmatch_df[
        crossmatch_df["cell_type"].isin(
            type_targets["cell_type"]
        )
    ][
        [
            "cell_type",
            "hemibrain_type",
            "morphology_group",
            "hemilineage",
            "region",
            "in_hemibrain"
        ]
    ]
    .copy()
)

print("Canonical molecular target types:")
print(type_targets["cell_type"].nunique())

print("\nHemiBrain mapping rows:")
print(len(hemibrain_mapping))

print("\nUnique canonical types represented:")
print(hemibrain_mapping["cell_type"].nunique())

print("\nUnique HemiBrain type labels:")
print(
    hemibrain_mapping["hemibrain_type"]
    .dropna()
    .nunique()
)

print("\nMapping rows per canonical type:")
display(
    hemibrain_mapping
    .groupby("cell_type")
    .size()
    .sort_values(ascending=False)
    .rename("n_hemibrain_mappings")
    .reset_index()
)
```

```python
# ============================================================
# CHECKPOINT 3
# CREATE MOLECULAR-LABELLED HEMIBRAIN MAPPING TABLE
# ============================================================

final_mapping = hemibrain_mapping.merge(
    type_targets,
    on="cell_type",
    how="inner"
)

# Keep only mappings explicitly marked as HemiBrain
final_mapping = final_mapping[
    final_mapping["in_hemibrain"] == True
].copy()

print("=" * 70)
print("MOLECULAR-LABELLED HEMIBRAIN MAPPING")
print("=" * 70)

print("Rows:", len(final_mapping))
print(
    "Canonical cell types:",
    final_mapping["cell_type"].nunique()
)
print(
    "HemiBrain type labels:",
    final_mapping["hemibrain_type"].dropna().nunique()
)

print(
    "Missing HemiBrain type labels:",
    final_mapping["hemibrain_type"].isna().sum()
)

display(
    final_mapping[
        [
            "cell_type",
            "hemibrain_type",
            "morphology_group",
            "hemilineage",
            "region",
            "n_molecular_records",
            "neurotransmitter"
        ]
    ]
)
```

```python
# ============================================================
# CHECKPOINT 3
# DOCUMENT NON-IDENTICAL / ONE-TO-MANY MAPPINGS
# ============================================================

print("=" * 70)
print("NON-IDENTICAL AND ONE-TO-MANY HEMIBRAIN MAPPINGS")
print("=" * 70)

# ------------------------------------------------------------
# Canonical type != HemiBrain type
# ------------------------------------------------------------

non_identical = final_mapping[
    final_mapping["cell_type"] !=
    final_mapping["hemibrain_type"]
].copy()

print("\nCanonical type != HemiBrain type:")
print(
    non_identical["cell_type"].nunique(),
    "canonical cell types"
)

display(
    non_identical[
        [
            "cell_type",
            "hemibrain_type",
            "neurotransmitter"
        ]
    ]
)

# ------------------------------------------------------------
# One canonical type -> multiple HemiBrain mappings
# ------------------------------------------------------------

mapping_counts = (
    final_mapping
    .groupby("cell_type")["hemibrain_type"]
    .nunique()
)

one_to_many = mapping_counts[
    mapping_counts > 1
]

print("\n" + "=" * 70)
print("ONE-TO-MANY MAPPINGS")
print("=" * 70)

print(
    "Canonical cell types with >1 HemiBrain type:",
    len(one_to_many)
)

if len(one_to_many) > 0:

    display(
        final_mapping[
            final_mapping["cell_type"].isin(
                one_to_many.index
            )
        ][
            [
                "cell_type",
                "hemibrain_type",
                "neurotransmitter"
            ]
        ]
        .sort_values(
            ["cell_type", "hemibrain_type"]
        )
    )
```

```python
# ============================================================
# CHECKPOINT 3
# NEUROTRANSMITTER CLASS BALANCE
# ============================================================

print("=" * 70)
print("NEUROTRANSMITTER CLASS BALANCE")
print("=" * 70)

class_counts = (
    type_targets["neurotransmitter"]
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_cell_types")
)

display(class_counts)

print("\nTotal canonical cell types:",
      len(type_targets))

print("Number of neurotransmitter classes:",
      type_targets["neurotransmitter"].nunique())

print("Largest class:",
      class_counts["n_cell_types"].max())

print("Smallest class:",
      class_counts["n_cell_types"].min())

print("\nClass proportions:")

class_counts["proportion"] = (
    class_counts["n_cell_types"] /
    len(type_targets)
)

display(class_counts)
```

```python
# ============================================================
# CHECKPOINT 3
# MOLECULAR EVIDENCE CONFIDENCE
# ============================================================

print("=" * 70)
print("MOLECULAR EVIDENCE CONFIDENCE")
print("=" * 70)

confidence_summary = (
    olf_nt
    .groupby("cell_type")
    .agg(
        n_records=(
            "neurotransmitter_verified_source",
            "size"
        ),
        max_confidence=(
            "neurotransmitter_verified_confidence",
            "max"
        ),
        min_confidence=(
            "neurotransmitter_verified_confidence",
            "min"
        ),
        n_sources=(
            "neurotransmitter_verified_source",
            "nunique"
        )
    )
    .reset_index()
)

confidence_summary = confidence_summary.merge(
    type_targets[
        [
            "cell_type",
            "neurotransmitter"
        ]
    ],
    on="cell_type",
    how="inner"
)

display(
    confidence_summary.sort_values(
        ["max_confidence", "cell_type"],
        ascending=[False, True]
    )
)

print("\nConfidence distribution:")
display(
    confidence_summary["max_confidence"]
    .value_counts()
    .sort_index()
    .rename_axis("max_confidence")
    .reset_index(name="n_cell_types")
)
```

```python
# ============================================================
# CHECKPOINT 3
# SAVE CHECKPOINT 3 OUTPUTS
# ============================================================

import os

os.makedirs("data/processed", exist_ok=True)

type_targets_path = (
    "data/processed/"
    "olfactory_type_level_neurotransmitter_targets.csv"
)

mapping_path = (
    "data/processed/"
    "olfactory_hemibrain_molecular_mapping.csv"
)

confidence_path = (
    "data/processed/"
    "olfactory_molecular_evidence_confidence.csv"
)

type_targets.to_csv(
    type_targets_path,
    index=False
)

final_mapping.to_csv(
    mapping_path,
    index=False
)

confidence_summary.to_csv(
    confidence_path,
    index=False
)

print("=" * 70)
print("CHECKPOINT 3 OUTPUTS SAVED")
print("=" * 70)

print(type_targets_path)
print(mapping_path)
print(confidence_path)
```

```python
# ============================================================
# CHECKPOINT 3
# FINAL CHECKPOINT 3 AUDIT
# ============================================================

print("=" * 70)
print("CHECKPOINT 3 — FINAL AUDIT")
print("=" * 70)

print(
    f"Canonical olfactory types: "
    f"{len(olfactory_type_set)}"
)

print(
    f"Molecularly represented types: "
    f"{type_targets['cell_type'].nunique()}"
)

print(
    f"Type-level molecular targets: "
    f"{len(type_targets)}"
)

print(
    f"HemiBrain mapping rows: "
    f"{len(final_mapping)}"
)

print(
    f"Unique HemiBrain type labels: "
    f"{final_mapping['hemibrain_type'].dropna().nunique()}"
)

print(
    f"Canonical types represented in HemiBrain mapping: "
    f"{final_mapping['cell_type'].nunique()}"
)

print(
    f"Missing molecular labels: "
    f"{type_targets['neurotransmitter'].isna().sum()}"
)

print(
    f"Missing HemiBrain labels: "
    f"{final_mapping['hemibrain_type'].isna().sum()}"
)

print(
    f"Neurotransmitter classes: "
    f"{type_targets['neurotransmitter'].nunique()}"
)

print("\n" + "=" * 70)
print("NEUROTRANSMITTER DISTRIBUTION")
print("=" * 70)

display(
    type_targets["neurotransmitter"]
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_cell_types")
)

print("\n" + "=" * 70)

if (
    type_targets["neurotransmitter"].isna().sum() == 0
    and final_mapping["hemibrain_type"].isna().sum() == 0
):
    print("✓ CHECKPOINT 3 DATA INTEGRITY PASSED")
else:
    print("⚠ CHECKPOINT 3 REQUIRES REVIEW")

print("=" * 70)
```

```python
# ============================================================
# CHECKPOINT 4
# PREPARE HEMIBRAIN TYPE LIST
# ============================================================

print("=" * 70)
print("CHECKPOINT 4 — HEMIBRAIN CONNECTIVITY TARGETS")
print("=" * 70)

# HemiBrain types associated with our 65 molecularly labelled
# canonical olfactory cell types.

connectivity_targets = (
    final_mapping[
        [
            "cell_type",
            "hemibrain_type",
            "neurotransmitter"
        ]
    ]
    .dropna(subset=["hemibrain_type"])
    .drop_duplicates()
    .copy()
)

print(
    "Canonical molecular cell types:",
    connectivity_targets["cell_type"].nunique()
)

print(
    "Unique HemiBrain types:",
    connectivity_targets["hemibrain_type"].nunique()
)

print("\nNeurotransmitter distribution at canonical type level:")

display(
    connectivity_targets[
        ["cell_type", "neurotransmitter"]
    ]
    .drop_duplicates()
    ["neurotransmitter"]
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_cell_types")
)

print("\nFirst HemiBrain targets:")

display(
    connectivity_targets.head(20)
)
```

```python
# ============================================================
# CHECKPOINT 4
# DIRECT NEUPRINT API CONFIGURATION
# ============================================================

print("=" * 70)
print("CHECKPOINT 4 — DIRECT NEUPRINT API")
print("=" * 70)

NEUPRINT_SERVER = "https://neuprint.janelia.org"
NEUPRINT_DATASET = "hemibrain:v1.2.1"
NEUPRINT_QUERY_URL = (
    NEUPRINT_SERVER + "/api/custom/custom"
)

NEUPRINT_HEADERS = {
    "Authorization": (
        "Bearer "
        + os.environ["NEUPRINT_APPLICATION_CREDENTIALS"]
    ),
    "Content-Type": "application/json"
}

print("Server:", NEUPRINT_SERVER)
print("Dataset:", NEUPRINT_DATASET)
print("Query endpoint:", NEUPRINT_QUERY_URL)
print("Token available:", "NEUPRINT_APPLICATION_CREDENTIALS" in os.environ)
```

```python
# ============================================================
# CHECKPOINT 4
# REUSABLE NEUPRINT CYPHER FUNCTION
# ============================================================

def run_neuprint_cypher(cypher, timeout=(10, 60)):
    """
    Execute a Cypher query against the HemiBrain neuPrint API
    and return the result as a pandas DataFrame.
    """

    payload = {
        "cypher": cypher,
        "dataset": NEUPRINT_DATASET
    }

    response = requests.post(
        NEUPRINT_QUERY_URL,
        headers=NEUPRINT_HEADERS,
        json=payload,
        timeout=timeout
    )

    response.raise_for_status()

    result = response.json()

    return pd.DataFrame(
        result["data"],
        columns=result["columns"]
    )


print("✓ Direct neuPrint Cypher function ready.")
```

```python
# ============================================================
# CHECKPOINT 4
# RETRIEVE TARGET OLFACTORY NEURONS
# ============================================================

print("=" * 70)
print("RETRIEVING TARGET OLFACTORY NEURONS")
print("=" * 70)

target_types = (
    connectivity_targets["hemibrain_type"]
    .dropna()
    .astype(str)
    .drop_duplicates()
    .tolist()
)

print("Target HemiBrain types:", len(target_types))

assert len(target_types) == 68, (
    f"Expected 68 HemiBrain types, "
    f"found {len(target_types)}"
)

# Escape strings for Cypher
escaped_types = [
    t.replace("\\", "\\\\").replace("'", "\\'")
    for t in target_types
]

type_list = ", ".join(
    f"'{t}'"
    for t in escaped_types
)

cypher = f"""
MATCH (n:`hemibrain_Neuron`)
WHERE n.type IN [{type_list}]
RETURN
    n.bodyId AS bodyId,
    n.type AS hemibrain_type,
    n.instance AS instance
ORDER BY n.type, n.bodyId
"""

print("Querying neuPrint...")

target_neurons = run_neuprint_cypher(cypher)

print("\n✓ Query successful.")
print("Total neurons retrieved:", len(target_neurons))
print(
    "Unique body IDs:",
    target_neurons["bodyId"].nunique()
)
print(
    "Unique HemiBrain types:",
    target_neurons["hemibrain_type"].nunique()
)

display(target_neurons.head(20))
```

```python
# ============================================================
# CHECKPOINT 4
# TARGET NEURON RETRIEVAL AUDIT
# ============================================================

print("=" * 70)
print("TARGET NEURON RETRIEVAL AUDIT")
print("=" * 70)

retrieved_types = set(
    target_neurons["hemibrain_type"]
    .dropna()
    .astype(str)
)

missing_types = sorted(
    set(target_types) - retrieved_types
)

unexpected_types = sorted(
    retrieved_types - set(target_types)
)

print("Requested HemiBrain types:", len(target_types))
print("Retrieved HemiBrain types:", len(retrieved_types))
print("Missing HemiBrain types:", len(missing_types))
print("Unexpected HemiBrain types:", len(unexpected_types))

if missing_types:
    print("\n⚠ Missing target types:")
    for cell_type in missing_types:
        print("  -", cell_type)
else:
    print("\n✓ All 68 target HemiBrain types retrieved.")

if unexpected_types:
    print("\n⚠ Unexpected types:")
    for cell_type in unexpected_types:
        print("  -", cell_type)
else:
    print("✓ No unexpected HemiBrain types retrieved.")
```

```python
# ============================================================
# CHECKPOINT 4
# NEURON COUNT PER HEMIBRAIN TYPE
# ============================================================

print("=" * 70)
print("NEURON COUNT PER HEMIBRAIN TYPE")
print("=" * 70)

neuron_counts = (
    target_neurons
    .groupby("hemibrain_type")
    .agg(
        neuron_count=("bodyId", "nunique"),
        instance_count=("instance", "nunique")
    )
    .reset_index()
    .sort_values(
        ["neuron_count", "hemibrain_type"],
        ascending=[False, True]
    )
)

display(neuron_counts)

print("\nSummary:")
print(
    "Total individual neurons:",
    target_neurons["bodyId"].nunique()
)
print(
    "Median neurons per type:",
    neuron_counts["neuron_count"].median()
)
print(
    "Maximum neurons in one type:",
    neuron_counts["neuron_count"].max()
)
print(
    "Minimum neurons in one type:",
    neuron_counts["neuron_count"].min()
)
```

```python
# ============================================================
# CHECKPOINT 4
# MAP NEURONS TO MOLECULAR LABELS
# ============================================================

print("=" * 70)
print("MAPPING NEURONS TO MOLECULAR LABELS")
print("=" * 70)

neuron_molecular_targets = target_neurons.merge(
    connectivity_targets[
        [
            "cell_type",
            "hemibrain_type",
            "neurotransmitter"
        ]
    ].drop_duplicates(),
    on="hemibrain_type",
    how="left"
)

print("Total neurons:", len(neuron_molecular_targets))
print(
    "Neurons with molecular labels:",
    neuron_molecular_targets["neurotransmitter"].notna().sum()
)
print(
    "Neurons without molecular labels:",
    neuron_molecular_targets["neurotransmitter"].isna().sum()
)

print("\nNeurotransmitter distribution:")
display(
    neuron_molecular_targets["neurotransmitter"]
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

display(
    neuron_molecular_targets.head(20)
)
```

```python
# ============================================================
# CHECKPOINT 4
# MOLECULAR MAPPING INTEGRITY AUDIT
# ============================================================

print("=" * 70)
print("MOLECULAR MAPPING INTEGRITY AUDIT")
print("=" * 70)

missing_labels = (
    neuron_molecular_targets[
        "neurotransmitter"
    ].isna().sum()
)

missing_canonical = (
    neuron_molecular_targets[
        "cell_type"
    ].isna().sum()
)

duplicate_body_ids = (
    neuron_molecular_targets["bodyId"]
    .duplicated()
    .sum()
)

print("Missing canonical cell type:", missing_canonical)
print("Missing neurotransmitter:", missing_labels)
print("Duplicate body IDs:", duplicate_body_ids)

if (
    missing_canonical == 0
    and missing_labels == 0
    and duplicate_body_ids == 0
):
    print("\n✓ MOLECULAR MAPPING INTEGRITY PASSED")
else:
    print("\n⚠ MOLECULAR MAPPING REQUIRES REVIEW")
```

```python
# ============================================================
# CHECKPOINT 4
# SAVE TARGET NEURON METADATA
# ============================================================

target_neurons_path = (
    "data/processed/"
    "olfactory_hemibrain_target_neurons.csv"
)

neuron_molecular_targets_path = (
    "data/processed/"
    "olfactory_neuron_molecular_targets.csv"
)

target_neurons.to_csv(
    target_neurons_path,
    index=False
)

neuron_molecular_targets.to_csv(
    neuron_molecular_targets_path,
    index=False
)

print("=" * 70)
print("TARGET NEURON DATA SAVED")
print("=" * 70)

print(target_neurons_path)
print(neuron_molecular_targets_path)
```

```python
# ============================================================
# CHECKPOINT 4
# RETRIEVE CONNECTIVITY FOR TARGET NEURONS
# ============================================================

print("=" * 70)
print("RETRIEVING TARGET NEURON CONNECTIVITY")
print("=" * 70)

target_body_ids = (
    target_neurons["bodyId"]
    .astype(int)
    .drop_duplicates()
    .tolist()
)

print("Target neurons:", len(target_body_ids))

assert len(target_body_ids) == 165, (
    f"Expected 165 target neurons, found {len(target_body_ids)}"
)

# Convert body IDs into a Cypher list
body_id_list = ", ".join(
    str(body_id)
    for body_id in target_body_ids
)

cypher = f"""
MATCH (a:`hemibrain_Neuron`)-[e:ConnectsTo]->(b:`hemibrain_Neuron`)
WHERE a.bodyId IN [{body_id_list}]
   OR b.bodyId IN [{body_id_list}]
RETURN
    a.bodyId AS pre_bodyId,
    a.type AS pre_type,
    b.bodyId AS post_bodyId,
    b.type AS post_type,
    e.weight AS synapse_count
"""

print("Querying neuPrint...")
print("This may take a little time depending on the response size.")

connectivity_edges = run_neuprint_cypher(
    cypher,
    timeout=(10, 120)
)

print("\n✓ Connectivity query completed.")
print("Total connectivity edges:", len(connectivity_edges))

display(connectivity_edges.head(20))
```

```python
# ============================================================
# CHECKPOINT 4
# CONNECTIVITY DATA QUALITY AUDIT
# ============================================================

print("=" * 70)
print("CONNECTIVITY DATA QUALITY AUDIT")
print("=" * 70)

print("Total connectivity edges:", len(connectivity_edges))

print("\nColumn names:")
print(list(connectivity_edges.columns))

print("\nMissing values:")
display(
    connectivity_edges.isna().sum()
)

print("\nUnique presynaptic neurons:")
print(
    connectivity_edges["pre_bodyId"].nunique()
)

print("Unique postsynaptic neurons:")
print(
    connectivity_edges["post_bodyId"].nunique()
)

print("Unique synaptic weights:")
print(
    connectivity_edges["synapse_count"].nunique()
)

print("\nSynapse-count summary:")
display(
    connectivity_edges["synapse_count"].describe()
)

print("\nMinimum synapse count:",
      connectivity_edges["synapse_count"].min())

print("Maximum synapse count:",
      connectivity_edges["synapse_count"].max())

# Basic integrity checks
assert connectivity_edges["pre_bodyId"].notna().all()
assert connectivity_edges["post_bodyId"].notna().all()
assert connectivity_edges["synapse_count"].notna().all()

assert (
    connectivity_edges["synapse_count"] >= 1
).all()

print("\n✓ Body IDs are complete.")
print("✓ Synapse counts are complete.")
print("✓ All synapse counts are positive.")
print("✓ CONNECTIVITY DATA QUALITY AUDIT PASSED")
```

```python
# ============================================================
# CHECKPOINT 4
# CLASSIFY TARGET CONNECTIVITY DIRECTION
# ============================================================

target_body_id_set = set(
    target_body_ids
)

connectivity_edges["target_is_pre"] = (
    connectivity_edges["pre_bodyId"]
    .isin(target_body_id_set)
)

connectivity_edges["target_is_post"] = (
    connectivity_edges["post_bodyId"]
    .isin(target_body_id_set)
)

connectivity_edges["direction"] = np.select(
    [
        (
            connectivity_edges["target_is_pre"]
            & ~connectivity_edges["target_is_post"]
        ),
        (
            connectivity_edges["target_is_post"]
            & ~connectivity_edges["target_is_pre"]
        ),
        (
            connectivity_edges["target_is_pre"]
            & connectivity_edges["target_is_post"]
        )
    ],
    [
        "outgoing",
        "incoming",
        "both"
    ],
    default="unexpected"
)

print("=" * 70)
print("TARGET CONNECTIVITY DIRECTION")
print("=" * 70)

display(
    connectivity_edges["direction"]
    .value_counts()
    .rename_axis("direction")
    .reset_index(name="edge_count")
)

unexpected_edges = (
    connectivity_edges["direction"]
    == "unexpected"
).sum()

print("\nUnexpected edges:", unexpected_edges)

assert unexpected_edges == 0

print("✓ Direction classification passed.")
```

```python
# ============================================================
# CHECKPOINT 4
# TARGET NEURON CONNECTIVITY COVERAGE
# ============================================================

outgoing_counts = (
    connectivity_edges[
        connectivity_edges["target_is_pre"]
    ]
    .groupby("pre_bodyId")
    .agg(
        outgoing_edges=("post_bodyId", "size"),
        outgoing_partners=("post_bodyId", "nunique"),
        outgoing_synapses=("synapse_count", "sum")
    )
    .rename_axis("bodyId")
)

incoming_counts = (
    connectivity_edges[
        connectivity_edges["target_is_post"]
    ]
    .groupby("post_bodyId")
    .agg(
        incoming_edges=("pre_bodyId", "size"),
        incoming_partners=("pre_bodyId", "nunique"),
        incoming_synapses=("synapse_count", "sum")
    )
    .rename_axis("bodyId")
)

connectivity_coverage = (
    target_neurons[
        ["bodyId", "hemibrain_type", "instance"]
    ]
    .drop_duplicates("bodyId")
    .set_index("bodyId")
    .join(outgoing_counts)
    .join(incoming_counts)
    .fillna(0)
    .reset_index()
)

print("=" * 70)
print("TARGET NEURON CONNECTIVITY COVERAGE")
print("=" * 70)

coverage_columns = [
    "outgoing_edges",
    "outgoing_partners",
    "outgoing_synapses",
    "incoming_edges",
    "incoming_partners",
    "incoming_synapses"
]

display(
    connectivity_coverage[coverage_columns]
    .describe()
)

zero_outgoing = (
    connectivity_coverage["outgoing_edges"] == 0
).sum()

zero_incoming = (
    connectivity_coverage["incoming_edges"] == 0
).sum()

print("\nTarget neurons:", len(connectivity_coverage))
print("Zero outgoing edges:", zero_outgoing)
print("Zero incoming edges:", zero_incoming)

print("\n✓ Connectivity coverage table constructed.")
```

```python
# ============================================================
# CHECKPOINT 4
# CONSTRUCT LABELLED CONNECTIVITY TABLE
# ============================================================

labelled_connectivity = connectivity_coverage.merge(
    neuron_molecular_targets[
        [
            "bodyId",
            "cell_type",
            "neurotransmitter"
        ]
    ],
    on="bodyId",
    how="left"
)

print("=" * 70)
print("LABELLED CONNECTIVITY TABLE")
print("=" * 70)

print("Rows:", len(labelled_connectivity))
print("Columns:", len(labelled_connectivity.columns))

print("\nNeurotransmitter distribution:")
display(
    labelled_connectivity["neurotransmitter"]
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

assert len(labelled_connectivity) == 165
assert labelled_connectivity["neurotransmitter"].notna().all()
assert labelled_connectivity["bodyId"].nunique() == 165

print("\n✓ 165 target neurons retained.")
print("✓ All neurons have molecular labels.")
print("✓ LABELLED CONNECTIVITY TABLE PASSED.")
```

```python
# ============================================================
# CHECKPOINT 4
# SAVE RAW CONNECTIVITY DATA
# ============================================================

connectivity_edges_path = (
    "data/raw/"
    "hemibrain_target_neuron_connectivity.csv"
)

connectivity_edges.to_csv(
    connectivity_edges_path,
    index=False
)

connectivity_coverage_path = (
    "data/processed/"
    "olfactory_neuron_connectivity_coverage.csv"
)

labelled_connectivity_path = (
    "data/processed/"
    "olfactory_labelled_connectivity_features_base.csv"
)

connectivity_coverage.to_csv(
    connectivity_coverage_path,
    index=False
)

labelled_connectivity.to_csv(
    labelled_connectivity_path,
    index=False
)

print("=" * 70)
print("CONNECTIVITY DATA SAVED")
print("=" * 70)

print("Raw edges:")
print(" ", connectivity_edges_path)

print("\nConnectivity coverage:")
print(" ", connectivity_coverage_path)

print("\nLabelled base table:")
print(" ", labelled_connectivity_path)

print("\n✓ Raw and processed connectivity data saved.")
```

```python
# ================================================================
# CHECKPOINT 4
# CONNECTIVITY ARCHITECTURE FEATURE ENGINEERING
# ================================================================

print("=" * 70)
print("CONNECTIVITY ARCHITECTURE FEATURE ENGINEERING")
print("=" * 70)

# ------------------------------------------------
# 1. LOCATE THE EXISTING CONNECTIVITY DATAFRAME
# ------------------------------------------------

# The connectivity query has already been run.
# We do NOT query neuPrint again.

possible_edge_variables = [
    "connectivity_edges",
    "target_connectivity",
    "connectivity_data",
    "edges",
    "connectivity_result",
    "connectivity_results",
    "raw_connectivity"
]

edge_df = None
edge_variable_name = None

for var_name in possible_edge_variables:
    if var_name in globals():
        candidate = globals()[var_name]

        if isinstance(candidate, pd.DataFrame):
            required_columns = {
                "pre_bodyId",
                "post_bodyId",
                "synapse_count"
            }

            if required_columns.issubset(candidate.columns):
                edge_df = candidate.copy()
                edge_variable_name = var_name
                break

if edge_df is None:
    print("Could not automatically identify the existing connectivity dataframe.")
    print("\nDataFrame variables currently in memory:")

    for name, obj in globals().items():
        if isinstance(obj, pd.DataFrame):
            print(
                f" - {name}: shape={obj.shape}, "
                f"columns={list(obj.columns)}"
            )

    raise NameError(
        "\nExisting connectivity dataframe could not be identified. "
        "No neuPrint query was repeated."
    )

print(f"\n✓ Using existing dataframe: {edge_variable_name}")
print(f"✓ Connectivity edges available: {len(edge_df):,}")

# ------------------------------------------------
# 2. BASIC VALIDATION
# ------------------------------------------------

required_columns = [
    "pre_bodyId",
    "post_bodyId",
    "synapse_count"
]

assert all(col in edge_df.columns for col in required_columns)

assert edge_df["pre_bodyId"].notna().all()
assert edge_df["post_bodyId"].notna().all()
assert edge_df["synapse_count"].notna().all()
assert (edge_df["synapse_count"] > 0).all()

print("✓ Required connectivity columns present.")
print("✓ Body IDs complete.")
print("✓ Synapse counts complete and positive.")

# ------------------------------------------------
# 3. IDENTIFY TARGET NEURONS
# ------------------------------------------------

target_ids = set(
    target_body_ids
)

print(f"\nTarget neurons: {len(target_ids)}")

# ------------------------------------------------
# 4. MARK INCOMING / OUTGOING CONNECTIONS
# ------------------------------------------------

edge_df["pre_is_target"] = (
    edge_df["pre_bodyId"].isin(target_ids)
)

edge_df["post_is_target"] = (
    edge_df["post_bodyId"].isin(target_ids)
)

# Only edges involving at least one target neuron
target_edges = edge_df[
    edge_df["pre_is_target"] |
    edge_df["post_is_target"]
].copy()

assert len(target_edges) == len(edge_df)

print(
    f"✓ All {len(target_edges):,} edges involve target neurons."
)

# ------------------------------------------------
# 5. OUTGOING CONNECTIVITY FEATURES
# ------------------------------------------------

outgoing_edges = target_edges[
    target_edges["pre_is_target"]
].copy()

outgoing_features = (
    outgoing_edges
    .groupby("pre_bodyId")
    .agg(
        outgoing_edges=("post_bodyId", "size"),
        outgoing_partners=("post_bodyId", "nunique"),
        outgoing_synapses=("synapse_count", "sum"),
        outgoing_mean_synapses=("synapse_count", "mean"),
        outgoing_median_synapses=("synapse_count", "median"),
        outgoing_max_synapses=("synapse_count", "max"),
        outgoing_synapse_sd=("synapse_count", "std")
    )
    .reset_index()
    .rename(columns={"pre_bodyId": "bodyId"})
)

# ------------------------------------------------
# 6. INCOMING CONNECTIVITY FEATURES
# ------------------------------------------------

incoming_edges = target_edges[
    target_edges["post_is_target"]
].copy()

incoming_features = (
    incoming_edges
    .groupby("post_bodyId")
    .agg(
        incoming_edges=("pre_bodyId", "size"),
        incoming_partners=("pre_bodyId", "nunique"),
        incoming_synapses=("synapse_count", "sum"),
        incoming_mean_synapses=("synapse_count", "mean"),
        incoming_median_synapses=("synapse_count", "median"),
        incoming_max_synapses=("synapse_count", "max"),
        incoming_synapse_sd=("synapse_count", "std")
    )
    .reset_index()
    .rename(columns={"post_bodyId": "bodyId"})
)

# ------------------------------------------------
# 7. CREATE ONE FEATURE MATRIX FOR ALL 165 NEURONS
# ------------------------------------------------

connectivity_features = target_neurons[
    ["bodyId"]
].copy()

connectivity_features = connectivity_features.merge(
    outgoing_features,
    on="bodyId",
    how="left"
)

connectivity_features = connectivity_features.merge(
    incoming_features,
    on="bodyId",
    how="left"
)

# Missing values mean that a particular feature was not
# observed for that direction. Zero-fill them.
feature_columns = [
    col for col in connectivity_features.columns
    if col != "bodyId"
]

connectivity_features[feature_columns] = (
    connectivity_features[feature_columns]
    .fillna(0)
)

# ------------------------------------------------
# 8. DERIVED ARCHITECTURAL FEATURES
# ------------------------------------------------

connectivity_features["total_edges"] = (
    connectivity_features["outgoing_edges"] +
    connectivity_features["incoming_edges"]
)

connectivity_features["total_partners"] = (
    connectivity_features["outgoing_partners"] +
    connectivity_features["incoming_partners"]
)

connectivity_features["total_synapses"] = (
    connectivity_features["outgoing_synapses"] +
    connectivity_features["incoming_synapses"]
)

connectivity_features["outgoing_incoming_edge_ratio"] = (
    connectivity_features["outgoing_edges"] /
    connectivity_features["incoming_edges"].replace(0, np.nan)
)

connectivity_features["outgoing_incoming_synapse_ratio"] = (
    connectivity_features["outgoing_synapses"] /
    connectivity_features["incoming_synapses"].replace(0, np.nan)
)

connectivity_features[
    [
        "outgoing_incoming_edge_ratio",
        "outgoing_incoming_synapse_ratio"
    ]
] = connectivity_features[
    [
        "outgoing_incoming_edge_ratio",
        "outgoing_incoming_synapse_ratio"
    ]
].replace(
    [np.inf, -np.inf],
    np.nan
).fillna(0)

# ------------------------------------------------
# 9. FINAL AUDIT
# ------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE MATRIX AUDIT")
print("=" * 70)

print(
    f"\nFeature matrix shape: "
    f"{connectivity_features.shape}"
)

print("\nFeature columns:")
for col in connectivity_features.columns:
    print(" -", col)

print("\nMissing values:")
print(connectivity_features.isna().sum())

assert len(connectivity_features) == 165
assert connectivity_features["bodyId"].nunique() == 165
assert connectivity_features.isna().sum().sum() == 0

print("\n✓ 165 target neurons represented.")
print("✓ No missing feature values.")
print("✓ Connectivity architecture features constructed.")
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 4
# FEATURE QUALITY, SCALE AND REDUNDANCY AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 4 — FEATURE QUALITY & REDUNDANCY AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. DEFINE ML FEATURE COLUMNS
# ------------------------------------------------

ml_feature_columns = [
    col for col in connectivity_features.columns
    if col != "bodyId"
]

print(f"\nNumber of candidate ML features: {len(ml_feature_columns)}")

# ------------------------------------------------
# 2. CHECK FEATURE TYPES
# ------------------------------------------------

print("\nFeature data types:")

feature_dtypes = connectivity_features[
    ml_feature_columns
].dtypes

display(feature_dtypes.to_frame("dtype"))

assert all(
    np.issubdtype(dtype, np.number)
    for dtype in feature_dtypes
)

print("✓ All candidate features are numeric.")

# ------------------------------------------------
# 3. CHECK CONSTANT FEATURES
# ------------------------------------------------

feature_variance = connectivity_features[
    ml_feature_columns
].var()

constant_features = feature_variance[
    feature_variance == 0
].index.tolist()

print("\nConstant features:")
print(constant_features)

if len(constant_features) == 0:
    print("✓ No constant features.")
else:
    print(
        f"⚠ {len(constant_features)} constant feature(s) detected."
    )

# ------------------------------------------------
# 4. CHECK FEATURE DISTRIBUTIONS
# ------------------------------------------------

feature_summary = connectivity_features[
    ml_feature_columns
].describe().T

feature_summary["skewness"] = (
    connectivity_features[
        ml_feature_columns
    ].skew()
)

print("\nFeature distribution summary:")
display(feature_summary)

# ------------------------------------------------
# 5. IDENTIFY HIGHLY SKEWED FEATURES
# ------------------------------------------------

high_skew_features = (
    feature_summary[
        feature_summary["skewness"].abs() > 2
    ]
    .index
    .tolist()
)

print("\nHighly skewed features (|skewness| > 2):")

if len(high_skew_features) == 0:
    print("None")
else:
    for feature in high_skew_features:
        print(
            f" - {feature}: "
            f"skewness = "
            f"{feature_summary.loc[feature, 'skewness']:.2f}"
        )

# ------------------------------------------------
# 6. CORRELATION ANALYSIS
# ------------------------------------------------

feature_correlation = connectivity_features[
    ml_feature_columns
].corr(method="spearman")

# Find highly correlated feature pairs
high_correlation_pairs = []

for i in range(len(ml_feature_columns)):
    for j in range(i + 1, len(ml_feature_columns)):

        feature_a = ml_feature_columns[i]
        feature_b = ml_feature_columns[j]

        correlation = feature_correlation.loc[
            feature_a,
            feature_b
        ]

        if abs(correlation) >= 0.90:
            high_correlation_pairs.append(
                (
                    feature_a,
                    feature_b,
                    correlation
                )
            )

print("\nHighly correlated feature pairs (|Spearman ρ| >= 0.90):")

if len(high_correlation_pairs) == 0:
    print("None")
else:
    for feature_a, feature_b, correlation in (
        high_correlation_pairs
    ):
        print(
            f" - {feature_a} ↔ {feature_b}: "
            f"ρ = {correlation:.3f}"
        )

# ------------------------------------------------
# 7. CHECK FOR IDENTICAL FEATURES
# ------------------------------------------------

duplicate_feature_pairs = []

for i in range(len(ml_feature_columns)):
    for j in range(i + 1, len(ml_feature_columns)):

        feature_a = ml_feature_columns[i]
        feature_b = ml_feature_columns[j]

        if connectivity_features[
            feature_a
        ].equals(
            connectivity_features[feature_b]
        ):
            duplicate_feature_pairs.append(
                (feature_a, feature_b)
            )

print("\nIdentical feature pairs:")

if len(duplicate_feature_pairs) == 0:
    print("None")
else:
    for pair in duplicate_feature_pairs:
        print(f" - {pair[0]} = {pair[1]}")

# ------------------------------------------------
# 8. VERIFY NO IDENTIFIER / LABEL LEAKAGE
# ------------------------------------------------

forbidden_feature_names = [
    "bodyId",
    "cell_type",
    "hemibrain_type",
    "neurotransmitter",
    "confidence",
    "evidence"
]

leakage_candidates = [
    feature
    for feature in ml_feature_columns
    if any(
        forbidden.lower() in feature.lower()
        for forbidden in forbidden_feature_names
    )
]

print("\nPotential identifier/label leakage features:")
print(leakage_candidates)

assert "bodyId" not in ml_feature_columns
assert "cell_type" not in ml_feature_columns
assert "hemibrain_type" not in ml_feature_columns
assert "neurotransmitter" not in ml_feature_columns

print("✓ No direct identifier or molecular-label features included.")

# ------------------------------------------------
# 9. SAVE AUDIT TABLE
# ------------------------------------------------

feature_summary.to_csv(
    "data/processed/connectivity_feature_quality_audit.csv"
)

feature_correlation.to_csv(
    "data/processed/connectivity_feature_correlation_matrix.csv"
)

print("\n✓ Feature quality audit saved.")
print("✓ Feature correlation matrix saved.")

# ------------------------------------------------
# 10. FINAL STATUS
# ------------------------------------------------

print("\n" + "=" * 70)
print("CHECKPOINT 4 — FEATURE AUDIT COMPLETE")
print("=" * 70)

print(f"Candidate features: {len(ml_feature_columns)}")
print(f"Constant features: {len(constant_features)}")
print(f"Highly skewed features: {len(high_skew_features)}")
print(f"Highly correlated pairs: {len(high_correlation_pairs)}")
print(f"Identical feature pairs: {len(duplicate_feature_pairs)}")
print(f"Leakage candidates: {len(leakage_candidates)}")

print("\n✓ CHECKPOINT 4 FEATURE AUDIT PASSED")
```

```python
# ================================================================
# CHECKPOINT 4
# FINALIZE CONNECTIVITY FEATURE MATRIX
# ================================================================

print("=" * 70)
print("CHECKPOINT 4 — FINALIZE CONNECTIVITY FEATURES")
print("=" * 70)

# ------------------------------------------------
# 1. REMOVE ONLY EXACT DUPLICATE FEATURES
# ------------------------------------------------

features_to_remove = [
    "outgoing_partners",
    "incoming_partners",
    "total_partners"
]

final_ml_feature_columns = [
    col
    for col in ml_feature_columns
    if col not in features_to_remove
]

print("\nRemoved exact duplicate features:")
for feature in features_to_remove:
    print(" -", feature)

print("\nFinal ML features:")
for i, feature in enumerate(final_ml_feature_columns, start=1):
    print(i, feature)

print("\nNumber of final ML features:", len(final_ml_feature_columns))

assert len(final_ml_feature_columns) == 16

# ------------------------------------------------
# 2. CREATE FINAL CONNECTIVITY FEATURE TABLE
# ------------------------------------------------

final_connectivity_features = connectivity_features[
    ["bodyId"] + final_ml_feature_columns
].copy()

# ------------------------------------------------
# 3. FINAL FEATURE AUDIT
# ------------------------------------------------

assert len(final_connectivity_features) == 165
assert final_connectivity_features["bodyId"].nunique() == 165
assert final_connectivity_features[final_ml_feature_columns].isna().sum().sum() == 0

print("\nFinal feature matrix shape:")
print(final_connectivity_features.shape)

print("\n✓ 165 neurons retained.")
print("✓ 16 ML features retained.")
print("✓ No missing feature values.")

# ------------------------------------------------
# 4. SAVE FINAL CONNECTIVITY FEATURES
# ------------------------------------------------

final_connectivity_features.to_csv(
    "data/processed/olfactory_final_connectivity_features.csv",
    index=False
)

print("\n✓ Final connectivity feature matrix saved.")
print("\nCHECKPOINT 4 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 5
# BUILD FINAL ML DATASET
# ================================================================

print("=" * 70)
print("CHECKPOINT 5 — BUILD FINAL ML DATASET")
print("=" * 70)

# ------------------------------------------------
# 1. START FROM THE FINAL CONNECTIVITY FEATURES
# ------------------------------------------------

final_features = final_connectivity_features.copy()

print("\nConnectivity feature matrix:")
print("Shape:", final_features.shape)

# ------------------------------------------------
# 2. SELECT MOLECULAR / GROUPING INFORMATION
# ------------------------------------------------

molecular_metadata = neuron_molecular_targets[
    [
        "bodyId",
        "hemibrain_type",
        "cell_type",
        "neurotransmitter"
    ]
].copy()

print("\nMolecular metadata:")
print("Shape:", molecular_metadata.shape)

# ------------------------------------------------
# 3. CHECK BODY IDs BEFORE MERGING
# ------------------------------------------------

assert final_features["bodyId"].nunique() == 165
assert molecular_metadata["bodyId"].nunique() == 165

assert set(final_features["bodyId"]) == set(
    molecular_metadata["bodyId"]
)

print("\n✓ Body IDs match between connectivity and molecular data.")

# ------------------------------------------------
# 4. MERGE MOLECULAR METADATA WITH CONNECTIVITY
# ------------------------------------------------

final_ml_dataset = molecular_metadata.merge(
    final_features,
    on="bodyId",
    how="inner",
    validate="one_to_one"
)

# ------------------------------------------------
# 5. DEFINE THE FINAL ML FEATURES
# ------------------------------------------------

ml_feature_columns = [
    col
    for col in final_ml_dataset.columns
    if col in final_ml_feature_columns
]

print("\nNumber of ML features:", len(ml_feature_columns))

print("\nML features:")
for i, feature in enumerate(ml_feature_columns, start=1):
    print(i, feature)

assert len(ml_feature_columns) == 16

# ------------------------------------------------
# 6. FINAL DATASET AUDIT
# ------------------------------------------------

print("\nFinal ML dataset shape:")
print(final_ml_dataset.shape)

print("\nNeurotransmitter distribution:")
display(
    final_ml_dataset[
        "neurotransmitter"
    ].value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

# Required columns
required_columns = [
    "bodyId",
    "hemibrain_type",
    "cell_type",
    "neurotransmitter"
]

for column in required_columns:
    assert column in final_ml_dataset.columns

# Row checks
assert len(final_ml_dataset) == 165
assert final_ml_dataset["bodyId"].nunique() == 165

# Molecular label checks
assert final_ml_dataset["cell_type"].notna().all()
assert final_ml_dataset["neurotransmitter"].notna().all()

# Feature checks
assert final_ml_dataset[
    ml_feature_columns
].isna().sum().sum() == 0

assert all(
    np.issubdtype(
        final_ml_dataset[feature].dtype,
        np.number
    )
    for feature in ml_feature_columns
)

print("\n✓ 165 neurons retained.")
print("✓ All neurons have cell-type labels.")
print("✓ All neurons have neurotransmitter labels.")
print("✓ All 16 ML features are numeric.")
print("✓ No missing ML feature values.")

# ------------------------------------------------
# 7. CHECK FOR DIRECT LABEL LEAKAGE
# ------------------------------------------------

forbidden_columns = [
    "bodyId",
    "hemibrain_type",
    "cell_type",
    "neurotransmitter"
]

for column in forbidden_columns:
    assert column not in ml_feature_columns

print("✓ No identifier or molecular-label columns used as ML features.")

# ------------------------------------------------
# 8. SAVE FINAL ML DATASET
# ------------------------------------------------

final_ml_dataset.to_csv(
    "data/processed/final_olfactory_neurotransmitter_ml_dataset.csv",
    index=False
)

# Save feature manifest
final_ml_feature_manifest = pd.DataFrame({
    "feature_number": range(1, len(ml_feature_columns) + 1),
    "feature": ml_feature_columns
})

final_ml_feature_manifest.to_csv(
    "data/processed/final_ml_feature_manifest.csv",
    index=False
)

print("\n✓ Final ML dataset saved:")
print(
    "data/processed/"
    "final_olfactory_neurotransmitter_ml_dataset.csv"
)

print("\n✓ ML feature manifest saved:")
print(
    "data/processed/"
    "final_ml_feature_manifest.csv"
)

print("\n" + "=" * 70)
print("CHECKPOINT 5 — FINAL ML DATASET COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 6
# EXPLORATORY ANALYSIS OF CONNECTIVITY VS MOLECULAR IDENTITY
# ================================================================

print("=" * 70)
print("CHECKPOINT 6 — EXPLORATORY ANALYSIS")
print("=" * 70)

# ------------------------------------------------
# 1. VERIFY FINAL DATASET
# ------------------------------------------------

print("\nFinal ML dataset:")
print("Shape:", final_ml_dataset.shape)

assert final_ml_dataset.shape == (165, 20)

# ------------------------------------------------
# 2. DEFINE THE TWO CLASSES
# ------------------------------------------------

print("\nNeurotransmitter classes:")

class_counts = (
    final_ml_dataset["neurotransmitter"]
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

display(class_counts)

# Confirm expected classes
assert set(
    final_ml_dataset["neurotransmitter"].unique()
) == {"acetylcholine", "gaba"}

# ------------------------------------------------
# 3. CLASS PROPORTIONS
# ------------------------------------------------

class_proportions = (
    final_ml_dataset["neurotransmitter"]
    .value_counts(normalize=True)
    .rename_axis("neurotransmitter")
    .reset_index(name="proportion")
)

class_proportions["percentage"] = (
    class_proportions["proportion"] * 100
)

print("\nClass proportions:")
display(class_proportions)

# ------------------------------------------------
# 4. CHECK GROUPING VARIABLE
# ------------------------------------------------

print("\nCanonical cell-type groups:")

n_cell_types = final_ml_dataset["cell_type"].nunique()

print("Number of canonical cell types:", n_cell_types)

assert n_cell_types == 65

# Number of neurons per canonical type
neurons_per_type = (
    final_ml_dataset
    .groupby("cell_type")
    .size()
    .sort_values(ascending=False)
)

print("\nNeuron count per canonical cell type:")
print(neurons_per_type.describe())

# ------------------------------------------------
# 5. CHECK WHETHER A CELL TYPE HAS MIXED LABELS
# ------------------------------------------------

labels_per_type = (
    final_ml_dataset
    .groupby("cell_type")["neurotransmitter"]
    .nunique()
)

mixed_label_types = labels_per_type[
    labels_per_type > 1
]

print("\nCell types with multiple neurotransmitter labels:")
print(len(mixed_label_types))

assert len(mixed_label_types) == 0

print("✓ Every canonical cell type has one neurotransmitter label.")

# ------------------------------------------------
# 6. IDENTIFY MULTI-NEURON CELL TYPES
# ------------------------------------------------

multi_neuron_types = neurons_per_type[
    neurons_per_type > 1
]

print("\nCanonical cell types represented by multiple neurons:")
print(len(multi_neuron_types))

print("\nLargest groups:")
display(
    multi_neuron_types.head(15)
    .rename("n_neurons")
    .reset_index()
)

# ------------------------------------------------
# 7. FINAL CHECKPOINT 6 SETUP AUDIT
# ------------------------------------------------

print("\n" + "=" * 70)
print("CHECKPOINT 6 SETUP AUDIT PASSED")
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 6
# FEATURE DISTRIBUTION SUMMARY
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.1 — FEATURE DISTRIBUTIONS")
print("=" * 70)

feature_summary = final_ml_dataset[ml_feature_columns].describe().T

feature_summary["skewness"] = (
    final_ml_dataset[ml_feature_columns].skew()
)

feature_summary["missing_values"] = (
    final_ml_dataset[ml_feature_columns].isna().sum()
)

display(feature_summary)

# ------------------------------------------------
# CHECKS
# ------------------------------------------------

assert len(feature_summary) == 16
assert feature_summary["missing_values"].sum() == 0

# Identify strongly skewed features
high_skew_features = feature_summary[
    feature_summary["skewness"].abs() > 2
].index.tolist()

print("\nFeatures with |skewness| > 2:")
for feature in high_skew_features:
    print("-", feature)

print("\n✓ Feature distribution summary complete.")
```

```python
# ================================================================
# CHECKPOINT 6
# FEATURE DISTRIBUTION PLOTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.2 — FEATURE DISTRIBUTION PLOTS")
print("=" * 70)

for feature in ml_feature_columns:

    plt.figure(figsize=(7, 4))

    sns.histplot(
        data=final_ml_dataset,
        x=feature,
        bins=20,
        kde=True
    )

    plt.title("Distribution of " + feature)
    plt.xlabel(feature)
    plt.ylabel("Number of neurons")
    plt.tight_layout()
    plt.show()

print("\n✓ Distribution plots generated.")
```

```python
# ================================================================
# CHECKPOINT 6
# MANN-WHITNEY U TESTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.5 — NON-PARAMETRIC GROUP COMPARISON")
print("=" * 70)

from scipy.stats import mannwhitneyu

ach_mask = (
    final_ml_dataset["neurotransmitter"] == "acetylcholine"
)

gaba_mask = (
    final_ml_dataset["neurotransmitter"] == "gaba"
)

statistical_results = []

for feature in ml_feature_columns:

    ach_values = final_ml_dataset.loc[
        ach_mask, feature
    ]

    gaba_values = final_ml_dataset.loc[
        gaba_mask, feature
    ]

    statistic, p_value = mannwhitneyu(
        ach_values,
        gaba_values,
        alternative="two-sided"
    )

    statistical_results.append({
        "feature": feature,
        "mann_whitney_U": statistic,
        "p_value": p_value
    })

mann_whitney_results = pd.DataFrame(
    statistical_results
)

# Sort by p-value
mann_whitney_results = (
    mann_whitney_results
    .sort_values("p_value")
    .reset_index(drop=True)
)

display(mann_whitney_results)

print("\n✓ Mann–Whitney tests completed.")
```

```python
# ================================================================
# CHECKPOINT 6
# EFFECT SIZE — RANK-BISERIAL CORRELATION
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.6 — EFFECT SIZES")
print("=" * 70)

effect_size_results = []

n_ach = ach_mask.sum()
n_gaba = gaba_mask.sum()

for feature in ml_feature_columns:

    ach_values = final_ml_dataset.loc[
        ach_mask, feature
    ].values

    gaba_values = final_ml_dataset.loc[
        gaba_mask, feature
    ].values

    U, p_value = mannwhitneyu(
        ach_values,
        gaba_values,
        alternative="two-sided"
    )

    # Rank-biserial correlation
    rank_biserial = (
        (2 * U) / (n_ach * n_gaba)
    ) - 1

    effect_size_results.append({
        "feature": feature,
        "U": U,
        "p_value": p_value,
        "rank_biserial_correlation": rank_biserial,
        "absolute_effect_size": abs(rank_biserial)
    })

effect_size_results = pd.DataFrame(
    effect_size_results
)

effect_size_results = (
    effect_size_results
    .sort_values(
        "absolute_effect_size",
        ascending=False
    )
    .reset_index(drop=True)
)

display(effect_size_results)

print("\n✓ Effect-size analysis complete.")
```

```python
# ================================================================
# CHECKPOINT 6
# MULTIPLE-TESTING CORRECTION
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.7 — MULTIPLE TESTING CORRECTION")
print("=" * 70)

from scipy.stats import false_discovery_control

corrected_results = mann_whitney_results.copy()

corrected_results["q_value"] = (
    false_discovery_control(
        corrected_results["p_value"].values,
        method="bh"
    )
)

corrected_results["significant_fdr_0.05"] = (
    corrected_results["q_value"] < 0.05
)

display(corrected_results)

print("\nNumber of features with FDR q < 0.05:")
print(
    corrected_results[
        "significant_fdr_0.05"
    ].sum()
)

print("\n✓ Benjamini–Hochberg correction completed.")
```

```python
# ================================================================
# CHECKPOINT 6
# FINAL FEATURE CORRELATION MATRIX
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.8 — FEATURE CORRELATIONS")
print("=" * 70)

final_correlation_matrix = (
    final_ml_dataset[ml_feature_columns]
    .corr(method="spearman")
)

plt.figure(figsize=(14, 11))

sns.heatmap(
    final_correlation_matrix,
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title(
    "Spearman Correlation of Final Connectivity Features"
)

plt.tight_layout()
plt.show()

# Save matrix
final_correlation_matrix.to_csv(
    "data/processed/final_ml_feature_spearman_correlation.csv"
)

print("\n✓ Correlation matrix generated.")
print("✓ Correlation matrix saved.")
```

```python
# ================================================================
# CHECKPOINT 6
# HIGH CORRELATION PAIRS
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.9 — CORRELATION AUDIT")
print("=" * 70)

correlation_pairs = []

for i in range(len(ml_feature_columns)):

    for j in range(i + 1, len(ml_feature_columns)):

        feature_a = ml_feature_columns[i]
        feature_b = ml_feature_columns[j]

        correlation = final_correlation_matrix.loc[
            feature_a,
            feature_b
        ]

        correlation_pairs.append({
            "feature_1": feature_a,
            "feature_2": feature_b,
            "spearman_rho": correlation,
            "absolute_rho": abs(correlation)
        })

correlation_pairs = pd.DataFrame(
    correlation_pairs
)

high_correlation_pairs = (
    correlation_pairs[
        correlation_pairs["absolute_rho"] >= 0.90
    ]
    .sort_values(
        "absolute_rho",
        ascending=False
    )
    .reset_index(drop=True)
)

print(
    "Number of feature pairs with |rho| >= 0.90:",
    len(high_correlation_pairs)
)

display(high_correlation_pairs)

print("\n✓ Correlation audit complete.")
```

```python
# ================================================================
# CHECKPOINT 6
# PCA OF CONNECTIVITY FEATURES
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.10 — PCA")
print("=" * 70)

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X_exploratory = final_ml_dataset[
    ml_feature_columns
].copy()

# Standardize features because they have very different scales
scaler_exploratory = StandardScaler()

X_scaled_exploratory = (
    scaler_exploratory.fit_transform(
        X_exploratory
    )
)

# PCA
pca_exploratory = PCA()

X_pca = pca_exploratory.fit_transform(
    X_scaled_exploratory
)

explained_variance = pd.DataFrame({
    "PC": [
        "PC" + str(i + 1)
        for i in range(len(
            pca_exploratory.explained_variance_ratio_
        ))
    ],
    "explained_variance_ratio":
        pca_exploratory.explained_variance_ratio_
})

explained_variance["cumulative_variance"] = (
    explained_variance[
        "explained_variance_ratio"
    ].cumsum()
)

display(
    explained_variance.head(10)
)

print(
    "\nVariance explained by first two PCs:",
    round(
        explained_variance.loc[
            :1,
            "explained_variance_ratio"
        ].sum(),
        4
    )
)

print("\n✓ PCA completed.")
```

```python
# ================================================================
# CHECKPOINT 6
# PCA VISUALIZATION
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.11 — PCA VISUALIZATION")
print("=" * 70)

pca_plot_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "neurotransmitter":
        final_ml_dataset["neurotransmitter"].values,
    "cell_type":
        final_ml_dataset["cell_type"].values
})

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=pca_plot_df,
    x="PC1",
    y="PC2",
    hue="neurotransmitter",
    s=70
)

plt.title(
    "PCA of Connectivity Features"
)

plt.xlabel(
    "PC1 "
    + str(
        round(
            explained_variance.loc[
                0,
                "explained_variance_ratio"
            ] * 100,
            1
        )
    )
    + "% variance"
)

plt.ylabel(
    "PC2 "
    + str(
        round(
            explained_variance.loc[
                1,
                "explained_variance_ratio"
            ] * 100,
            1
        )
    )
    + "% variance"
)

plt.tight_layout()
plt.show()

print("\n✓ PCA visualization generated.")
```

```python
# ================================================================
# CHECKPOINT 6
# PCA FEATURE LOADINGS
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.12 — PCA LOADINGS")
print("=" * 70)

pca_loadings = pd.DataFrame(
    pca_exploratory.components_[:2].T,
    index=ml_feature_columns,
    columns=["PC1_loading", "PC2_loading"]
)

pca_loadings["PC1_absolute"] = (
    pca_loadings["PC1_loading"].abs()
)

pca_loadings["PC2_absolute"] = (
    pca_loadings["PC2_loading"].abs()
)

display(
    pca_loadings.sort_values(
        "PC1_absolute",
        ascending=False
    )
)

print("\n✓ PCA loading analysis complete.")
```

```python
# ================================================================
# CHECKPOINT 6
# DM3_vPN MAPPING SENSITIVITY
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.13 — DM3_vPN SENSITIVITY")
print("=" * 70)

dm3_mask = (
    final_ml_dataset["cell_type"] == "DM3_vPN"
)

dm3_data = final_ml_dataset.loc[
    dm3_mask,
    [
        "bodyId",
        "hemibrain_type",
        "cell_type",
        "neurotransmitter"
    ] + ml_feature_columns
].copy()

print("\nDM3_vPN neurons:")
print("Number:", len(dm3_data))

display(dm3_data)

# ------------------------------------------------
# Compare DM3_vPN feature means with all neurons
# ------------------------------------------------

dm3_feature_comparison = pd.DataFrame({
    "feature": ml_feature_columns,
    "DM3_vPN_mean": [
        dm3_data[feature].mean()
        for feature in ml_feature_columns
    ],
    "all_neurons_mean": [
        final_ml_dataset[feature].mean()
        for feature in ml_feature_columns
    ]
})

dm3_feature_comparison["difference"] = (
    dm3_feature_comparison["DM3_vPN_mean"]
    -
    dm3_feature_comparison["all_neurons_mean"]
)

display(dm3_feature_comparison)

print("\n✓ DM3_vPN sensitivity inspection complete.")
```

```python
# ================================================================
# CHECKPOINT 6
# CELL-TYPE-LEVEL EXPLORATORY DATASET
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.14 — CELL-TYPE LEVEL SUMMARY")
print("=" * 70)

cell_type_exploratory = (
    final_ml_dataset
    .groupby(
        [
            "cell_type",
            "neurotransmitter"
        ],
        as_index=False
    )[ml_feature_columns]
    .mean()
)

print("\nCell-type-level dataset shape:")
print(cell_type_exploratory.shape)

print(
    "\nNumber of canonical cell types:",
    cell_type_exploratory["cell_type"].nunique()
)

print("\nNeurotransmitter distribution:")
display(
    cell_type_exploratory[
        "neurotransmitter"
    ].value_counts()
)

assert len(cell_type_exploratory) == 65
assert cell_type_exploratory["cell_type"].nunique() == 65

print("\n✓ Cell-type-level exploratory dataset created.")
```

```python
# ================================================================
# CHECKPOINT 6
# NEURON-LEVEL VS CELL-TYPE-LEVEL COMPARISON
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.15 — GROUPING STRUCTURE AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. NEURON-LEVEL COUNTS
# ------------------------------------------------

neuron_level_counts = (
    final_ml_dataset["neurotransmitter"]
    .value_counts()
    .rename("neuron_count")
)

# ------------------------------------------------
# 2. CELL-TYPE-LEVEL COUNTS
# ------------------------------------------------

cell_type_level_counts = (
    cell_type_exploratory["neurotransmitter"]
    .value_counts()
    .rename("cell_type_count")
)

# ------------------------------------------------
# 3. COMBINE
# ------------------------------------------------

grouping_comparison = pd.concat(
    [
        neuron_level_counts,
        cell_type_level_counts
    ],
    axis=1
)

display(grouping_comparison)

# ------------------------------------------------
# 4. BASIC GROUPING INFORMATION
# ------------------------------------------------

print("\nNeuron-level observations:")
print(len(final_ml_dataset))

print("\nCanonical cell-type groups:")
print(final_ml_dataset["cell_type"].nunique())

print("\nCell types represented by more than one neuron:")
print(
    (
        final_ml_dataset
        .groupby("cell_type")
        .size()
        .gt(1)
        .sum()
    )
)

# ------------------------------------------------
# 5. AUDIT
# ------------------------------------------------

assert len(final_ml_dataset) == 165
assert final_ml_dataset["cell_type"].nunique() == 65
assert len(cell_type_exploratory) == 65

print("\n✓ 165 neuron-level observations confirmed.")
print("✓ 65 canonical cell-type groups confirmed.")
print("✓ Neuron-level and cell-type-level grouping documented.")
print("\n✓ CHECKPOINT 6.15 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 6
# EXPLORATORY RESULTS SUMMARY
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.16 — EXPLORATORY RESULTS SUMMARY")
print("=" * 70)

# ------------------------------------------------
# 1. IDENTIFY TOP FEATURES BY EFFECT SIZE
# ------------------------------------------------

top_effect_features = (
    effect_size_results[
        [
            "feature",
            "rank_biserial_correlation",
            "absolute_effect_size",
            "p_value"
        ]
    ]
    .copy()
)

# ------------------------------------------------
# 2. ADD FDR VALUES
# ------------------------------------------------

top_effect_features = top_effect_features.merge(
    corrected_results[
        [
            "feature",
            "q_value",
            "significant_fdr_0.05"
        ]
    ],
    on="feature",
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------
# 3. SORT BY EFFECT SIZE
# ------------------------------------------------

top_effect_features = (
    top_effect_features
    .sort_values(
        "absolute_effect_size",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nFeatures ranked by effect size:")
display(top_effect_features)

# ------------------------------------------------
# 4. HIGHLY CORRELATED PAIRS
# ------------------------------------------------

print("\nNumber of feature pairs with |Spearman rho| >= 0.90:")
print(len(high_correlation_pairs))

# ------------------------------------------------
# 5. PCA SUMMARY
# ------------------------------------------------

pc1_variance = (
    pca_exploratory
    .explained_variance_ratio_[0]
)

pc2_variance = (
    pca_exploratory
    .explained_variance_ratio_[1]
)

print("\nPCA variance explained:")
print("PC1:", round(pc1_variance * 100, 2), "%")
print("PC2:", round(pc2_variance * 100, 2), "%")
print(
    "PC1 + PC2:",
    round((pc1_variance + pc2_variance) * 100, 2),
    "%"
)

# ------------------------------------------------
# 6. SAVE SUMMARY
# ------------------------------------------------

top_effect_features.to_csv(
    "data/processed/checkpoint6_exploratory_feature_ranking.csv",
    index=False
)

print("\n✓ Exploratory feature ranking saved.")
print("✓ CHECKPOINT 6.16 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 6
# SAVE PCA AND EXPLORATORY VISUALIZATION DATA
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.17 — SAVE PCA DATA")
print("=" * 70)

# ------------------------------------------------
# 1. SAVE PCA COORDINATES
# ------------------------------------------------

pca_plot_df.to_csv(
    "data/processed/checkpoint6_pca_coordinates.csv",
    index=False
)

# ------------------------------------------------
# 2. SAVE PCA LOADINGS
# ------------------------------------------------

pca_loadings.to_csv(
    "data/processed/checkpoint6_pca_feature_loadings.csv"
)

# ------------------------------------------------
# 3. SAVE EXPLAINED VARIANCE
# ------------------------------------------------

explained_variance.to_csv(
    "data/processed/checkpoint6_pca_variance.csv",
    index=False
)

print("\n✓ PCA coordinates saved.")
print("✓ PCA feature loadings saved.")
print("✓ PCA explained variance saved.")

print("\n✓ CHECKPOINT 6.17 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 6
# SAVE COMPLETE CHECKPOINT 6 AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 6.18 — SAVE COMPLETE AUDIT")
print("=" * 70)

# ------------------------------------------------
# FEATURE AUDIT
# ------------------------------------------------

checkpoint6_feature_audit = pd.DataFrame({
    "feature": ml_feature_columns,
    "mean": [
        final_ml_dataset[f].mean()
        for f in ml_feature_columns
    ],
    "median": [
        final_ml_dataset[f].median()
        for f in ml_feature_columns
    ],
    "std": [
        final_ml_dataset[f].std()
        for f in ml_feature_columns
    ],
    "skewness": [
        final_ml_dataset[f].skew()
        for f in ml_feature_columns
    ]
})

# ------------------------------------------------
# MERGE STATISTICAL RESULTS
# ------------------------------------------------

checkpoint6_feature_audit = (
    checkpoint6_feature_audit
    .merge(
        effect_size_results[
            [
                "feature",
                "rank_biserial_correlation",
                "absolute_effect_size"
            ]
        ],
        on="feature",
        how="left",
        validate="one_to_one"
    )
    .merge(
        corrected_results[
            [
                "feature",
                "p_value",
                "q_value",
                "significant_fdr_0.05"
            ]
        ],
        on="feature",
        how="left",
        validate="one_to_one"
    )
)

display(checkpoint6_feature_audit)

# ------------------------------------------------
# SAVE
# ------------------------------------------------

checkpoint6_feature_audit.to_csv(
    "data/processed/checkpoint6_complete_feature_audit.csv",
    index=False
)

print("\n✓ Complete feature audit saved.")
print("\n✓ CHECKPOINT 6.18 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 6
# FINAL AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 6 — FINAL SCIENTIFIC AUDIT")
print("=" * 70)

# ------------------------------------------------
# DATASET
# ------------------------------------------------

assert final_ml_dataset.shape == (165, 20)

assert final_ml_dataset["bodyId"].nunique() == 165

assert final_ml_dataset["cell_type"].nunique() == 65

# ------------------------------------------------
# FEATURES
# ------------------------------------------------

assert len(ml_feature_columns) == 16

assert final_ml_dataset[
    ml_feature_columns
].isna().sum().sum() == 0

# ------------------------------------------------
# TARGET
# ------------------------------------------------

assert set(
    final_ml_dataset["neurotransmitter"].unique()
) == {
    "acetylcholine",
    "gaba"
}

# ------------------------------------------------
# GROUPING
# ------------------------------------------------

labels_per_type = (
    final_ml_dataset
    .groupby("cell_type")["neurotransmitter"]
    .nunique()
)

assert (labels_per_type == 1).all()

# ------------------------------------------------
# STATISTICAL ANALYSIS
# ------------------------------------------------

assert len(mann_whitney_results) == 16

assert len(effect_size_results) == 16

assert len(corrected_results) == 16

# ------------------------------------------------
# CORRELATION ANALYSIS
# ------------------------------------------------

assert final_correlation_matrix.shape == (16, 16)

# ------------------------------------------------
# PCA
# ------------------------------------------------

assert X_pca.shape[0] == 165

assert X_pca.shape[1] == 16

assert len(pca_loadings) == 16

# ------------------------------------------------
# CELL-TYPE LEVEL ANALYSIS
# ------------------------------------------------

assert len(cell_type_exploratory) == 65

# ------------------------------------------------
# DM3_vPN
# ------------------------------------------------

assert "DM3_vPN" in final_ml_dataset["cell_type"].values

# ------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------

print("\nDATASET")
print("-------")
print("Neurons:", len(final_ml_dataset))
print("Canonical cell types:", final_ml_dataset["cell_type"].nunique())
print("Connectivity features:", len(ml_feature_columns))

print("\nTARGET")
print("------")
print(
    final_ml_dataset[
        "neurotransmitter"
    ].value_counts()
)

print("\nEXPLORATORY ANALYSES")
print("--------------------")
print("Mann–Whitney tests:", len(mann_whitney_results))
print("Effect-size analyses:", len(effect_size_results))
print("Correlation matrix: 16 × 16")
print("PCA dimensions:", X_pca.shape)
print("Cell-type-level observations:", len(cell_type_exploratory))

print("\nGROUPING")
print("--------")
print("165 neuron observations")
print("65 canonical cell-type groups")
print("43 cell types represented by multiple neurons")

print("\n" + "=" * 70)
print("✓ CHECKPOINT 6 COMPLETE")
print("=" * 70)
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 7
# ESTABLISH BASELINE MODELS
# ================================================================

print("=" * 70)
print("CHECKPOINT 7 — BASELINE MODELS")
print("=" * 70)

# ------------------------------------------------
# 1. VERIFY FINAL DATASET
# ------------------------------------------------

print("\nFinal dataset shape:")
print(final_ml_dataset.shape)

assert final_ml_dataset.shape == (165, 20)

# ------------------------------------------------
# 2. DEFINE FEATURES AND TARGET
# ------------------------------------------------

X_baseline = final_ml_dataset[
    ml_feature_columns
].copy()

y_baseline = final_ml_dataset[
    "neurotransmitter"
].copy()

print("\nFeature matrix shape:")
print(X_baseline.shape)

print("\nTarget vector shape:")
print(y_baseline.shape)

print("\nTarget distribution:")
display(
    y_baseline
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

# ------------------------------------------------
# 3. BASIC CHECKS
# ------------------------------------------------

assert X_baseline.shape == (165, 16)
assert len(y_baseline) == 165
assert X_baseline.isna().sum().sum() == 0

assert set(y_baseline.unique()) == {
    "acetylcholine",
    "gaba"
}

print("\n✓ 165 observations confirmed.")
print("✓ 16 connectivity features confirmed.")
print("✓ No missing feature values.")
print("✓ Two neurotransmitter classes confirmed.")

print("\n✓ CHECKPOINT 7 SETUP COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 7.1
# STRATIFIED TRAIN/TEST SPLIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.1 — TRAIN/TEST SPLIT")
print("=" * 70)

from sklearn.model_selection import train_test_split

X_train_base, X_test_base, y_train_base, y_test_base = (
    train_test_split(
        X_baseline,
        y_baseline,
        test_size=0.25,
        random_state=42,
        stratify=y_baseline
    )
)

print("\nTraining observations:")
print(len(X_train_base))

print("\nTest observations:")
print(len(X_test_base))

print("\nTraining class distribution:")
display(
    y_train_base
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

print("\nTest class distribution:")
display(
    y_test_base
    .value_counts()
    .rename_axis("neurotransmitter")
    .reset_index(name="n_neurons")
)

# ------------------------------------------------
# CHECKS
# ------------------------------------------------

assert len(X_train_base) + len(X_test_base) == 165
assert len(y_train_base) + len(y_test_base) == 165

assert set(y_train_base.unique()) == {
    "acetylcholine",
    "gaba"
}

assert set(y_test_base.unique()) == {
    "acetylcholine",
    "gaba"
}

print("\n✓ Stratified split completed.")
print("✓ Both classes present in training and test sets.")
```

```python
# ================================================================
# CHECKPOINT 7.2
# SCALE FEATURES FOR LOGISTIC REGRESSION
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.2 — FEATURE STANDARDIZATION")
print("=" * 70)

from sklearn.preprocessing import StandardScaler

baseline_scaler = StandardScaler()

X_train_scaled = baseline_scaler.fit_transform(
    X_train_base
)

X_test_scaled = baseline_scaler.transform(
    X_test_base
)

print("\nTraining matrix:")
print(X_train_scaled.shape)

print("\nTest matrix:")
print(X_test_scaled.shape)

# ------------------------------------------------
# CHECKS
# ------------------------------------------------

assert X_train_scaled.shape[1] == 16
assert X_test_scaled.shape[1] == 16

print("\n✓ Scaler fitted only on training data.")
print("✓ Test data transformed without refitting.")
```

```python
# ================================================================
# CHECKPOINT 7.3
# DUMMY CLASSIFIER BASELINE
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.3 — DUMMY CLASSIFIER")
print("=" * 70)

from sklearn.dummy import DummyClassifier

dummy_model = DummyClassifier(
    strategy="most_frequent"
)

dummy_model.fit(
    X_train_base,
    y_train_base
)

dummy_predictions = dummy_model.predict(
    X_test_base
)

print("\nDummy prediction distribution:")
display(
    pd.Series(dummy_predictions)
    .value_counts()
    .rename_axis("predicted_class")
    .reset_index(name="n_predictions")
)

print("\n✓ Dummy classifier trained.")
```

```python
# ================================================================
# CHECKPOINT 7.4
# DUMMY CLASSIFIER METRICS
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.4 — DUMMY BASELINE METRICS")
print("=" * 70)

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    recall_score,
    confusion_matrix
)

dummy_accuracy = accuracy_score(
    y_test_base,
    dummy_predictions
)

dummy_balanced_accuracy = balanced_accuracy_score(
    y_test_base,
    dummy_predictions
)

dummy_macro_f1 = f1_score(
    y_test_base,
    dummy_predictions,
    average="macro",
    zero_division=0
)

dummy_gaba_recall = recall_score(
    y_test_base,
    dummy_predictions,
    pos_label="gaba",
    zero_division=0
)

dummy_ach_recall = recall_score(
    y_test_base,
    dummy_predictions,
    pos_label="acetylcholine",
    zero_division=0
)

dummy_cm = confusion_matrix(
    y_test_base,
    dummy_predictions,
    labels=["acetylcholine", "gaba"]
)

print("\nAccuracy:")
print(round(dummy_accuracy, 4))

print("\nBalanced accuracy:")
print(round(dummy_balanced_accuracy, 4))

print("\nMacro F1:")
print(round(dummy_macro_f1, 4))

print("\nACh recall:")
print(round(dummy_ach_recall, 4))

print("\nGABA recall:")
print(round(dummy_gaba_recall, 4))

print("\nConfusion matrix:")
print(dummy_cm)

print("\n✓ Dummy baseline evaluated.")
```

```python
# ================================================================
# CHECKPOINT 7.5
# LOGISTIC REGRESSION BASELINE
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.5 — LOGISTIC REGRESSION")
print("=" * 70)

from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

logistic_model.fit(
    X_train_scaled,
    y_train_base
)

logistic_predictions = logistic_model.predict(
    X_test_scaled
)

print("\n✓ Logistic Regression trained.")
print("\nPredicted classes:")
display(
    pd.Series(logistic_predictions)
    .value_counts()
    .rename_axis("predicted_class")
    .reset_index(name="n_predictions")
)
```

```python
# ================================================================
# CHECKPOINT 7.6
# LOGISTIC REGRESSION METRICS
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.6 — LOGISTIC REGRESSION METRICS")
print("=" * 70)

logistic_accuracy = accuracy_score(
    y_test_base,
    logistic_predictions
)

logistic_balanced_accuracy = balanced_accuracy_score(
    y_test_base,
    logistic_predictions
)

logistic_macro_f1 = f1_score(
    y_test_base,
    logistic_predictions,
    average="macro",
    zero_division=0
)

logistic_gaba_recall = recall_score(
    y_test_base,
    logistic_predictions,
    pos_label="gaba",
    zero_division=0
)

logistic_ach_recall = recall_score(
    y_test_base,
    logistic_predictions,
    pos_label="acetylcholine",
    zero_division=0
)

logistic_cm = confusion_matrix(
    y_test_base,
    logistic_predictions,
    labels=["acetylcholine", "gaba"]
)

print("\nAccuracy:")
print(round(logistic_accuracy, 4))

print("\nBalanced accuracy:")
print(round(logistic_balanced_accuracy, 4))

print("\nMacro F1:")
print(round(logistic_macro_f1, 4))

print("\nACh recall:")
print(round(logistic_ach_recall, 4))

print("\nGABA recall:")
print(round(logistic_gaba_recall, 4))

print("\nConfusion matrix:")
print(logistic_cm)

print("\n✓ Logistic Regression evaluated.")
```

```python
# ================================================================
# CHECKPOINT 7.7
# CLASSIFICATION REPORTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.7 — CLASSIFICATION REPORT")
print("=" * 70)

from sklearn.metrics import classification_report

print("\nDUMMY CLASSIFIER")
print("-" * 50)

print(
    classification_report(
        y_test_base,
        dummy_predictions,
        labels=["acetylcholine", "gaba"],
        zero_division=0
    )
)

print("\nLOGISTIC REGRESSION")
print("-" * 50)

print(
    classification_report(
        y_test_base,
        logistic_predictions,
        labels=["acetylcholine", "gaba"],
        zero_division=0
    )
)

print("\n✓ Classification reports generated.")
```

```python
# ================================================================
# CHECKPOINT 7.8
# BASELINE MODEL COMPARISON
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.8 — BASELINE COMPARISON")
print("=" * 70)

baseline_results = pd.DataFrame({
    "model": [
        "DummyClassifier",
        "LogisticRegression"
    ],
    "accuracy": [
        dummy_accuracy,
        logistic_accuracy
    ],
    "balanced_accuracy": [
        dummy_balanced_accuracy,
        logistic_balanced_accuracy
    ],
    "macro_f1": [
        dummy_macro_f1,
        logistic_macro_f1
    ],
    "ACh_recall": [
        dummy_ach_recall,
        logistic_ach_recall
    ],
    "GABA_recall": [
        dummy_gaba_recall,
        logistic_gaba_recall
    ]
})

display(baseline_results)

print("\n✓ Baseline models compared.")
```

```python
# ================================================================
# CHECKPOINT 7.9
# LOGISTIC REGRESSION FEATURE COEFFICIENTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.9 — LOGISTIC REGRESSION COEFFICIENTS")
print("=" * 70)

# Determine which class is represented by the positive coefficient
print("\nModel classes:")
print(logistic_model.classes_)

logistic_coefficients = pd.DataFrame({
    "feature": ml_feature_columns,
    "coefficient": logistic_model.coef_[0]
})

logistic_coefficients["absolute_coefficient"] = (
    logistic_coefficients["coefficient"].abs()
)

logistic_coefficients = (
    logistic_coefficients
    .sort_values(
        "absolute_coefficient",
        ascending=False
    )
    .reset_index(drop=True)
)

display(logistic_coefficients)

print("\n✓ Logistic Regression coefficients extracted.")
```

```python
# ================================================================
# CHECKPOINT 7.10
# CONFUSION MATRICES
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.10 — CONFUSION MATRICES")
print("=" * 70)

plt.figure(figsize=(6, 5))

sns.heatmap(
    dummy_cm,
    annot=True,
    fmt="d",
    xticklabels=["acetylcholine", "gaba"],
    yticklabels=["acetylcholine", "gaba"]
)

plt.title("Dummy Classifier Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()


plt.figure(figsize=(6, 5))

sns.heatmap(
    logistic_cm,
    annot=True,
    fmt="d",
    xticklabels=["acetylcholine", "gaba"],
    yticklabels=["acetylcholine", "gaba"]
)

plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

print("\n✓ Confusion matrices generated.")
```

```python
# ================================================================
# CHECKPOINT 7.11
# SAVE BASELINE RESULTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 7.11 — SAVE BASELINE RESULTS")
print("=" * 70)

# ------------------------------------------------
# SAVE MODEL COMPARISON
# ------------------------------------------------

baseline_results.to_csv(
    "results/tables/checkpoint7_baseline_model_comparison.csv",
    index=False
)

# ------------------------------------------------
# SAVE LOGISTIC COEFFICIENTS
# ------------------------------------------------

logistic_coefficients.to_csv(
    "results/tables/checkpoint7_logistic_coefficients.csv",
    index=False
)

# ------------------------------------------------
# SAVE CONFUSION MATRICES
# ------------------------------------------------

dummy_cm_df = pd.DataFrame(
    dummy_cm,
    index=["actual_acetylcholine", "actual_gaba"],
    columns=["predicted_acetylcholine", "predicted_gaba"]
)

logistic_cm_df = pd.DataFrame(
    logistic_cm,
    index=["actual_acetylcholine", "actual_gaba"],
    columns=["predicted_acetylcholine", "predicted_gaba"]
)

dummy_cm_df.to_csv(
    "results/tables/checkpoint7_dummy_confusion_matrix.csv"
)

logistic_cm_df.to_csv(
    "results/tables/checkpoint7_logistic_confusion_matrix.csv"
)

print("\n✓ Baseline comparison saved.")
print("✓ Logistic coefficients saved.")
print("✓ Dummy confusion matrix saved.")
print("✓ Logistic confusion matrix saved.")

print("\n✓ CHECKPOINT 7.11 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 7
# FINAL SCIENTIFIC AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 7 — FINAL AUDIT")
print("=" * 70)

# ------------------------------------------------
# DATA
# ------------------------------------------------

assert X_baseline.shape == (165, 16)
assert len(y_baseline) == 165

# ------------------------------------------------
# SPLIT
# ------------------------------------------------

assert len(X_train_base) == len(y_train_base)
assert len(X_test_base) == len(y_test_base)

assert len(X_train_base) + len(X_test_base) == 165

# ------------------------------------------------
# MODELS
# ------------------------------------------------

assert len(dummy_predictions) == len(y_test_base)

assert len(logistic_predictions) == len(y_test_base)

# ------------------------------------------------
# METRICS
# ------------------------------------------------

assert 0 <= dummy_balanced_accuracy <= 1
assert 0 <= logistic_balanced_accuracy <= 1

assert 0 <= dummy_macro_f1 <= 1
assert 0 <= logistic_macro_f1 <= 1

# ------------------------------------------------
# RESULTS TABLE
# ------------------------------------------------

assert len(baseline_results) == 2

# ------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------

print("\nDATA")
print("----")
print("Total neurons:", len(final_ml_dataset))
print("Connectivity features:", len(ml_feature_columns))
print("Canonical cell types:", final_ml_dataset["cell_type"].nunique())

print("\nBASELINE MODELS")
print("---------------")

display(
    baseline_results
)

print("\nKEY METRICS")
print("-----------")

print(
    "Dummy balanced accuracy:",
    round(dummy_balanced_accuracy, 4)
)

print(
    "Logistic balanced accuracy:",
    round(logistic_balanced_accuracy, 4)
)

print(
    "Dummy macro F1:",
    round(dummy_macro_f1, 4)
)

print(
    "Logistic macro F1:",
    round(logistic_macro_f1, 4)
)

print("\n" + "=" * 70)
print("✓ CHECKPOINT 7 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 8
# NONLINEAR MOLECULAR-IDENTITY PREDICTORS
# ================================================================

print("=" * 70)
print("CHECKPOINT 8 — RANDOM FOREST + XGBOOST")
print("=" * 70)

# Reuse the exact train/test split from Checkpoint 7
assert X_train_base.shape == (123, 16)
assert X_test_base.shape == (42, 16)

assert len(y_train_base) == 123
assert len(y_test_base) == 42

assert len(ml_feature_columns) == 16

print("\nTraining observations:")
print(len(X_train_base))

print("\nTest observations:")
print(len(X_test_base))

print("\nConnectivity features:")
print(len(ml_feature_columns))

print("\nTraining class distribution:")
display(
    y_train_base.value_counts()
)

print("\nTest class distribution:")
display(
    y_test_base.value_counts()
)

print("\n✓ Checkpoint 7 train/test split confirmed.")
print("✓ Same split will be used for all nonlinear models.")
```

```python
# ================================================================
# CHECKPOINT 8.1 — MODEL DEPENDENCIES
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.1 — MODEL DEPENDENCIES")
print("=" * 70)

# Random Forest
from sklearn.ensemble import RandomForestClassifier

# XGBoost
import importlib.util

if importlib.util.find_spec("xgboost") is None:
    print("\nXGBoost not found. Installing...")
    !pip install -q xgboost

from xgboost import XGBClassifier

# Required evaluation functions
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    recall_score,
    confusion_matrix,
    classification_report
)

print("\n✓ RandomForestClassifier imported.")
print("✓ XGBClassifier imported.")
print("✓ Evaluation metrics imported.")
print("✓ Model dependencies ready.")
```

```python
# ================================================================
# CHECKPOINT 8.2 — RANDOM FOREST
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.2 — RANDOM FOREST")
print("=" * 70)

rf_model = RandomForestClassifier(
    n_estimators=500,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train_base,
    y_train_base
)

rf_predictions = rf_model.predict(
    X_test_base
)

print("\n✓ Random Forest trained.")

print("\nPredicted class distribution:")
display(
    pd.Series(
        rf_predictions,
        name="predicted_class"
    ).value_counts()
)
```

```python
# ================================================================
# CHECKPOINT 8.3 — RANDOM FOREST METRICS
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.3 — RANDOM FOREST METRICS")
print("=" * 70)

rf_accuracy = accuracy_score(
    y_test_base,
    rf_predictions
)

rf_balanced_accuracy = balanced_accuracy_score(
    y_test_base,
    rf_predictions
)

rf_macro_f1 = f1_score(
    y_test_base,
    rf_predictions,
    average="macro"
)

rf_class_recalls = recall_score(
    y_test_base,
    rf_predictions,
    labels=["acetylcholine", "gaba"],
    average=None
)

rf_ach_recall = rf_class_recalls[0]
rf_gaba_recall = rf_class_recalls[1]

rf_confusion_matrix = confusion_matrix(
    y_test_base,
    rf_predictions,
    labels=["acetylcholine", "gaba"]
)

print("\nAccuracy:")
print(f"{rf_accuracy:.4f}")

print("\nBalanced accuracy:")
print(f"{rf_balanced_accuracy:.4f}")

print("\nMacro F1:")
print(f"{rf_macro_f1:.4f}")

print("\nACh recall:")
print(f"{rf_ach_recall:.4f}")

print("\nGABA recall:")
print(f"{rf_gaba_recall:.4f}")

print("\nConfusion matrix:")
print(rf_confusion_matrix)

print("\n✓ Random Forest evaluated.")
```

```python
# ================================================================
# CHECKPOINT 8.4 — RANDOM FOREST CLASSIFICATION REPORT
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.4 — RANDOM FOREST CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test_base,
        rf_predictions,
        labels=["acetylcholine", "gaba"],
        digits=4
    )
)

print("✓ Random Forest classification report generated.")
```

```python
# ================================================================
# CHECKPOINT 8.5 — XGBOOST TARGET ENCODING
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.5 — XGBOOST TARGET ENCODING")
print("=" * 70)

xgb_label_map = {
    "acetylcholine": 0,
    "gaba": 1
}

xgb_inverse_label_map = {
    0: "acetylcholine",
    1: "gaba"
}

y_train_xgb = (
    y_train_base
    .map(xgb_label_map)
    .astype(int)
)

y_test_xgb = (
    y_test_base
    .map(xgb_label_map)
    .astype(int)
)

assert set(y_train_xgb.unique()) == {0, 1}
assert set(y_test_xgb.unique()) == {0, 1}

print("\nLabel encoding:")
print("acetylcholine → 0")
print("gaba → 1")

print("\nTraining labels:")
display(
    y_train_xgb.value_counts().sort_index()
)

print("\nTest labels:")
display(
    y_test_xgb.value_counts().sort_index()
)

print("\n✓ XGBoost target encoding complete.")
```

```python
# ================================================================
# CHECKPOINT 8.6 — XGBOOST
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.6 — XGBOOST")
print("=" * 70)

# Calculate class imbalance correction
n_ach_train = (y_train_xgb == 0).sum()
n_gaba_train = (y_train_xgb == 1).sum()

xgb_scale_pos_weight = (
    n_ach_train / n_gaba_train
)

print("\nTraining ACh neurons:")
print(n_ach_train)

print("\nTraining GABA neurons:")
print(n_gaba_train)

print("\nGABA class weighting factor:")
print(f"{xgb_scale_pos_weight:.4f}")

xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_weight=3,
    reg_lambda=1.0,
    reg_alpha=0.0,
    scale_pos_weight=xgb_scale_pos_weight,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1,
    tree_method="hist"
)

xgb_model.fit(
    X_train_base,
    y_train_xgb
)

xgb_predictions_numeric = xgb_model.predict(
    X_test_base
)

xgb_predictions = (
    pd.Series(
        xgb_predictions_numeric,
        index=y_test_base.index
    )
    .map(xgb_inverse_label_map)
)

print("\n✓ XGBoost trained.")

print("\nPredicted class distribution:")
display(
    xgb_predictions.value_counts()
)
```

```python
# ================================================================
# CHECKPOINT 8.7 — XGBOOST METRICS
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.7 — XGBOOST METRICS")
print("=" * 70)

xgb_accuracy = accuracy_score(
    y_test_base,
    xgb_predictions
)

xgb_balanced_accuracy = balanced_accuracy_score(
    y_test_base,
    xgb_predictions
)

xgb_macro_f1 = f1_score(
    y_test_base,
    xgb_predictions,
    average="macro"
)

xgb_class_recalls = recall_score(
    y_test_base,
    xgb_predictions,
    labels=["acetylcholine", "gaba"],
    average=None
)

xgb_ach_recall = xgb_class_recalls[0]
xgb_gaba_recall = xgb_class_recalls[1]

xgb_confusion_matrix = confusion_matrix(
    y_test_base,
    xgb_predictions,
    labels=["acetylcholine", "gaba"]
)

print("\nAccuracy:")
print(f"{xgb_accuracy:.4f}")

print("\nBalanced accuracy:")
print(f"{xgb_balanced_accuracy:.4f}")

print("\nMacro F1:")
print(f"{xgb_macro_f1:.4f}")

print("\nACh recall:")
print(f"{xgb_ach_recall:.4f}")

print("\nGABA recall:")
print(f"{xgb_gaba_recall:.4f}")

print("\nConfusion matrix:")
print(xgb_confusion_matrix)

print("\n✓ XGBoost evaluated.")
```

```python
# ================================================================
# CHECKPOINT 8.8 — XGBOOST CLASSIFICATION REPORT
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.8 — XGBOOST CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test_base,
        xgb_predictions,
        labels=["acetylcholine", "gaba"],
        digits=4
    )
)

print("✓ XGBoost classification report generated.")
```

```python
# ================================================================
# CHECKPOINT 8.9 — MODEL COMPARISON
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.9 — MODEL COMPARISON")
print("=" * 70)

# Extract previously calculated Checkpoint 7 results
dummy_row = baseline_results[
    baseline_results["model"] == "DummyClassifier"
].iloc[0]

logistic_row = baseline_results[
    baseline_results["model"] == "LogisticRegression"
].iloc[0]

checkpoint8_results = pd.DataFrame([
    {
        "model": "DummyClassifier",
        "accuracy": dummy_row["accuracy"],
        "balanced_accuracy": dummy_row["balanced_accuracy"],
        "macro_f1": dummy_row["macro_f1"],
        "ACh_recall": dummy_row["ACh_recall"],
        "GABA_recall": dummy_row["GABA_recall"]
    },
    {
        "model": "LogisticRegression",
        "accuracy": logistic_row["accuracy"],
        "balanced_accuracy": logistic_row["balanced_accuracy"],
        "macro_f1": logistic_row["macro_f1"],
        "ACh_recall": logistic_row["ACh_recall"],
        "GABA_recall": logistic_row["GABA_recall"]
    },
    {
        "model": "RandomForest",
        "accuracy": rf_accuracy,
        "balanced_accuracy": rf_balanced_accuracy,
        "macro_f1": rf_macro_f1,
        "ACh_recall": rf_ach_recall,
        "GABA_recall": rf_gaba_recall
    },
    {
        "model": "XGBoost",
        "accuracy": xgb_accuracy,
        "balanced_accuracy": xgb_balanced_accuracy,
        "macro_f1": xgb_macro_f1,
        "ACh_recall": xgb_ach_recall,
        "GABA_recall": xgb_gaba_recall
    }
])

checkpoint8_results = (
    checkpoint8_results
    .sort_values(
        "balanced_accuracy",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    checkpoint8_results.round(4)
)

print("\n✓ All four models compared.")
```

```python
# ================================================================
# CHECKPOINT 8.10 — IMPROVEMENT OVER DUMMY
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.10 — IMPROVEMENT OVER DUMMY")
print("=" * 70)

dummy_balanced_accuracy = checkpoint8_results.loc[
    checkpoint8_results["model"] == "DummyClassifier",
    "balanced_accuracy"
].iloc[0]

dummy_macro_f1 = checkpoint8_results.loc[
    checkpoint8_results["model"] == "DummyClassifier",
    "macro_f1"
].iloc[0]

checkpoint8_results[
    "balanced_accuracy_gain_vs_dummy"
] = (
    checkpoint8_results["balanced_accuracy"]
    - dummy_balanced_accuracy
)

checkpoint8_results[
    "macro_f1_gain_vs_dummy"
] = (
    checkpoint8_results["macro_f1"]
    - dummy_macro_f1
)

display(
    checkpoint8_results.round(4)
)

print("\n✓ Improvement relative to dummy baseline calculated.")
```

```python
# ================================================================
# CHECKPOINT 8.11 — RANDOM FOREST FEATURE IMPORTANCE
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.11 — RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 70)

rf_feature_importance = pd.DataFrame({
    "feature": ml_feature_columns,
    "importance": rf_model.feature_importances_
})

rf_feature_importance = (
    rf_feature_importance
    .sort_values(
        "importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    rf_feature_importance.round(6)
)

print("\n✓ Random Forest feature importance extracted.")
```

```python
# ================================================================
# CHECKPOINT 8.12 — XGBOOST FEATURE IMPORTANCE
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.12 — XGBOOST FEATURE IMPORTANCE")
print("=" * 70)

xgb_feature_importance = pd.DataFrame({
    "feature": ml_feature_columns,
    "importance": xgb_model.feature_importances_
})

xgb_feature_importance = (
    xgb_feature_importance
    .sort_values(
        "importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    xgb_feature_importance.round(6)
)

print("\n✓ XGBoost feature importance extracted.")
```

```python
# ================================================================
# CHECKPOINT 8.13 — FEATURE IMPORTANCE COMPARISON
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.13 — FEATURE IMPORTANCE COMPARISON")
print("=" * 70)

rf_ranked = (
    rf_feature_importance[
        ["feature", "importance"]
    ]
    .rename(
        columns={
            "importance": "RF_importance"
        }
    )
)

xgb_ranked = (
    xgb_feature_importance[
        ["feature", "importance"]
    ]
    .rename(
        columns={
            "importance": "XGB_importance"
        }
    )
)

feature_importance_comparison = (
    rf_ranked
    .merge(
        xgb_ranked,
        on="feature",
        how="outer"
    )
)

feature_importance_comparison[
    "RF_rank"
] = (
    feature_importance_comparison[
        "RF_importance"
    ]
    .rank(
        ascending=False,
        method="min"
    )
)

feature_importance_comparison[
    "XGB_rank"
] = (
    feature_importance_comparison[
        "XGB_importance"
    ]
    .rank(
        ascending=False,
        method="min"
    )
)

feature_importance_comparison[
    "mean_rank"
] = (
    feature_importance_comparison[
        ["RF_rank", "XGB_rank"]
    ]
    .mean(axis=1)
)

feature_importance_comparison = (
    feature_importance_comparison
    .sort_values(
        "mean_rank"
    )
    .reset_index(drop=True)
)

display(
    feature_importance_comparison.round(6)
)

print("\n✓ Feature rankings compared across nonlinear models.")
```

```python
# ================================================================
# CHECKPOINT 8.14 — NONLINEAR MODEL CONFUSION MATRICES
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.14 — CONFUSION MATRICES")
print("=" * 70)

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5)
)

sns.heatmap(
    rf_confusion_matrix,
    annot=True,
    fmt="d",
    cmap="magma",
    xticklabels=[
        "acetylcholine",
        "gaba"
    ],
    yticklabels=[
        "acetylcholine",
        "gaba"
    ],
    ax=axes[0]
)

axes[0].set_title(
    "Random Forest Confusion Matrix"
)

axes[0].set_xlabel(
    "Predicted"
)

axes[0].set_ylabel(
    "Actual"
)


sns.heatmap(
    xgb_confusion_matrix,
    annot=True,
    fmt="d",
    cmap="magma",
    xticklabels=[
        "acetylcholine",
        "gaba"
    ],
    yticklabels=[
        "acetylcholine",
        "gaba"
    ],
    ax=axes[1]
)

axes[1].set_title(
    "XGBoost Confusion Matrix"
)

axes[1].set_xlabel(
    "Predicted"
)

axes[1].set_ylabel(
    "Actual"
)

plt.tight_layout()
plt.show()

print("\n✓ Nonlinear model confusion matrices generated.")
```

```python
# ================================================================
# CHECKPOINT 8.16 — SAVE RESULTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 8.16 — SAVE RESULTS")
print("=" * 70)

os.makedirs(
    "results/tables",
    exist_ok=True
)

os.makedirs(
    "results/figures",
    exist_ok=True
)

# ------------------------------------------------
# Model comparison
# ------------------------------------------------

checkpoint8_results.to_csv(
    "results/tables/checkpoint8_model_comparison.csv",
    index=False
)

# ------------------------------------------------
# Random Forest feature importance
# ------------------------------------------------

rf_feature_importance.to_csv(
    "results/tables/random_forest_feature_importance.csv",
    index=False
)

# ------------------------------------------------
# XGBoost feature importance
# ------------------------------------------------

xgb_feature_importance.to_csv(
    "results/tables/xgboost_feature_importance.csv",
    index=False
)

# ------------------------------------------------
# Combined feature importance
# ------------------------------------------------

feature_importance_comparison.to_csv(
    "results/tables/nonlinear_feature_importance_comparison.csv",
    index=False
)

# ------------------------------------------------
# Random Forest confusion matrix
# ------------------------------------------------

pd.DataFrame(
    rf_confusion_matrix,
    index=[
        "actual_acetylcholine",
        "actual_gaba"
    ],
    columns=[
        "predicted_acetylcholine",
        "predicted_gaba"
    ]
).to_csv(
    "results/tables/random_forest_confusion_matrix.csv"
)

# ------------------------------------------------
# XGBoost confusion matrix
# ------------------------------------------------

pd.DataFrame(
    xgb_confusion_matrix,
    index=[
        "actual_acetylcholine",
        "actual_gaba"
    ],
    columns=[
        "predicted_acetylcholine",
        "predicted_gaba"
    ]
).to_csv(
    "results/tables/xgboost_confusion_matrix.csv"
)

print("\n✓ Model comparison saved.")
print("✓ Random Forest feature importance saved.")
print("✓ XGBoost feature importance saved.")
print("✓ Combined feature importance saved.")
print("✓ Random Forest confusion matrix saved.")
print("✓ XGBoost confusion matrix saved.")

print("\n✓ CHECKPOINT 8 RESULTS SAVED.")
```

```python
# ================================================================
# CHECKPOINT 8 — FINAL AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 8 — FINAL AUDIT")
print("=" * 70)

# ================================================================
# DATA AUDIT
# ================================================================

assert len(final_ml_dataset) == 165

assert len(ml_feature_columns) == 16

assert X_train_base.shape == (123, 16)

assert X_test_base.shape == (42, 16)

assert len(y_train_base) == 123

assert len(y_test_base) == 42


# ================================================================
# MODEL AUDIT
# ================================================================

assert hasattr(
    rf_model,
    "predict"
)

assert hasattr(
    xgb_model,
    "predict"
)

assert len(rf_predictions) == 42

assert len(xgb_predictions) == 42


# ================================================================
# METRIC AUDIT
# ================================================================

for metric_value in [
    rf_accuracy,
    rf_balanced_accuracy,
    rf_macro_f1,
    rf_ach_recall,
    rf_gaba_recall,
    xgb_accuracy,
    xgb_balanced_accuracy,
    xgb_macro_f1,
    xgb_ach_recall,
    xgb_gaba_recall
]:

    assert 0 <= metric_value <= 1


# ================================================================
# CLASS AUDIT
# ================================================================

assert set(
    rf_predictions
).issubset(
    {
        "acetylcholine",
        "gaba"
    }
)

assert set(
    xgb_predictions
).issubset(
    {
        "acetylcholine",
        "gaba"
    }
)


# ================================================================
# FEATURE IMPORTANCE AUDIT
# ================================================================

assert len(
    rf_feature_importance
) == 16

assert len(
    xgb_feature_importance
) == 16

assert set(
    rf_feature_importance["feature"]
) == set(
    ml_feature_columns
)

assert set(
    xgb_feature_importance["feature"]
) == set(
    ml_feature_columns
)


# ================================================================
# RESULTS TABLE AUDIT
# ================================================================

assert len(
    checkpoint8_results
) == 4

assert set(
    checkpoint8_results["model"]
) == {
    "DummyClassifier",
    "LogisticRegression",
    "RandomForest",
    "XGBoost"
}


# ================================================================
# FINAL SUMMARY
# ================================================================

print("\nDATA")
print("----")

print(
    "Total neurons:",
    len(final_ml_dataset)
)

print(
    "Connectivity features:",
    len(ml_feature_columns)
)

print(
    "Training observations:",
    len(X_train_base)
)

print(
    "Test observations:",
    len(X_test_base)
)


print("\nMODELS")
print("------")

print("DummyClassifier")
print("LogisticRegression")
print("RandomForest")
print("XGBoost")


print("\nMODEL PERFORMANCE")
print("-----------------")

display(
    checkpoint8_results[
        [
            "model",
            "accuracy",
            "balanced_accuracy",
            "macro_f1",
            "ACh_recall",
            "GABA_recall"
        ]
    ].round(4)
)


print("\n✓ All model predictions valid.")
print("✓ All metrics valid.")
print("✓ All 16 features represented.")
print("✓ Same train/test split used for all models.")
print("✓ No test-set refitting performed.")
print("✓ Random Forest class balancing applied.")
print("✓ XGBoost class imbalance correction applied.")

print("\n" + "=" * 70)
print("✓ CHECKPOINT 8 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 9
# CELL-TYPE-GROUPED CROSS-VALIDATION
# ================================================================

print("=" * 70)
print("CHECKPOINT 9 — CELL-TYPE-GROUPED CROSS-VALIDATION")
print("=" * 70)

# Use the final ML dataset already constructed
X_grouped = final_ml_dataset[
    ml_feature_columns
].copy()

y_grouped = final_ml_dataset[
    "neurotransmitter"
].copy()

groups_grouped = final_ml_dataset[
    "cell_type"
].copy()

# ------------------------------------------------
# Basic validation
# ------------------------------------------------

assert X_grouped.shape == (165, 16)
assert len(y_grouped) == 165
assert len(groups_grouped) == 165

assert X_grouped.isna().sum().sum() == 0
assert y_grouped.isna().sum() == 0
assert groups_grouped.isna().sum() == 0

assert groups_grouped.nunique() == 65

assert set(
    y_grouped.unique()
) == {
    "acetylcholine",
    "gaba"
}

print("\nDataset shape:")
print(X_grouped.shape)

print("\nNumber of neurons:")
print(len(X_grouped))

print("\nNumber of canonical cell-type groups:")
print(groups_grouped.nunique())

print("\nNumber of connectivity features:")
print(len(ml_feature_columns))

print("\nNeurotransmitter distribution:")
display(
    y_grouped.value_counts()
)

print("\nCell-type group distribution:")
display(
    groups_grouped.value_counts().describe()
)

print("\n✓ 165 neurons confirmed.")
print("✓ 65 canonical cell-type groups confirmed.")
print("✓ 16 connectivity features confirmed.")
print("✓ No missing values.")
print("✓ Both neurotransmitter classes present.")
```

```python
# ================================================================
# CHECKPOINT 9.1 — GROUPED CROSS-VALIDATION TOOLS
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.1 — GROUPED CROSS-VALIDATION TOOLS")
print("=" * 70)

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression

# These should already exist from Checkpoint 8,
# but explicitly confirm availability.
assert RandomForestClassifier is not None
assert XGBClassifier is not None

print("\n✓ StratifiedGroupKFold available.")
print("✓ Pipeline available.")
print("✓ StandardScaler available.")
print("✓ DummyClassifier available.")
print("✓ LogisticRegression available.")
print("✓ RandomForestClassifier available.")
print("✓ XGBClassifier available.")
```

```python
# ================================================================
# CHECKPOINT 9.2 — STRATIFIED GROUPED SPLITTER
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.2 — STRATIFIED GROUPED SPLITTER")
print("=" * 70)

N_SPLITS = 5
CV_RANDOM_STATE = 42

group_cv = StratifiedGroupKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=CV_RANDOM_STATE
)

print("\nNumber of folds:")
print(N_SPLITS)

print("\nGrouping variable:")
print("canonical cell_type")

print("\nStratification target:")
print("neurotransmitter")

print("\nRandom state:")
print(CV_RANDOM_STATE)

print("\n✓ StratifiedGroupKFold configured.")
```

```python
# ================================================================
# CHECKPOINT 9.3 — FOLD INTEGRITY AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.3 — FOLD INTEGRITY AUDIT")
print("=" * 70)

fold_assignments = np.full(
    len(X_grouped),
    -1,
    dtype=int
)

fold_audit_records = []

for fold_number, (train_idx, val_idx) in enumerate(
    group_cv.split(
        X_grouped,
        y_grouped,
        groups_grouped
    ),
    start=1
):

    train_groups = set(
        groups_grouped.iloc[train_idx]
    )

    val_groups = set(
        groups_grouped.iloc[val_idx]
    )

    overlap = train_groups.intersection(
        val_groups
    )

    # Assign fold number to validation observations
    fold_assignments[val_idx] = fold_number

    train_ach = (
        y_grouped.iloc[train_idx]
        == "acetylcholine"
    ).sum()

    train_gaba = (
        y_grouped.iloc[train_idx]
        == "gaba"
    ).sum()

    val_ach = (
        y_grouped.iloc[val_idx]
        == "acetylcholine"
    ).sum()

    val_gaba = (
        y_grouped.iloc[val_idx]
        == "gaba"
    ).sum()

    fold_audit_records.append({
        "fold": fold_number,
        "train_neurons": len(train_idx),
        "validation_neurons": len(val_idx),
        "train_cell_types": len(train_groups),
        "validation_cell_types": len(val_groups),
        "train_ACh": train_ach,
        "train_GABA": train_gaba,
        "validation_ACh": val_ach,
        "validation_GABA": val_gaba,
        "cell_type_overlap": len(overlap)
    })

fold_audit = pd.DataFrame(
    fold_audit_records
)

display(
    fold_audit
)

# ------------------------------------------------
# Integrity checks
# ------------------------------------------------

assert (fold_assignments >= 1).all()
assert (fold_assignments <= N_SPLITS).all()

# Every observation must be assigned to exactly one
# validation fold.
assert np.bincount(
    fold_assignments
)[1:].sum() == 165

# Every fold must contain both classes.
assert (
    fold_audit["validation_ACh"] > 0
).all()

assert (
    fold_audit["validation_GABA"] > 0
).all()

# No canonical cell type may occur in both train
# and validation within the same fold.
assert (
    fold_audit["cell_type_overlap"] == 0
).all()

print("\n✓ Every neuron assigned to exactly one validation fold.")
print("✓ All five validation folds contain both classes.")
print("✓ No canonical cell-type overlap between train and validation.")
print("\n✓ FOLD INTEGRITY AUDIT PASSED.")
```

```python
# ================================================================
# CHECKPOINT 9.4 — MODEL DEFINITIONS
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.4 — MODEL DEFINITIONS")
print("=" * 70)

def make_logistic_model():

    return Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "logistic",
            LogisticRegression(
                max_iter=2000,
                class_weight="balanced",
                random_state=42
            )
        )
    ])


def make_random_forest_model():

    return RandomForestClassifier(
        n_estimators=500,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )


def make_xgboost_model(
    y_train_encoded
):

    n_ach = (
        y_train_encoded == 0
    ).sum()

    n_gaba = (
        y_train_encoded == 1
    ).sum()

    scale_pos_weight = (
        n_ach / n_gaba
    )

    return XGBClassifier(
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        min_child_weight=3,
        reg_lambda=1.0,
        reg_alpha=0.0,
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1,
        tree_method="hist"
    )


print("\n✓ Logistic Regression factory defined.")
print("✓ Random Forest factory defined.")
print("✓ XGBoost factory defined.")
print("✓ Models will be freshly instantiated for every fold.")
```

```python
# ================================================================
# CHECKPOINT 9.5 — XGBOOST TARGET ENCODING
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.5 — XGBOOST TARGET ENCODING")
print("=" * 70)

xgb_grouped_label_map = {
    "acetylcholine": 0,
    "gaba": 1
}

xgb_grouped_inverse_map = {
    0: "acetylcholine",
    1: "gaba"
}

y_grouped_xgb = (
    y_grouped
    .map(xgb_grouped_label_map)
    .astype(int)
)

assert set(
    y_grouped_xgb.unique()
) == {0, 1}

print("\nEncoding:")
print("acetylcholine → 0")
print("gaba → 1")

print("\nEncoded target distribution:")
display(
    y_grouped_xgb.value_counts()
    .sort_index()
)

print("\n✓ XGBoost encoding complete.")
```

```python
# ================================================================
# CHECKPOINT 9.6 — RUN CELL-TYPE-GROUPED CROSS-VALIDATION
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.6 — RUN CELL-TYPE-GROUPED CROSS-VALIDATION")
print("=" * 70)

grouped_cv_records = []

# Recreate the splitter so iteration is identical
group_cv = StratifiedGroupKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=CV_RANDOM_STATE
)

for fold_number, (train_idx, val_idx) in enumerate(
    group_cv.split(
        X_grouped,
        y_grouped,
        groups_grouped
    ),
    start=1
):

    print("\n" + "-" * 70)
    print(f"FOLD {fold_number}")
    print("-" * 70)

    X_train_fold = X_grouped.iloc[
        train_idx
    ].copy()

    X_val_fold = X_grouped.iloc[
        val_idx
    ].copy()

    y_train_fold = y_grouped.iloc[
        train_idx
    ].copy()

    y_val_fold = y_grouped.iloc[
        val_idx
    ].copy()

    # ------------------------------------------------
    # Fold information
    # ------------------------------------------------

    print(
        "Training neurons:",
        len(train_idx)
    )

    print(
        "Validation neurons:",
        len(val_idx)
    )

    print(
        "Training cell types:",
        groups_grouped.iloc[
            train_idx
        ].nunique()
    )

    print(
        "Validation cell types:",
        groups_grouped.iloc[
            val_idx
        ].nunique()
    )

    print("\nValidation class distribution:")
    print(
        y_val_fold.value_counts().to_dict()
    )

    # ------------------------------------------------
    # 1. Dummy
    # ------------------------------------------------

    dummy_fold = DummyClassifier(
        strategy="most_frequent"
    )

    dummy_fold.fit(
        X_train_fold,
        y_train_fold
    )

    dummy_pred = dummy_fold.predict(
        X_val_fold
    )

    # ------------------------------------------------
    # 2. Logistic Regression
    # ------------------------------------------------

    logistic_fold = make_logistic_model()

    logistic_fold.fit(
        X_train_fold,
        y_train_fold
    )

    logistic_pred = logistic_fold.predict(
        X_val_fold
    )

    # ------------------------------------------------
    # 3. Random Forest
    # ------------------------------------------------

    rf_fold = make_random_forest_model()

    rf_fold.fit(
        X_train_fold,
        y_train_fold
    )

    rf_pred = rf_fold.predict(
        X_val_fold
    )

    # ------------------------------------------------
    # 4. XGBoost
    # ------------------------------------------------

    y_train_fold_xgb = (
        y_train_fold
        .map(xgb_grouped_label_map)
        .astype(int)
    )

    xgb_fold = make_xgboost_model(
        y_train_fold_xgb
    )

    xgb_fold.fit(
        X_train_fold,
        y_train_fold_xgb
    )

    xgb_pred_numeric = xgb_fold.predict(
        X_val_fold
    )

    xgb_pred = (
        pd.Series(
            xgb_pred_numeric,
            index=y_val_fold.index
        )
        .map(
            xgb_grouped_inverse_map
        )
    )

    # ------------------------------------------------
    # Store predictions
    # ------------------------------------------------

    model_predictions = {
        "DummyClassifier": dummy_pred,
        "LogisticRegression": logistic_pred,
        "RandomForest": rf_pred,
        "XGBoost": xgb_pred
    }

    # ------------------------------------------------
    # Calculate metrics
    # ------------------------------------------------

    for model_name, predictions in model_predictions.items():

        fold_accuracy = accuracy_score(
            y_val_fold,
            predictions
        )

        fold_balanced_accuracy = (
            balanced_accuracy_score(
                y_val_fold,
                predictions
            )
        )

        fold_macro_f1 = f1_score(
            y_val_fold,
            predictions,
            average="macro"
        )

        fold_recalls = recall_score(
            y_val_fold,
            predictions,
            labels=[
                "acetylcholine",
                "gaba"
            ],
            average=None,
            zero_division=0
        )

        grouped_cv_records.append({
            "fold": fold_number,
            "model": model_name,
            "accuracy": fold_accuracy,
            "balanced_accuracy": fold_balanced_accuracy,
            "macro_f1": fold_macro_f1,
            "ACh_recall": fold_recalls[0],
            "GABA_recall": fold_recalls[1],
            "n_validation_neurons": len(val_idx),
            "n_validation_cell_types": (
                groups_grouped.iloc[
                    val_idx
                ].nunique()
            )
        })

    print(
        f"✓ Fold {fold_number} complete."
    )

grouped_cv_results = pd.DataFrame(
    grouped_cv_records
)

print("\n" + "=" * 70)
print("✓ ALL GROUPED CV FOLDS COMPLETE")
print("=" * 70)

display(
    grouped_cv_results.round(4)
)
```

```python
# ================================================================
# CHECKPOINT 9.7 — FOLD-LEVEL RESULTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.7 — FOLD-LEVEL RESULTS")
print("=" * 70)

display(
    grouped_cv_results[
        [
            "fold",
            "model",
            "accuracy",
            "balanced_accuracy",
            "macro_f1",
            "ACh_recall",
            "GABA_recall",
            "n_validation_neurons",
            "n_validation_cell_types"
        ]
    ].round(4)
)

# Basic checks
assert len(grouped_cv_results) == 20
assert grouped_cv_results["fold"].nunique() == 5
assert grouped_cv_results["model"].nunique() == 4

assert set(
    grouped_cv_results["model"]
) == {
    "DummyClassifier",
    "LogisticRegression",
    "RandomForest",
    "XGBoost"
}

assert (
    grouped_cv_results[
        "n_validation_neurons"
    ].sum()
    == 165 * 4
)

print("\n✓ 5 folds × 4 models = 20 evaluations.")
print("✓ Fold-level results validated.")
```

```python
# ================================================================
# CHECKPOINT 9.8 — GROUPED CV PERFORMANCE SUMMARY
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.8 — GROUPED CV PERFORMANCE SUMMARY")
print("=" * 70)

grouped_cv_summary = (
    grouped_cv_results
    .groupby("model")
    .agg(
        accuracy_mean=(
            "accuracy",
            "mean"
        ),
        accuracy_sd=(
            "accuracy",
            "std"
        ),
        balanced_accuracy_mean=(
            "balanced_accuracy",
            "mean"
        ),
        balanced_accuracy_sd=(
            "balanced_accuracy",
            "std"
        ),
        macro_f1_mean=(
            "macro_f1",
            "mean"
        ),
        macro_f1_sd=(
            "macro_f1",
            "std"
        ),
        ACh_recall_mean=(
            "ACh_recall",
            "mean"
        ),
        ACh_recall_sd=(
            "ACh_recall",
            "std"
        ),
        GABA_recall_mean=(
            "GABA_recall",
            "mean"
        ),
        GABA_recall_sd=(
            "GABA_recall",
            "std"
        )
    )
    .reset_index()
)

# Rank by mean balanced accuracy
grouped_cv_summary = (
    grouped_cv_summary
    .sort_values(
        "balanced_accuracy_mean",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    grouped_cv_summary.round(4)
)

print("\n✓ Mean and standard deviation calculated across five grouped folds.")
```

```python
# ================================================================
# CHECKPOINT 9.9 — PUBLICATION-STYLE PERFORMANCE TABLE
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.9 — PUBLICATION-STYLE PERFORMANCE TABLE")
print("=" * 70)

grouped_cv_publication_table = pd.DataFrame({
    "model": grouped_cv_summary["model"],

    "accuracy": (
        grouped_cv_summary["accuracy_mean"]
        .map(lambda x: f"{x:.3f}")
        + " ± "
        + grouped_cv_summary["accuracy_sd"]
        .map(lambda x: f"{x:.3f}")
    ),

    "balanced_accuracy": (
        grouped_cv_summary[
            "balanced_accuracy_mean"
        ]
        .map(lambda x: f"{x:.3f}")
        + " ± "
        + grouped_cv_summary[
            "balanced_accuracy_sd"
        ]
        .map(lambda x: f"{x:.3f}")
    ),

    "macro_F1": (
        grouped_cv_summary["macro_f1_mean"]
        .map(lambda x: f"{x:.3f}")
        + " ± "
        + grouped_cv_summary["macro_f1_sd"]
        .map(lambda x: f"{x:.3f}")
    ),

    "ACh_recall": (
        grouped_cv_summary[
            "ACh_recall_mean"
        ]
        .map(lambda x: f"{x:.3f}")
        + " ± "
        + grouped_cv_summary[
            "ACh_recall_sd"
        ]
        .map(lambda x: f"{x:.3f}")
    ),

    "GABA_recall": (
        grouped_cv_summary[
            "GABA_recall_mean"
        ]
        .map(lambda x: f"{x:.3f}")
        + " ± "
        + grouped_cv_summary[
            "GABA_recall_sd"
        ]
        .map(lambda x: f"{x:.3f}")
    )
})

display(
    grouped_cv_publication_table
)

print("\n✓ Publication-style grouped CV table generated.")
```

```python
# ================================================================
# CHECKPOINT 9.10 — RANDOM SPLIT VS GROUPED CV
# ================================================================

print("=" * 70)
print("CHECKPOINT 9.10 — RANDOM SPLIT VS GROUPED CV")
print("=" * 70)

# ------------------------------------------------
# Extract Checkpoint 8 random-split performance
# ------------------------------------------------

random_split_comparison = checkpoint8_results[
    [
        "model",
        "accuracy",
        "balanced_accuracy",
        "macro_f1"
    ]
].copy()

random_split_comparison = (
    random_split_comparison
    .rename(
        columns={
            "accuracy": "random_split_accuracy",
            "balanced_accuracy":
                "random_split_balanced_accuracy",
            "macro_f1":
                "random_split_macro_f1"
        }
    )
)

# ------------------------------------------------
# Extract grouped CV mean performance
# ------------------------------------------------

grouped_mean_comparison = grouped_cv_summary[
    [
        "model",
        "accuracy_mean",
        "balanced_accuracy_mean",
        "macro_f1_mean"
    ]
].copy()

grouped_mean_comparison = (
    grouped_mean_comparison
    .rename(
        columns={
            "accuracy_mean":
                "grouped_cv_accuracy",
            "balanced_accuracy_mean":
                "grouped_cv_balanced_accuracy",
            "macro_f1_mean":
                "grouped_cv_macro_f1"
        }
    )
)

# ------------------------------------------------
# Merge
# ------------------------------------------------

random_vs_grouped = (
    random_split_comparison
    .merge(
        grouped_mean_comparison,
        on="model",
        how="inner"
    )
)

# ------------------------------------------------
# Calculate differences
# ------------------------------------------------

random_vs_grouped[
    "balanced_accuracy_change"
] = (
    random_vs_grouped[
        "grouped_cv_balanced_accuracy"
    ]
    -
    random_vs_grouped[
        "random_split_balanced_accuracy"
    ]
)

random_vs_grouped[
    "macro_f1_change"
] = (
    random_vs_grouped[
        "grouped_cv_macro_f1"
    ]
    -
    random_vs_grouped[
        "random_split_macro_f1"
    ]
)

display(
    random_vs_grouped.round(4)
)

print(
    "\n✓ Random-split and grouped-CV performance compared."
)
```

```python
# ================================================================
# CHECKPOINT 10.1 — FIXED GROUPED FOLD STRUCTURE
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.1 — FIXED GROUPED FOLD STRUCTURE")
print("=" * 70)

CP10_N_SPLITS = 5
CP10_RANDOM_STATE = 42

cp10_cv = StratifiedGroupKFold(
    n_splits=CP10_N_SPLITS,
    shuffle=True,
    random_state=CP10_RANDOM_STATE
)

cp10_splits = list(
    cp10_cv.split(
        X_cp10,
        y_cp10,
        groups_cp10
    )
)

assert len(cp10_splits) == 5

# ------------------------------------------------
# Verify group separation
# ------------------------------------------------

for fold_number, (train_idx, val_idx) in enumerate(
    cp10_splits,
    start=1
):

    train_groups = set(
        groups_cp10.iloc[train_idx]
    )

    val_groups = set(
        groups_cp10.iloc[val_idx]
    )

    assert len(
        train_groups.intersection(val_groups)
    ) == 0

print("\n✓ Five fixed grouped folds generated.")
print("✓ No canonical cell type appears in both train and validation.")
print("✓ These folds will be reused for sensitivity analyses.")
```

```python
# ================================================================
# CHECKPOINT 10.2 — DATA LEAKAGE AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.2 — DATA LEAKAGE AUDIT")
print("=" * 70)

# ------------------------------------------------
# Forbidden identifiers / labels
# ------------------------------------------------

forbidden_columns = {
    "bodyId",
    "hemibrain_type",
    "cell_type",
    "neurotransmitter"
}

feature_overlap = (
    set(ml_feature_columns)
    .intersection(forbidden_columns)
)

assert len(feature_overlap) == 0

# ------------------------------------------------
# Confirm feature names are the intended features
# ------------------------------------------------

assert set(
    ml_feature_columns
) == set(
    X_cp10.columns
)

# ------------------------------------------------
# Confirm no target-derived columns
# ------------------------------------------------

for feature in ml_feature_columns:

    assert feature != "neurotransmitter"
    assert feature != "cell_type"
    assert feature != "bodyId"
    assert feature != "hemibrain_type"

# ------------------------------------------------
# Confirm group integrity in every fold
# ------------------------------------------------

for train_idx, val_idx in cp10_splits:

    train_groups = set(
        groups_cp10.iloc[train_idx]
    )

    val_groups = set(
        groups_cp10.iloc[val_idx]
    )

    assert (
        train_groups.isdisjoint(
            val_groups
        )
    )

# ------------------------------------------------
# Confirm all features numeric
# ------------------------------------------------

assert all(
    pd.api.types.is_numeric_dtype(
        X_cp10[column]
    )
    for column in X_cp10.columns
)

print("\nForbidden feature overlap:")
print(feature_overlap)

print("\n✓ No body IDs used as features.")
print("✓ No cell-type labels used as features.")
print("✓ No neurotransmitter labels used as features.")
print("✓ No molecular metadata used as features.")
print("✓ All features numeric.")
print("✓ Group separation verified in every fold.")

print("\n✓ DATA LEAKAGE AUDIT PASSED.")
```

```python
# ================================================================
# CHECKPOINT 10.3 — ROBUST GROUPED-CV EVALUATOR
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.3 — ROBUST GROUPED-CV EVALUATOR")
print("=" * 70)


def evaluate_logistic_grouped_cv(
    X,
    y,
    groups,
    splits
):

    fold_records = []

    for fold_number, (train_idx, val_idx) in enumerate(
        splits,
        start=1
    ):

        X_train = X.iloc[
            train_idx
        ].copy()

        X_val = X.iloc[
            val_idx
        ].copy()

        y_train = y.iloc[
            train_idx
        ].copy()

        y_val = y.iloc[
            val_idx
        ].copy()

        # ------------------------------------------------
        # Scaler + Logistic Regression
        # Both are fitted using training data only.
        # ------------------------------------------------

        model = Pipeline([
            (
                "scaler",
                StandardScaler()
            ),
            (
                "logistic",
                LogisticRegression(
                    max_iter=2000,
                    class_weight="balanced",
                    random_state=42
                )
            )
        ])

        model.fit(
            X_train,
            y_train
        )

        predictions = model.predict(
            X_val
        )

        # ------------------------------------------------
        # Metrics
        # ------------------------------------------------

        accuracy = accuracy_score(
            y_val,
            predictions
        )

        balanced_accuracy = (
            balanced_accuracy_score(
                y_val,
                predictions
            )
        )

        macro_f1 = f1_score(
            y_val,
            predictions,
            average="macro"
        )

        recalls = recall_score(
            y_val,
            predictions,
            labels=[
                "acetylcholine",
                "gaba"
            ],
            average=None,
            zero_division=0
        )

        fold_records.append({
            "fold": fold_number,
            "accuracy": accuracy,
            "balanced_accuracy":
                balanced_accuracy,
            "macro_f1": macro_f1,
            "ACh_recall": recalls[0],
            "GABA_recall": recalls[1],
            "n_validation_neurons":
                len(val_idx),
            "n_validation_cell_types":
                groups.iloc[val_idx].nunique()
        })

    fold_df = pd.DataFrame(
        fold_records
    )

    summary = {
        "accuracy_mean":
            fold_df["accuracy"].mean(),

        "accuracy_sd":
            fold_df["accuracy"].std(),

        "balanced_accuracy_mean":
            fold_df["balanced_accuracy"].mean(),

        "balanced_accuracy_sd":
            fold_df["balanced_accuracy"].std(),

        "macro_f1_mean":
            fold_df["macro_f1"].mean(),

        "macro_f1_sd":
            fold_df["macro_f1"].std(),

        "ACh_recall_mean":
            fold_df["ACh_recall"].mean(),

        "ACh_recall_sd":
            fold_df["ACh_recall"].std(),

        "GABA_recall_mean":
            fold_df["GABA_recall"].mean(),

        "GABA_recall_sd":
            fold_df["GABA_recall"].std()
    }

    return fold_df, summary


print("\n✓ Reusable grouped Logistic Regression evaluator created.")
print("✓ Standardization occurs independently inside every fold.")
```

```python
# ================================================================
# CHECKPOINT 10.4 — REFERENCE PERFORMANCE
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.4 — REFERENCE PERFORMANCE")
print("=" * 70)

reference_fold_results, reference_summary = (
    evaluate_logistic_grouped_cv(
        X_cp10,
        y_cp10,
        groups_cp10,
        cp10_splits
    )
)

reference_balanced_accuracy = (
    reference_summary[
        "balanced_accuracy_mean"
    ]
)

reference_macro_f1 = (
    reference_summary[
        "macro_f1_mean"
    ]
)

print("\nReference Logistic Regression:")
print(
    f"Balanced accuracy: "
    f"{reference_balanced_accuracy:.4f} "
    f"± "
    f"{reference_summary['balanced_accuracy_sd']:.4f}"
)

print(
    f"Macro F1: "
    f"{reference_macro_f1:.4f} "
    f"± "
    f"{reference_summary['macro_f1_sd']:.4f}"
)

display(
    reference_fold_results.round(4)
)

assert len(
    reference_fold_results
) == 5

assert 0 <= reference_balanced_accuracy <= 1

print("\n✓ Reference grouped-CV performance established.")
```

```python
# ================================================================
# CHECKPOINT 10.5 — FEATURE-FAMILY DEFINITIONS
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.5 — FEATURE-FAMILY DEFINITIONS")
print("=" * 70)

outgoing_features = [
    "outgoing_edges",
    "outgoing_synapses",
    "outgoing_mean_synapses",
    "outgoing_median_synapses",
    "outgoing_max_synapses",
    "outgoing_synapse_sd"
]

incoming_features = [
    "incoming_edges",
    "incoming_synapses",
    "incoming_mean_synapses",
    "incoming_median_synapses",
    "incoming_max_synapses",
    "incoming_synapse_sd"
]

overall_features = [
    "total_edges",
    "total_synapses"
]

directionality_features = [
    "outgoing_incoming_edge_ratio",
    "outgoing_incoming_synapse_ratio"
]

feature_family_sets = {
    "All_features": ml_feature_columns,

    "Outgoing_only": outgoing_features,

    "Incoming_only": incoming_features,

    "Overall_only": overall_features,

    "Directionality_only": directionality_features,

    "Outgoing_plus_incoming": (
        outgoing_features
        + incoming_features
    ),

    "Counts_only": [
        "outgoing_edges",
        "outgoing_synapses",
        "incoming_edges",
        "incoming_synapses",
        "total_edges",
        "total_synapses"
    ]
}

# ------------------------------------------------
# Audit
# ------------------------------------------------

for family_name, features in feature_family_sets.items():

    assert len(features) > 0

    assert set(features).issubset(
        set(ml_feature_columns)
    )

print("\nFeature families:")

for family_name, features in feature_family_sets.items():

    print(
        f"  {family_name}: "
        f"{len(features)} features"
    )

print("\n✓ Feature-family definitions validated.")
```

```python
# ================================================================
# CHECKPOINT 10.7 — LOG-TRANSFORMATION SENSITIVITY
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.7 — LOG-TRANSFORMATION SENSITIVITY")
print("=" * 70)

log_transform_features = [
    "outgoing_edges",
    "outgoing_synapses",
    "outgoing_mean_synapses",
    "outgoing_median_synapses",
    "outgoing_max_synapses",
    "outgoing_synapse_sd",

    "incoming_edges",
    "incoming_synapses",
    "incoming_mean_synapses",
    "incoming_median_synapses",
    "incoming_max_synapses",
    "incoming_synapse_sd",

    "total_edges",
    "total_synapses"
]

X_log = X_cp10.copy()

for feature in log_transform_features:

    # All these features should be nonnegative.
    assert (
        X_log[feature] >= 0
    ).all()

    X_log[feature] = np.log1p(
        X_log[feature]
    )

# Directionality ratios intentionally remain
# on their original scale.

assert X_log.shape == X_cp10.shape

log_fold_results, log_summary = (
    evaluate_logistic_grouped_cv(
        X_log,
        y_cp10,
        groups_cp10,
        cp10_splits
    )
)

log_sensitivity_result = pd.DataFrame([
    {
        "analysis":
            "Raw features",

        "balanced_accuracy_mean":
            reference_summary[
                "balanced_accuracy_mean"
            ],

        "balanced_accuracy_sd":
            reference_summary[
                "balanced_accuracy_sd"
            ],

        "macro_f1_mean":
            reference_summary[
                "macro_f1_mean"
            ],

        "macro_f1_sd":
            reference_summary[
                "macro_f1_sd"
            ]
    },

    {
        "analysis":
            "Log1p transformed connectivity",

        "balanced_accuracy_mean":
            log_summary[
                "balanced_accuracy_mean"
            ],

        "balanced_accuracy_sd":
            log_summary[
                "balanced_accuracy_sd"
            ],

        "macro_f1_mean":
            log_summary[
                "macro_f1_mean"
            ],

        "macro_f1_sd":
            log_summary[
                "macro_f1_sd"
            ]
    }
])

log_sensitivity_result[
    "balanced_accuracy_change_vs_raw"
] = (
    log_sensitivity_result[
        "balanced_accuracy_mean"
    ]
    - reference_balanced_accuracy
)

display(
    log_sensitivity_result.round(4)
)

print("\n✓ Log-transformation sensitivity analysis complete.")
```

```python
# ================================================================
# CHECKPOINT 10.8 — CONNECTIVITY THRESHOLD SENSITIVITY
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.8 — CONNECTIVITY THRESHOLD SENSITIVITY")
print("=" * 70)


def build_features_from_edges(
    edges,
    target_body_ids
):

    target_body_ids = set(
        target_body_ids
    )

    work = edges.copy()

    # ------------------------------------------------
    # Ensure numeric synapse counts
    # ------------------------------------------------

    work["synapse_count"] = pd.to_numeric(
        work["synapse_count"],
        errors="coerce"
    )

    assert work[
        "synapse_count"
    ].notna().all()

    # ------------------------------------------------
    # Target flags
    # ------------------------------------------------

    work["pre_is_target"] = (
        work["pre_bodyId"]
        .isin(target_body_ids)
    )

    work["post_is_target"] = (
        work["post_bodyId"]
        .isin(target_body_ids)
    )

    # ------------------------------------------------
    # Base target table
    # ------------------------------------------------

    target_table = pd.DataFrame({
        "bodyId":
            sorted(target_body_ids)
    })

    # ------------------------------------------------
    # OUTGOING
    # ------------------------------------------------

    outgoing = work[
        work["pre_is_target"]
    ].copy()

    outgoing_summary = (
        outgoing
        .groupby("pre_bodyId")
        ["synapse_count"]
        .agg(
            outgoing_edges="size",
            outgoing_synapses="sum",
            outgoing_mean_synapses="mean",
            outgoing_median_synapses="median",
            outgoing_max_synapses="max",
            outgoing_synapse_sd="std"
        )
        .reset_index()
        .rename(
            columns={
                "pre_bodyId": "bodyId"
            }
        )
    )

    # ------------------------------------------------
    # INCOMING
    # ------------------------------------------------

    incoming = work[
        work["post_is_target"]
    ].copy()

    incoming_summary = (
        incoming
        .groupby("post_bodyId")
        ["synapse_count"]
        .agg(
            incoming_edges="size",
            incoming_synapses="sum",
            incoming_mean_synapses="mean",
            incoming_median_synapses="median",
            incoming_max_synapses="max",
            incoming_synapse_sd="std"
        )
        .reset_index()
        .rename(
            columns={
                "post_bodyId": "bodyId"
            }
        )
    )

    # ------------------------------------------------
    # Merge
    # ------------------------------------------------

    features = (
        target_table
        .merge(
            outgoing_summary,
            on="bodyId",
            how="left"
        )
        .merge(
            incoming_summary,
            on="bodyId",
            how="left"
        )
    )

    # ------------------------------------------------
    # Fill missing values with zero.
    #
    # SD is undefined for a single edge, so it is also
    # set to zero in this feature representation.
    # ------------------------------------------------

    numeric_features = [
        "outgoing_edges",
        "outgoing_synapses",
        "outgoing_mean_synapses",
        "outgoing_median_synapses",
        "outgoing_max_synapses",
        "outgoing_synapse_sd",

        "incoming_edges",
        "incoming_synapses",
        "incoming_mean_synapses",
        "incoming_median_synapses",
        "incoming_max_synapses",
        "incoming_synapse_sd"
    ]

    features[
        numeric_features
    ] = features[
        numeric_features
    ].fillna(0)

    # ------------------------------------------------
    # Overall connectivity
    # ------------------------------------------------

    features[
        "total_edges"
    ] = (
        features["outgoing_edges"]
        + features["incoming_edges"]
    )

    features[
        "total_synapses"
    ] = (
        features["outgoing_synapses"]
        + features["incoming_synapses"]
    )

    # ------------------------------------------------
    # Directionality ratios
    # ------------------------------------------------

    features[
        "outgoing_incoming_edge_ratio"
    ] = np.divide(
        features["outgoing_edges"],
        features["incoming_edges"],
        out=np.zeros(
            len(features),
            dtype=float
        ),
        where=(
            features[
                "incoming_edges"
            ].to_numpy() != 0
        )
    )

    features[
        "outgoing_incoming_synapse_ratio"
    ] = np.divide(
        features["outgoing_synapses"],
        features["incoming_synapses"],
        out=np.zeros(
            len(features),
            dtype=float
        ),
        where=(
            features[
                "incoming_synapses"
            ].to_numpy() != 0
        )
    )

    return features[
        [
            "bodyId"
        ]
        + ml_feature_columns
    ]


# ------------------------------------------------
# Target IDs
# ------------------------------------------------

target_body_ids_cp10 = (
    final_ml_dataset[
        "bodyId"
    ]
    .astype(int)
    .tolist()
)

assert len(
    set(target_body_ids_cp10)
) == 165


# ------------------------------------------------
# Threshold analysis
# ------------------------------------------------

threshold_values = [
    1,
    2,
    3,
    5
]

threshold_records = []

for threshold in threshold_values:

    print(
        f"\nEvaluating synapse threshold >= {threshold}"
    )

    threshold_edges = connectivity_edges[
        connectivity_edges[
            "synapse_count"
        ] >= threshold
    ].copy()

    threshold_features = (
        build_features_from_edges(
            threshold_edges,
            target_body_ids_cp10
        )
    )

    # ------------------------------------------------
    # Align with ML dataset
    # ------------------------------------------------

    threshold_features = (
        final_ml_dataset[
            ["bodyId"]
        ]
        .merge(
            threshold_features,
            on="bodyId",
            how="left",
            validate="one_to_one"
        )
    )

    threshold_features = threshold_features[
        ml_feature_columns
    ]

    assert threshold_features.shape == (
        165,
        16
    )

    assert (
        threshold_features
        .isna()
        .sum()
        .sum()
        == 0
    )

    fold_results, summary = (
        evaluate_logistic_grouped_cv(
            threshold_features,
            y_cp10,
            groups_cp10,
            cp10_splits
        )
    )

    threshold_records.append({
        "synapse_threshold":
            threshold,

        "n_edges_retained":
            len(threshold_edges),

        "balanced_accuracy_mean":
            summary[
                "balanced_accuracy_mean"
            ],

        "balanced_accuracy_sd":
            summary[
                "balanced_accuracy_sd"
            ],

        "macro_f1_mean":
            summary[
                "macro_f1_mean"
            ],

        "macro_f1_sd":
            summary[
                "macro_f1_sd"
            ],

        "ACh_recall_mean":
            summary[
                "ACh_recall_mean"
            ],

        "GABA_recall_mean":
            summary[
                "GABA_recall_mean"
            ]
    })

threshold_sensitivity_results = pd.DataFrame(
    threshold_records
)

threshold_sensitivity_results[
    "balanced_accuracy_change_vs_threshold1"
] = (
    threshold_sensitivity_results[
        "balanced_accuracy_mean"
    ]
    - threshold_sensitivity_results.loc[
        threshold_sensitivity_results[
            "synapse_threshold"
        ] == 1,
        "balanced_accuracy_mean"
    ].iloc[0]
)

display(
    threshold_sensitivity_results.round(4)
)

print(
    "\n✓ Connectivity-threshold sensitivity analysis complete."
)
```

```python
# ================================================================
# CHECKPOINT 10.10 — GROUP-PRESERVING PERMUTATION SANITY CHECK
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.10 — GROUP-PRESERVING PERMUTATION SANITY CHECK")
print("=" * 70)

N_PERMUTATIONS = 100
PERMUTATION_RANDOM_STATE = 42

permutation_rng = np.random.default_rng(
    PERMUTATION_RANDOM_STATE
)

# ------------------------------------------------
# One molecular label per canonical cell type
# ------------------------------------------------

cell_type_labels = (
    final_ml_dataset[
        [
            "cell_type",
            "neurotransmitter"
        ]
    ]
    .drop_duplicates(
        subset=["cell_type"]
    )
    .copy()
)

assert (
    cell_type_labels[
        "cell_type"
    ].nunique()
    == 65
)

# Every canonical type must have exactly one label.
label_counts = (
    final_ml_dataset
    .groupby("cell_type")
    ["neurotransmitter"]
    .nunique()
)

assert (
    label_counts == 1
).all()

original_group_labels = (
    cell_type_labels[
        "neurotransmitter"
    ]
    .to_numpy()
)

permutation_records = []

print(
    f"\nRunning {N_PERMUTATIONS} "
    "group-preserving permutations..."
)

for permutation_number in range(
    N_PERMUTATIONS
):

    # ------------------------------------------------
    # Shuffle labels BETWEEN cell types.
    # ------------------------------------------------

    shuffled_labels = (
        permutation_rng
        .permutation(
            original_group_labels
        )
    )

    shuffled_group_label_map = dict(
        zip(
            cell_type_labels[
                "cell_type"
            ],
            shuffled_labels
        )
    )

    y_permuted = groups_cp10.map(
        shuffled_group_label_map
    )

    assert len(y_permuted) == 165

    # ------------------------------------------------
    # Confirm label remains constant within
    # each canonical cell type.
    # ------------------------------------------------

    assert (
        y_permuted
        .groupby(groups_cp10)
        .nunique()
        .eq(1)
        .all()
    )

    # ------------------------------------------------
    # Evaluate Logistic Regression
    # ------------------------------------------------

    _, perm_summary = (
        evaluate_logistic_grouped_cv(
            X_cp10,
            y_permuted,
            groups_cp10,
            cp10_splits
        )
    )

    permutation_records.append({
        "permutation":
            permutation_number + 1,

        "balanced_accuracy":
            perm_summary[
                "balanced_accuracy_mean"
            ],

        "macro_f1":
            perm_summary[
                "macro_f1_mean"
            ]
    })

permutation_results = pd.DataFrame(
    permutation_records
)

print("\nPermutation results:")
display(
    permutation_results.describe().round(4)
)

print(
    "\n✓ 100 group-preserving permutations completed."
)
```

```python
# ================================================================
# CHECKPOINT 10.11 — PERMUTATION NULL COMPARISON
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.11 — PERMUTATION NULL COMPARISON")
print("=" * 70)

observed_ba = (
    reference_balanced_accuracy
)

null_ba = (
    permutation_results[
        "balanced_accuracy"
    ]
)

null_mean = null_ba.mean()
null_sd = null_ba.std()

null_95_low = null_ba.quantile(
    0.025
)

null_95_high = null_ba.quantile(
    0.975
)

# Empirical one-sided permutation p-value
permutation_p_value = (
    1
    + (
        null_ba >= observed_ba
    ).sum()
) / (
    len(null_ba) + 1
)

permutation_summary = pd.DataFrame([
    {
        "observed_balanced_accuracy":
            observed_ba,

        "null_mean":
            null_mean,

        "null_sd":
            null_sd,

        "null_2.5_percentile":
            null_95_low,

        "null_97.5_percentile":
            null_95_high,

        "empirical_p_value":
            permutation_p_value
    }
])

display(
    permutation_summary.round(4)
)

print(
    f"\nObserved grouped-CV BA: "
    f"{observed_ba:.4f}"
)

print(
    f"Permutation null mean: "
    f"{null_mean:.4f}"
)

print(
    f"Permutation null SD: "
    f"{null_sd:.4f}"
)

print(
    f"Permutation empirical p-value: "
    f"{permutation_p_value:.4f}"
)

print(
    "\n✓ Observed-vs-null comparison completed."
)
```

```python
# ================================================================
# CHECKPOINT 10.12 — PERMUTATION NULL DISTRIBUTION
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.12 — PERMUTATION NULL DISTRIBUTION")
print("=" * 70)

plt.figure(
    figsize=(9, 6)
)

plt.hist(
    permutation_results[
        "balanced_accuracy"
    ],
    bins=20
)

plt.axvline(
    observed_ba,
    linestyle="--",
    linewidth=2,
    label=(
        f"Observed BA = "
        f"{observed_ba:.3f}"
    )
)

plt.axvline(
    0.5,
    linestyle=":",
    linewidth=2,
    label="Chance = 0.500"
)

plt.xlabel(
    "Balanced Accuracy"
)

plt.ylabel(
    "Number of Permutations"
)

plt.title(
    "Group-Preserving Permutation Null Distribution"
)

plt.legend()

plt.tight_layout()

plt.show()

print(
    "\n✓ Permutation null distribution plotted."
)
```

```python
# ================================================================
# CHECKPOINT 10.13 — ROBUSTNESS SUMMARY
# ================================================================

print("=" * 70)
print("CHECKPOINT 10.13 — ROBUSTNESS SUMMARY")
print("=" * 70)

# ------------------------------------------------
# Reference
# ------------------------------------------------

robustness_records = []

robustness_records.append({
    "analysis_type":
        "Reference",

    "condition":
        "All 16 raw features",

    "n_features":
        16,

    "balanced_accuracy_mean":
        reference_summary[
            "balanced_accuracy_mean"
        ],

    "balanced_accuracy_sd":
        reference_summary[
            "balanced_accuracy_sd"
        ],

    "macro_f1_mean":
        reference_summary[
            "macro_f1_mean"
        ],

    "macro_f1_sd":
        reference_summary[
            "macro_f1_sd"
        ]
})

# ------------------------------------------------
# Feature families
# ------------------------------------------------

for _, row in feature_family_results.iterrows():

    if row[
        "feature_set"
    ] == "All_features":

        continue

    robustness_records.append({
        "analysis_type":
            "Feature family",

        "condition":
            row["feature_set"],

        "n_features":
            row["n_features"],

        "balanced_accuracy_mean":
            row[
                "balanced_accuracy_mean"
            ],

        "balanced_accuracy_sd":
            row[
                "balanced_accuracy_sd"
            ],

        "macro_f1_mean":
            row[
                "macro_f1_mean"
            ],

        "macro_f1_sd":
            row[
                "macro_f1_sd"
            ]
    })

# ------------------------------------------------
# Log transformation
# ------------------------------------------------

robustness_records.append({
    "analysis_type":
        "Transformation",

    "condition":
        "Log1p connectivity features",

    "n_features":
        16,

    "balanced_accuracy_mean":
        log_summary[
            "balanced_accuracy_mean"
        ],

    "balanced_accuracy_sd":
        log_summary[
            "balanced_accuracy_sd"
        ],

    "macro_f1_mean":
        log_summary[
            "macro_f1_mean"
        ],

    "macro_f1_sd":
        log_summary[
            "macro_f1_sd"
        ]
})

# ------------------------------------------------
# Thresholds
# ------------------------------------------------

for _, row in (
    threshold_sensitivity_results
    .iterrows()
):

    threshold = int(
        row[
            "synapse_threshold"
        ]
    )

    if threshold == 1:
        continue

    robustness_records.append({
        "analysis_type":
            "Connectivity threshold",

        "condition":
            f"Minimum synapses = "
            f"{threshold}",

        "n_features":
            16,

        "balanced_accuracy_mean":
            row[
                "balanced_accuracy_mean"
            ],

        "balanced_accuracy_sd":
            row[
                "balanced_accuracy_sd"
            ],

        "macro_f1_mean":
            row[
                "macro_f1_mean"
            ],

        "macro_f1_sd":
            row[
                "macro_f1_sd"
            ]
    })

robustness_summary = pd.DataFrame(
    robustness_records
)

robustness_summary[
    "change_vs_reference"
] = (
    robustness_summary[
        "balanced_accuracy_mean"
    ]
    - reference_balanced_accuracy
)

display(
    robustness_summary.round(4)
)

print(
    "\n✓ Robustness summary assembled."
)
```

```python
# ================================================================
# CHECKPOINT 10 — FINAL AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 10 — FINAL AUDIT")
print("=" * 70)

# ================================================================
# DATA
# ================================================================

assert X_cp10.shape == (
    165,
    16
)

assert len(y_cp10) == 165

assert len(groups_cp10) == 165

assert groups_cp10.nunique() == 65

assert X_cp10.isna().sum().sum() == 0

assert y_cp10.isna().sum() == 0

assert groups_cp10.isna().sum() == 0


# ================================================================
# FEATURE AUDIT
# ================================================================

assert len(
    ml_feature_columns
) == 16

assert feature_overlap == set()

assert set(
    X_cp10.columns
) == set(
    ml_feature_columns
)


# ================================================================
# GROUPED FOLD AUDIT
# ================================================================

assert len(cp10_splits) == 5

for train_idx, val_idx in cp10_splits:

    train_groups = set(
        groups_cp10.iloc[train_idx]
    )

    val_groups = set(
        groups_cp10.iloc[val_idx]
    )

    assert (
        train_groups.isdisjoint(
            val_groups
        )
    )


# ================================================================
# REFERENCE PERFORMANCE
# ================================================================

assert len(
    reference_fold_results
) == 5

assert (
    0
    <= reference_balanced_accuracy
    <= 1
)


# ================================================================
# FEATURE FAMILY ANALYSIS
# ================================================================

assert len(
    feature_family_results
) == len(
    feature_family_sets
)

assert (
    feature_family_results[
        "n_features"
    ] > 0
).all()


# ================================================================
# LOG TRANSFORMATION
# ================================================================

assert X_log.shape == (
    165,
    16
)

assert (
    X_log.isna().sum().sum()
    == 0
)

assert len(
    log_sensitivity_result
) == 2


# ================================================================
# THRESHOLD ANALYSIS
# ================================================================

assert set(
    threshold_sensitivity_results[
        "synapse_threshold"
    ]
) == {
    1,
    2,
    3,
    5
}

assert (
    threshold_sensitivity_results[
        "n_edges_retained"
    ]
    .diff()
    .iloc[1:]
    <= 0
).all()


# ================================================================
# PERMUTATION ANALYSIS
# ================================================================

assert len(
    permutation_results
) == N_PERMUTATIONS

assert (
    permutation_results[
        "balanced_accuracy"
    ]
    .between(0, 1)
    .all()
)

assert (
    permutation_results[
        "macro_f1"
    ]
    .between(0, 1)
    .all()
)

assert 0 <= (
    permutation_p_value
) <= 1


# ================================================================
# ROBUSTNESS SUMMARY
# ================================================================

assert len(
    robustness_summary
) > 1

assert (
    robustness_summary[
        "balanced_accuracy_mean"
    ]
    .between(0, 1)
    .all()
)


# ================================================================
# FINAL OUTPUT
# ================================================================

print("\nDATA")
print("----")

print(
    "Total neurons:",
    len(X_cp10)
)

print(
    "Canonical cell types:",
    groups_cp10.nunique()
)

print(
    "Connectivity features:",
    len(ml_feature_columns)
)


print("\nVALIDATION")
print("----------")

print(
    "Method:",
    "5-fold StratifiedGroupKFold"
)

print(
    "Grouping:",
    "canonical cell_type"
)

print(
    "Primary model:",
    "Logistic Regression"
)


print("\nREFERENCE PERFORMANCE")
print("---------------------")

print(
    f"Balanced accuracy: "
    f"{reference_balanced_accuracy:.4f} "
    f"± "
    f"{reference_summary['balanced_accuracy_sd']:.4f}"
)

print(
    f"Macro F1: "
    f"{reference_macro_f1:.4f} "
    f"± "
    f"{reference_summary['macro_f1_sd']:.4f}"
)


print("\nPERMUTATION SANITY CHECK")
print("------------------------")

print(
    f"Observed BA: "
    f"{observed_ba:.4f}"
)

print(
    f"Permutation null mean: "
    f"{null_mean:.4f}"
)

print(
    f"Permutation null SD: "
    f"{null_sd:.4f}"
)

print(
    f"Empirical p-value: "
    f"{permutation_p_value:.4f}"
)


print("\nROBUSTNESS TESTS")
print("----------------")

print(
    "Feature-family sensitivity: COMPLETE"
)

print(
    "Log1p sensitivity: COMPLETE"
)

print(
    "Connectivity-threshold sensitivity: COMPLETE"
)

print(
    "Group-preserving permutation: COMPLETE"
)

print(
    "Leakage audit: PASSED"
)

print(
    "\n✓ No identifier or molecular-label leakage."
)

print(
    "✓ Group separation preserved throughout."
)

print(
    "✓ Preprocessing performed within validation folds."
)

print(
    "✓ Robustness analyses completed."
)

print(
    "\n" + "=" * 70
)

print(
    "✓ CHECKPOINT 10 COMPLETE"
)

print(
    "=" * 70
)
```

```python
# ================================================================
# CHECKPOINT 11.1 — FEATURE INTERPRETATION SETUP
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.1 — FEATURE INTERPRETATION SETUP")
print("=" * 70)

# ------------------------------------------------
# Primary interpretation model
# ------------------------------------------------

CP11_MODEL = "Logistic Regression"

# ------------------------------------------------
# Verify the exact dataset used in Checkpoint 10
# ------------------------------------------------

assert X_cp10.shape == (165, 16)
assert len(y_cp10) == 165
assert len(groups_cp10) == 165
assert groups_cp10.nunique() == 65

assert list(X_cp10.columns) == ml_feature_columns

assert X_cp10.isna().sum().sum() == 0
assert y_cp10.isna().sum() == 0
assert groups_cp10.isna().sum() == 0

# ------------------------------------------------
# Verify the fixed grouped folds
# ------------------------------------------------

assert len(cp10_splits) == 5

for train_idx, val_idx in cp10_splits:

    train_groups = set(
        groups_cp10.iloc[train_idx]
    )

    val_groups = set(
        groups_cp10.iloc[val_idx]
    )

    assert train_groups.isdisjoint(val_groups)

print("\nDataset:")
print("  Neurons:", len(X_cp10))
print("  Canonical cell types:", groups_cp10.nunique())
print("  Connectivity features:", len(ml_feature_columns))

print("\nInterpretation strategy:")
print("  1. Standardized Logistic Regression coefficients")
print("  2. Held-out grouped-CV permutation importance")
print("  3. Feature-rank stability across folds")
print("  4. Combined interpretation")

print("\n✓ Checkpoint 11 setup validated.")
```

```python
# ================================================================
# CHECKPOINT 11.2 — GROUPED-CV LOGISTIC MODELS
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.2 — GROUPED-CV LOGISTIC MODELS")
print("=" * 70)

# Import only the new inspection function needed for this checkpoint.
from sklearn.inspection import permutation_importance

cp11_fold_models = []

for fold_number, (train_idx, val_idx) in enumerate(
    cp10_splits,
    start=1
):

    X_train = X_cp10.iloc[train_idx].copy()
    X_val = X_cp10.iloc[val_idx].copy()

    y_train = y_cp10.iloc[train_idx].copy()
    y_val = y_cp10.iloc[val_idx].copy()

    model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "logistic",
            LogisticRegression(
                max_iter=2000,
                class_weight="balanced",
                random_state=42
            )
        )
    ])

    model.fit(
        X_train,
        y_train
    )

    predictions = model.predict(
        X_val
    )

    cp11_fold_models.append({
        "fold": fold_number,
        "model": model,
        "train_idx": train_idx,
        "val_idx": val_idx,
        "X_train": X_train,
        "X_val": X_val,
        "y_train": y_train,
        "y_val": y_val,
        "predictions": predictions
    })

print(
    f"\n✓ {len(cp11_fold_models)} grouped Logistic Regression models fitted."
)

print("✓ Every model was fitted using training neurons only.")
print("✓ Validation neurons remained held out.")
```

```python
# ================================================================
# CHECKPOINT 11.3 — STANDARDIZED LOGISTIC COEFFICIENTS
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.3 — STANDARDIZED LOGISTIC COEFFICIENTS")
print("=" * 70)

coefficient_records = []

for fold_record in cp11_fold_models:

    fold_number = fold_record["fold"]
    model = fold_record["model"]

    logistic_step = model.named_steps["logistic"]

    coefficients = logistic_step.coef_[0]

    for feature, coefficient in zip(
        ml_feature_columns,
        coefficients
    ):

        coefficient_records.append({
            "fold": fold_number,
            "feature": feature,
            "coefficient": coefficient,
            "absolute_coefficient": abs(coefficient)
        })

coefficient_by_fold = pd.DataFrame(
    coefficient_records
)

print("\nCoefficient table shape:")
print(coefficient_by_fold.shape)

assert coefficient_by_fold.shape == (80, 4)
assert coefficient_by_fold["feature"].nunique() == 16
assert coefficient_by_fold["fold"].nunique() == 5

print("\n✓ Standardized coefficients extracted from all five folds.")
```

```python
# ================================================================
# CHECKPOINT 11.4 — COEFFICIENT STABILITY
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.4 — COEFFICIENT STABILITY")
print("=" * 70)

coefficient_summary = (
    coefficient_by_fold
    .groupby("feature")
    .agg(
        mean_coefficient=(
            "coefficient",
            "mean"
        ),
        sd_coefficient=(
            "coefficient",
            "std"
        ),
        mean_absolute_coefficient=(
            "absolute_coefficient",
            "mean"
        ),
        median_absolute_coefficient=(
            "absolute_coefficient",
            "median"
        )
    )
    .reset_index()
)

# ------------------------------------------------
# Sign stability
# ------------------------------------------------

def calculate_sign_consistency(values):

    values = np.asarray(values)

    positive = np.sum(values > 0)
    negative = np.sum(values < 0)

    if positive >= negative:
        dominant_sign = "positive"
        consistency = positive / len(values)
    else:
        dominant_sign = "negative"
        consistency = negative / len(values)

    return pd.Series({
        "dominant_coefficient_sign": dominant_sign,
        "sign_consistency": consistency
    })


sign_summary = (
    coefficient_by_fold
    .groupby("feature")["coefficient"]
    .apply(calculate_sign_consistency)
    .unstack()
    .reset_index()
)

coefficient_summary = coefficient_summary.merge(
    sign_summary,
    on="feature",
    how="left"
)

coefficient_summary = (
    coefficient_summary
    .sort_values(
        "mean_absolute_coefficient",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    coefficient_summary.round(6)
)

print(
    "\n✓ Coefficient magnitude and sign stability summarized."
)
```

```python
# ================================================================
# CHECKPOINT 11.5 — HELD-OUT GROUPED-CV PERMUTATION IMPORTANCE
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.5 — HELD-OUT GROUPED-CV PERMUTATION IMPORTANCE")
print("=" * 70)

permutation_importance_records = []

for fold_record in cp11_fold_models:

    fold_number = fold_record["fold"]
    model = fold_record["model"]
    X_val = fold_record["X_val"]
    y_val = fold_record["y_val"]

    print(
        f"Calculating permutation importance for fold {fold_number}..."
    )

    fold_importance = permutation_importance(
        estimator=model,
        X=X_val,
        y=y_val,
        scoring="balanced_accuracy",
        n_repeats=30,
        random_state=1000 + fold_number,
        n_jobs=-1
    )

    for feature_index, feature in enumerate(
        ml_feature_columns
    ):

        permutation_importance_records.append({
            "fold": fold_number,
            "feature": feature,
            "permutation_importance_mean": (
                fold_importance.importances_mean[
                    feature_index
                ]
            ),
            "permutation_importance_sd": (
                fold_importance.importances_std[
                    feature_index
                ]
            )
        })

permutation_importance_by_fold = pd.DataFrame(
    permutation_importance_records
)

assert permutation_importance_by_fold.shape[0] == 80
assert permutation_importance_by_fold["fold"].nunique() == 5
assert permutation_importance_by_fold["feature"].nunique() == 16

print(
    "\n✓ Held-out permutation importance calculated."
)
print("✓ 30 permutations per feature per fold.")
print("✓ Balanced accuracy used as the importance metric.")
print("✓ Validation data were never used to fit the models.")
```

```python
# ================================================================
# CHECKPOINT 11.6 — PERMUTATION IMPORTANCE SUMMARY
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.6 — PERMUTATION IMPORTANCE SUMMARY")
print("=" * 70)

permutation_importance_summary = (
    permutation_importance_by_fold
    .groupby("feature")
    .agg(
        mean_permutation_importance=(
            "permutation_importance_mean",
            "mean"
        ),
        sd_permutation_importance=(
            "permutation_importance_mean",
            "std"
        ),
        median_permutation_importance=(
            "permutation_importance_mean",
            "median"
        ),
        mean_within_fold_permutation_sd=(
            "permutation_importance_sd",
            "mean"
        )
    )
    .reset_index()
)

# ------------------------------------------------
# Rank features
# ------------------------------------------------

permutation_importance_summary = (
    permutation_importance_summary
    .sort_values(
        "mean_permutation_importance",
        ascending=False
    )
    .reset_index(drop=True)
)

permutation_importance_summary[
    "permutation_rank"
] = np.arange(
    1,
    len(permutation_importance_summary) + 1
)

display(
    permutation_importance_summary.round(6)
)

print(
    "\n✓ Permutation importance ranked across the five grouped folds."
)
```

```python
# ================================================================
# CHECKPOINT 11.7 — PERMUTATION IMPORTANCE STABILITY
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.7 — PERMUTATION IMPORTANCE STABILITY")
print("=" * 70)

# ------------------------------------------------
# Count folds where feature importance is positive
# ------------------------------------------------

importance_stability = (
    permutation_importance_by_fold
    .groupby("feature")
    .agg(
        positive_importance_folds=(
            "permutation_importance_mean",
            lambda x: int((x > 0).sum())
        ),
        negative_importance_folds=(
            "permutation_importance_mean",
            lambda x: int((x < 0).sum())
        ),
        near_zero_importance_folds=(
            "permutation_importance_mean",
            lambda x: int((x.abs() < 1e-6).sum())
        )
    )
    .reset_index()
)

importance_stability[
    "positive_fold_fraction"
] = (
    importance_stability[
        "positive_importance_folds"
    ] / 5
)

importance_stability[
    "stable_positive_importance"
] = (
    importance_stability[
        "positive_importance_folds"
    ] >= 4
)

importance_stability = (
    importance_stability
    .sort_values(
        [
            "stable_positive_importance",
            "positive_fold_fraction"
        ],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

display(
    importance_stability
)

print(
    "\n✓ Cross-fold importance stability calculated."
)

print(
    "\nInterpretation rule:"
)

print(
    "  Stable positive = importance > 0 in at least 4 of 5 folds."
)
```

```python
# ================================================================
# CHECKPOINT 11.8 — COMBINED FEATURE INTERPRETATION
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.8 — COMBINED FEATURE INTERPRETATION")
print("=" * 70)

feature_interpretation = (
    coefficient_summary[
        [
            "feature",
            "mean_coefficient",
            "sd_coefficient",
            "mean_absolute_coefficient",
            "dominant_coefficient_sign",
            "sign_consistency"
        ]
    ]
    .merge(
        permutation_importance_summary[
            [
                "feature",
                "mean_permutation_importance",
                "sd_permutation_importance",
                "median_permutation_importance",
                "permutation_rank"
            ]
        ],
        on="feature",
        how="inner"
    )
    .merge(
        importance_stability[
            [
                "feature",
                "positive_importance_folds",
                "positive_fold_fraction",
                "stable_positive_importance"
            ]
        ],
        on="feature",
        how="inner"
    )
)

# ------------------------------------------------
# Rank by held-out permutation importance
# ------------------------------------------------

feature_interpretation = (
    feature_interpretation
    .sort_values(
        "mean_permutation_importance",
        ascending=False
    )
    .reset_index(drop=True)
)

assert len(feature_interpretation) == 16
assert feature_interpretation["feature"].nunique() == 16

display(
    feature_interpretation.round(6)
)

print(
    "\n✓ Combined feature interpretation table generated."
)
```

```python
# ================================================================
# CHECKPOINT 11.9 — TOP AND STABLE FEATURES
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.9 — TOP AND STABLE FEATURES")
print("=" * 70)

# ------------------------------------------------
# Top features by held-out permutation importance
# ------------------------------------------------

top_features = (
    feature_interpretation
    .head(10)
    .copy()
)

print("\nTop 10 features by mean held-out permutation importance:")
print("-" * 70)

display(
    top_features[
        [
            "feature",
            "mean_permutation_importance",
            "sd_permutation_importance",
            "positive_importance_folds",
            "mean_coefficient",
            "sign_consistency"
        ]
    ].round(6)
)

# ------------------------------------------------
# Most stable positive features
# ------------------------------------------------

stable_features = (
    feature_interpretation[
        feature_interpretation[
            "stable_positive_importance"
        ]
    ]
    .copy()
)

print(
    "\nFeatures with positive permutation importance in ≥4/5 folds:"
)
print("-" * 70)

display(
    stable_features[
        [
            "feature",
            "mean_permutation_importance",
            "positive_importance_folds",
            "positive_fold_fraction",
            "mean_coefficient",
            "sign_consistency"
        ]
    ].round(6)
)

print(
    "\n✓ Top and cross-fold-stable features identified."
)
```

```python
# ================================================================
# CHECKPOINT 11.10 — INTERPRETATION METHOD AGREEMENT
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.10 — INTERPRETATION METHOD AGREEMENT")
print("=" * 70)

feature_interpretation[
    "coefficient_rank"
] = (
    feature_interpretation[
        "mean_absolute_coefficient"
    ]
    .rank(
        ascending=False,
        method="average"
    )
)

# Spearman correlation between the two rankings

rank_correlation = (
    feature_interpretation[
        [
            "coefficient_rank",
            "permutation_rank"
        ]
    ]
    .corr(
        method="spearman"
    )
    .loc[
        "coefficient_rank",
        "permutation_rank"
    ]
)

print(
    f"\nSpearman correlation between coefficient and "
    f"permutation-importance rankings: {rank_correlation:.4f}"
)

# ------------------------------------------------
# Ranking table
# ------------------------------------------------

ranking_comparison = (
    feature_interpretation[
        [
            "feature",
            "coefficient_rank",
            "permutation_rank",
            "mean_absolute_coefficient",
            "mean_permutation_importance"
        ]
    ]
    .sort_values(
        "permutation_rank"
    )
    .reset_index(drop=True)
)

display(
    ranking_comparison.round(6)
)

print(
    "\n✓ Agreement between interpretation methods assessed."
)
```

```python
# ================================================================
# CHECKPOINT 11.11 — FEATURE-FAMILY IMPORTANCE — CORRECTED
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.11 — FEATURE-FAMILY IMPORTANCE — CORRECTED")
print("=" * 70)

# ------------------------------------------------
# Correct feature-family assignment
# IMPORTANT:
# Directionality ratios must be checked BEFORE
# the generic outgoing_ / incoming_ prefixes.
# ------------------------------------------------

feature_family_map = {}

for feature in ml_feature_columns:

    # Directionality must come first
    if feature in [
        "outgoing_incoming_edge_ratio",
        "outgoing_incoming_synapse_ratio"
    ]:
        family = "Directionality"

    elif feature.startswith("outgoing_"):
        family = "Outgoing connectivity"

    elif feature.startswith("incoming_"):
        family = "Incoming connectivity"

    elif feature in [
        "total_edges",
        "total_synapses"
    ]:
        family = "Overall connectivity"

    else:
        family = "Other"

    feature_family_map[feature] = family


# ------------------------------------------------
# Apply mapping
# ------------------------------------------------

feature_interpretation[
    "feature_family"
] = (
    feature_interpretation["feature"]
    .map(feature_family_map)
)

assert (
    feature_interpretation[
        "feature_family"
    ].isna().sum()
    == 0
)


# ------------------------------------------------
# Verify expected family sizes
# ------------------------------------------------

family_counts = (
    feature_interpretation[
        "feature_family"
    ]
    .value_counts()
)

print("\nFeature-family counts:")
print(family_counts)


assert family_counts[
    "Outgoing connectivity"
] == 6

assert family_counts[
    "Incoming connectivity"
] == 6

assert family_counts[
    "Overall connectivity"
] == 2

assert family_counts[
    "Directionality"
] == 2


# ------------------------------------------------
# Summarize family-level importance
# ------------------------------------------------

feature_family_importance = (
    feature_interpretation
    .groupby("feature_family")
    .agg(
        n_features=(
            "feature",
            "count"
        ),

        mean_permutation_importance=(
            "mean_permutation_importance",
            "mean"
        ),

        total_absolute_permutation_importance=(
            "mean_permutation_importance",
            lambda x: np.abs(x).sum()
        ),

        mean_absolute_coefficient=(
            "mean_absolute_coefficient",
            "mean"
        )
    )
    .reset_index()
)


# ------------------------------------------------
# Relative contribution
# ------------------------------------------------

total_importance = (
    feature_family_importance[
        "total_absolute_permutation_importance"
    ].sum()
)

assert total_importance > 0

feature_family_importance[
    "relative_permutation_importance"
] = (
    feature_family_importance[
        "total_absolute_permutation_importance"
    ]
    / total_importance
)


feature_family_importance = (
    feature_family_importance
    .sort_values(
        "relative_permutation_importance",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------
# Display
# ------------------------------------------------

display(
    feature_family_importance.round(6)
)


# ------------------------------------------------
# Save corrected result
# ------------------------------------------------

feature_family_importance.to_csv(
    "results/tables/checkpoint11_feature_family_importance.csv",
    index=False
)

print(
    "\n✓ Feature-family assignment corrected."
)

print(
    "✓ Directionality ratios correctly separated."
)

print(
    "✓ Corrected family-level importance saved."
)
```

```python
# ================================================================
# CHECKPOINT 11.15 — FINAL SCIENTIFIC AUDIT
# ================================================================

print("=" * 70)
print("CHECKPOINT 11.15 — FINAL SCIENTIFIC AUDIT")
print("=" * 70)

# ================================================================
# DATA AUDIT
# ================================================================

assert X_cp10.shape == (165, 16)

assert len(y_cp10) == 165

assert len(groups_cp10) == 165

assert groups_cp10.nunique() == 65

assert len(ml_feature_columns) == 16

assert (
    X_cp10.isna().sum().sum()
    == 0
)

# ================================================================
# FOLD AUDIT
# ================================================================

assert len(cp10_splits) == 5

for train_idx, val_idx in cp10_splits:

    train_groups = set(
        groups_cp10.iloc[train_idx]
    )

    val_groups = set(
        groups_cp10.iloc[val_idx]
    )

    assert train_groups.isdisjoint(
        val_groups
    )

# ================================================================
# COEFFICIENT AUDIT
# ================================================================

assert len(
    coefficient_by_fold
) == 80

assert (
    coefficient_by_fold[
        "feature"
    ].nunique()
    == 16
)

assert (
    coefficient_by_fold[
        "fold"
    ].nunique()
    == 5
)

# ================================================================
# PERMUTATION AUDIT
# ================================================================

assert len(
    permutation_importance_by_fold
) == 80

assert (
    permutation_importance_by_fold[
        "feature"
    ].nunique()
    == 16
)

assert (
    permutation_importance_by_fold[
        "fold"
    ].nunique()
    == 5
)

assert (
    permutation_importance_summary[
        "feature"
    ].nunique()
    == 16
)

# ================================================================
# COMBINED INTERPRETATION AUDIT
# ================================================================

assert len(
    feature_interpretation
) == 16

assert (
    feature_interpretation[
        "mean_permutation_importance"
    ].notna().all()
)

assert (
    feature_interpretation[
        "mean_absolute_coefficient"
    ].notna().all()
)

# ================================================================
# OUTPUT AUDIT
# ================================================================

assert os.path.exists(
    "results/figures/checkpoint11_permutation_feature_importance.png"
)

assert os.path.exists(
    "results/figures/checkpoint11_logistic_coefficients.png"
)

print("\nDATA")
print("----")
print("Total neurons:", len(X_cp10))
print("Canonical cell types:", groups_cp10.nunique())
print("Connectivity features:", len(ml_feature_columns))

print("\nINTERPRETATION")
print("-------------")
print("Primary model:", CP11_MODEL)
print("Validation:", "5-fold StratifiedGroupKFold")
print("Grouping:", "canonical cell_type")
print("Permutation repeats per fold:", 30)

print("\nMETHODS")
print("-------")
print("✓ Standardized Logistic Regression coefficients")
print("✓ Held-out permutation importance")
print("✓ Cross-fold importance stability")
print("✓ Feature-family interpretation")
print("✓ Coefficient/permutation rank agreement")

print("\nSCIENTIFIC SAFEGUARDS")
print("---------------------")
print("✓ No body IDs used as features.")
print("✓ No molecular labels used as features.")
print("✓ No cell-type labels used as features.")
print("✓ Group separation preserved.")
print("✓ Models fitted independently within each fold.")
print("✓ Permutation importance calculated on held-out data.")
print("✓ No feature selected using Checkpoint 6 p-values.")

print("\n" + "=" * 70)
print("✓ CHECKPOINT 11 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CHECKPOINT 12.4 — CELL-TYPE LEVEL EFFECT SIZES
# ================================================================

print("=" * 70)
print("CHECKPOINT 12.4 — CELL-TYPE LEVEL EFFECT SIZES")
print("=" * 70)

# ------------------------------------------------
# Cliff's delta helper
# ------------------------------------------------

def cliffs_delta(x, y):
    """
    Cliff's delta.

    Positive values:
        values in x tend to be larger than values in y.

    Negative values:
        values in x tend to be smaller than values in y.
    """

    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    x = x[np.isfinite(x)]
    y = y[np.isfinite(y)]

    if len(x) == 0 or len(y) == 0:
        return np.nan

    comparisons = (
        (x[:, None] > y[None, :]).sum()
        -
        (x[:, None] < y[None, :]).sum()
    )

    return comparisons / (len(x) * len(y))


# ------------------------------------------------
# Calculate effect sizes
# ------------------------------------------------

effect_size_records = []

ach_types = cell_type_level_dataset.loc[
    cell_type_level_dataset["neurotransmitter"] == "acetylcholine"
]

gaba_types = cell_type_level_dataset.loc[
    cell_type_level_dataset["neurotransmitter"] == "gaba"
]

for feature in ml_feature_columns:

    ach_values = ach_types[feature].values
    gaba_values = gaba_types[feature].values

    delta = cliffs_delta(
        ach_values,
        gaba_values
    )

    effect_size_records.append({
        "feature": feature,
        "ACh_median": np.median(ach_values),
        "GABA_median": np.median(gaba_values),
        "Cliffs_delta_ACh_vs_GABA": delta,
        "absolute_Cliffs_delta": abs(delta)
    })


cell_type_effect_sizes = (
    pd.DataFrame(effect_size_records)
    .sort_values(
        "absolute_Cliffs_delta",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    cell_type_effect_sizes.round(4)
)

print("\n✓ Cell-type-level effect sizes calculated.")
print("✓ Effect sizes are descriptive and do not treat neurons within a cell type as independent.")

# ------------------------------------------------
# Save
# ------------------------------------------------

cell_type_effect_sizes.to_csv(
    "results/tables/checkpoint12_cell_type_effect_sizes.csv",
    index=False
)

print(
    "\n✓ Effect-size table saved."
)

print("\n✓ CHECKPOINT 12.4 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 12.5 — PREDICTIVE IMPORTANCE VS TYPE-LEVEL SEPARATION
# ================================================================

print("=" * 70)
print("CHECKPOINT 12.5 — PREDICTIVE IMPORTANCE VS TYPE-LEVEL SEPARATION")
print("=" * 70)

# ------------------------------------------------
# Merge with Checkpoint 11 interpretation results
# ------------------------------------------------

biological_feature_comparison = (
    feature_interpretation[
        [
            "feature",
            "mean_permutation_importance",
            "permutation_rank",
            "mean_absolute_coefficient",
            "sign_consistency",
            "positive_importance_folds"
        ]
    ]
    .merge(
        cell_type_effect_sizes[
            [
                "feature",
                "Cliffs_delta_ACh_vs_GABA",
                "absolute_Cliffs_delta"
            ]
        ],
        on="feature",
        how="inner",
        validate="one_to_one"
    )
)

# ------------------------------------------------
# Rank type-level separation
# ------------------------------------------------

biological_feature_comparison[
    "type_level_separation_rank"
] = (
    biological_feature_comparison[
        "absolute_Cliffs_delta"
    ]
    .rank(
        ascending=False,
        method="min"
    )
)

biological_feature_comparison = (
    biological_feature_comparison
    .sort_values(
        "permutation_rank"
    )
    .reset_index(drop=True)
)

display(
    biological_feature_comparison.round(4)
)

# ------------------------------------------------
# Rank agreement
# ------------------------------------------------

rank_agreement_bio = scipy.stats.spearmanr(
    biological_feature_comparison[
        "permutation_rank"
    ],
    biological_feature_comparison[
        "type_level_separation_rank"
    ]
)

print(
    "\nSpearman correlation between:"
)
print(
    "  Predictive importance rank"
)
print(
    "  Type-level separation rank"
)
print(
    f"\nρ = {rank_agreement_bio.statistic:.4f}"
)
print(
    f"p = {rank_agreement_bio.pvalue:.4g}"
)

print(
    "\n✓ Predictive importance compared with independent type-level separation."
)

# ------------------------------------------------
# Save
# ------------------------------------------------

biological_feature_comparison.to_csv(
    "results/tables/checkpoint12_predictive_vs_biological_feature_comparison.csv",
    index=False
)

print(
    "\n✓ Comparison table saved."
)

print("\n✓ CHECKPOINT 12.5 COMPLETE")
```

```python
# ================================================================
# CHECKPOINT 12.6 — CONSISTENT BIOLOGICAL FEATURES
# ================================================================

print("=" * 70)
print("CHECKPOINT 12.6 — CONSISTENT BIOLOGICAL FEATURES")
print("=" * 70)

# ------------------------------------------------
# Normalize ranks
# ------------------------------------------------

biological_feature_comparison[
    "predictive_rank_score"
] = (
    1
    -
    (
        biological_feature_comparison[
            "permutation_rank"
        ] - 1
    )
    /
    (
        len(ml_feature_columns) - 1
    )
)

biological_feature_comparison[
    "type_level_separation_score"
] = (
    biological_feature_comparison[
        "absolute_Cliffs_delta"
    ]
    /
    biological_feature_comparison[
        "absolute_Cliffs_delta"
    ].max()
)

biological_feature_comparison[
    "stability_score"
] = (
    biological_feature_comparison[
        "positive_importance_folds"
    ]
    / 5
)

# ------------------------------------------------
# Combined descriptive score
#
# This is NOT a statistical significance score.
# It is only a ranking aid for interpretation.
# ------------------------------------------------

biological_feature_comparison[
    "combined_interpretation_score"
] = (
    biological_feature_comparison[
        "predictive_rank_score"
    ]
    +
    biological_feature_comparison[
        "type_level_separation_score"
    ]
    +
    biological_feature_comparison[
        "stability_score"
    ]
) / 3

consistent_biological_features = (
    biological_feature_comparison
    .sort_values(
        "combined_interpretation_score",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    consistent_biological_features[
        [
            "feature",
            "mean_permutation_importance",
            "permutation_rank",
            "absolute_Cliffs_delta",
            "positive_importance_folds",
            "sign_consistency",
            "combined_interpretation_score"
        ]
    ].head(10).round(4)
)

print(
    "\n✓ Predictive importance, type-level separation and fold stability combined."
)

print(
    "✓ Combined score is descriptive only and is not a hypothesis-testing statistic."
)

print("\n✓ CHECKPOINT 12.6 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 12.7 — WITHIN-CELL-TYPE CONNECTIVITY VARIABILITY
# ============================================================

print("CHECKPOINT 12.7 — WITHIN-CELL-TYPE VARIABILITY")
print("=" * 70)

within_type_cv_records = []

for feature in ml_feature_columns:

    grouped = final_ml_dataset.groupby("cell_type")[feature]

    group_sizes = grouped.size()
    within_type_mean = grouped.mean()
    within_type_std = grouped.std()

    # A coefficient of variation is meaningful only when:
    # 1. the cell type contains >=2 neurons
    # 2. the mean is non-zero
    # 3. the standard deviation is finite
    valid = (
        (group_sizes >= 2)
        & within_type_mean.notna()
        & within_type_mean.ne(0)
        & within_type_std.notna()
    )

    cv_values = (
        within_type_std[valid]
        / within_type_mean[valid].abs()
    )

    within_type_cv_records.append({
        "feature": feature,
        "n_cell_types_total": group_sizes.size,
        "n_cell_types_with_multiple_neurons": int((group_sizes >= 2).sum()),
        "n_cell_types_used_for_cv": int(valid.sum()),
        "median_within_type_cv": cv_values.median(),
        "mean_within_type_cv": cv_values.mean(),
        "max_within_type_cv": cv_values.max()
    })

within_type_cv_df = pd.DataFrame(within_type_cv_records)

print(within_type_cv_df.to_string(index=False))

# Save results
within_type_cv_df.to_csv(
    "results/checkpoint12_within_cell_type_cv.csv",
    index=False
)

print("\n✓ Within-cell-type variability analysis saved.")
print("  results/checkpoint12_within_cell_type_cv.csv")

print("\nInterpretation:")
print(
    "CV estimates are calculated only for cell types represented by "
    "at least two neurons and with non-zero mean feature values."
)
```

```python
# ================================================================
# CHECKPOINT 12.8 — BIOLOGICAL CONNECTIVITY DIMENSIONS
# ================================================================

print("=" * 70)
print("CHECKPOINT 12.8 — BIOLOGICAL CONNECTIVITY DIMENSIONS")
print("=" * 70)

biological_dimension_map = {

    "outgoing_mean_synapses":
        "Mean strength of outgoing synaptic connections",

    "outgoing_median_synapses":
        "Typical strength of outgoing synaptic connections",

    "outgoing_synapses":
        "Total outgoing synaptic weight",

    "outgoing_max_synapses":
        "Strongest individual outgoing connection",

    "outgoing_synapse_sd":
        "Heterogeneity of outgoing connection strengths",

    "outgoing_edges":
        "Number of distinct outgoing partners",

    "incoming_mean_synapses":
        "Mean strength of incoming synaptic connections",

    "incoming_median_synapses":
        "Typical strength of incoming synaptic connections",

    "incoming_synapses":
        "Total incoming synaptic weight",

    "incoming_max_synapses":
        "Strongest individual incoming connection",

    "incoming_synapse_sd":
        "Heterogeneity of incoming connection strengths",

    "incoming_edges":
        "Number of distinct incoming partners",

    "total_edges":
        "Overall number of distinct connectivity partners",

    "total_synapses":
        "Overall synaptic weight",

    "outgoing_incoming_edge_ratio":
        "Balance between outgoing and incoming partner counts",

    "outgoing_incoming_synapse_ratio":
        "Balance between outgoing and incoming synaptic weight"
}


biological_interpretation_table = (
    biological_feature_comparison[
        [
            "feature",
            "mean_permutation_importance",
            "permutation_rank",
            "absolute_Cliffs_delta",
            "positive_importance_folds",
            "sign_consistency"
        ]
    ]
    .copy()
)

biological_interpretation_table[
    "connectivity_dimension"
] = (
    biological_interpretation_table[
        "feature"
    ]
    .map(biological_dimension_map)
)

assert (
    biological_interpretation_table[
        "connectivity_dimension"
    ].notna().all()
)

# ------------------------------------------------
# Interpretation category
# ------------------------------------------------

def interpretation_category(row):

    if (
        row["positive_importance_folds"] >= 4
        and
        row["permutation_rank"] <= 3
    ):
        return "Strong and cross-fold consistent predictive signal"

    elif (
        row["positive_importance_folds"] >= 3
        and
        row["permutation_rank"] <= 8
    ):
        return "Moderate predictive signal with partial stability"

    elif row["mean_permutation_importance"] > 0:
        return "Weak or unstable predictive contribution"

    else:
        return "Little evidence of predictive contribution"


biological_interpretation_table[
    "interpretation_category"
] = (
    biological_interpretation_table
    .apply(
        interpretation_category,
        axis=1
    )
)

display(
    biological_interpretation_table.round(4)
)

# ------------------------------------------------
# Save
# ------------------------------------------------

biological_interpretation_table.to_csv(
    "results/tables/checkpoint12_biological_interpretation_table.csv",
    index=False
)

print(
    "\n✓ Biological connectivity dimensions assigned."
)

print(
    "✓ Interpretation categories based on predictive stability, "
    "not statistical significance."
)

print("\n✓ CHECKPOINT 12.8 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 12.12 — FINAL SCIENTIFIC AUDIT
# ============================================================

print("=" * 80)
print("CHECKPOINT 12 — FINAL SCIENTIFIC AUDIT")
print("=" * 80)

audit = {}

# ============================================================
# 1. DATASET STRUCTURE
# ============================================================

audit["165 neuron-level observations"] = (
    len(final_ml_dataset) == 165
)

audit["165 unique neuron body IDs"] = (
    final_ml_dataset["bodyId"].nunique() == 165
)

audit["65 canonical cell types"] = (
    final_ml_dataset["cell_type"].nunique() == 65
)

audit["16 connectivity features"] = (
    len(ml_feature_columns) == 16
)

audit["No missing connectivity features"] = (
    final_ml_dataset[ml_feature_columns]
    .notna()
    .all()
    .all()
)

audit["No missing neurotransmitter labels"] = (
    final_ml_dataset["neurotransmitter"]
    .notna()
    .all()
)

# ============================================================
# 2. CELL-TYPE-LEVEL AGGREGATION
# ============================================================

audit["Cell-type dataset contains 65 observations"] = (
    len(cell_type_level_dataset) == 65
)

audit["65 unique canonical cell types after aggregation"] = (
    cell_type_level_dataset["cell_type"].nunique() == 65
)

audit["Exactly one neurotransmitter label per cell type"] = (
    cell_type_level_dataset
    .groupby("cell_type")["neurotransmitter"]
    .nunique()
    .max()
    == 1
)

audit["Cell-type aggregation contains no missing features"] = (
    cell_type_level_dataset[ml_feature_columns]
    .notna()
    .all()
    .all()
)

audit["Only ACh and GABA represented"] = (
    set(cell_type_level_dataset["neurotransmitter"].unique())
    == {"acetylcholine", "gaba"}
)

# ============================================================
# 3. WITHIN-CELL-TYPE VARIABILITY
# ============================================================

group_sizes = (
    final_ml_dataset
    .groupby("cell_type")
    .size()
)

n_multi_neuron_types = int(
    (group_sizes >= 2).sum()
)

audit["Within-type analysis identifies multi-neuron cell types"] = (
    within_type_cv_df[
        "n_cell_types_with_multiple_neurons"
    ].nunique() == 1
    and
    int(
        within_type_cv_df[
            "n_cell_types_with_multiple_neurons"
        ].iloc[0]
    )
    == n_multi_neuron_types
)

audit["Within-type CV excludes single-neuron groups"] = (
    (
        within_type_cv_df[
            "n_cell_types_used_for_cv"
        ]
        <=
        within_type_cv_df[
            "n_cell_types_with_multiple_neurons"
        ]
    ).all()
)

# ============================================================
# 4. PREDICTIVE INTERPRETATION OUTPUTS
# ============================================================

audit["Feature interpretation available"] = (
    "feature_interpretation" in globals()
)

audit["Feature-family importance available"] = (
    "feature_family_importance" in globals()
)

audit["Biological feature comparison available"] = (
    "biological_feature_comparison" in globals()
)

audit["Cell-type effect-size analysis available"] = (
    "cell_type_effect_sizes" in globals()
)

audit["Biological interpretation table available"] = (
    "biological_interpretation_table" in globals()
)

# ============================================================
# 5. FEATURE INTERPRETATION COMPLETENESS
# ============================================================

audit["All 16 features have interpretation results"] = (
    len(feature_interpretation) == len(ml_feature_columns)
)

audit["All 16 features have biological effect sizes"] = (
    len(cell_type_effect_sizes) == len(ml_feature_columns)
)

audit["All 16 features mapped to biological dimensions"] = (
    len(biological_interpretation_table)
    == len(ml_feature_columns)
)

audit["No unmapped biological dimensions"] = (
    biological_interpretation_table[
        "connectivity_dimension"
    ].notna().all()
)

# ============================================================
# 6. SCIENTIFIC CLAIM BOUNDARIES
# ============================================================

claim_boundaries = {

    "Predictive association supported": True,

    "Mechanistic causality demonstrated": False,

    "Universal prediction across all Drosophila neurons demonstrated": False,

    "Deterministic molecular identity from connectivity demonstrated": False,

    "Generalization beyond examined olfactory population demonstrated": False,

    "Connectivity features shown to be independently causal": False
}

audit["Predictive association appropriately supported"] = (
    claim_boundaries[
        "Predictive association supported"
    ] is True
)

audit["Mechanistic causality appropriately rejected"] = (
    claim_boundaries[
        "Mechanistic causality demonstrated"
    ] is False
)

audit["Universal generalization appropriately rejected"] = (
    claim_boundaries[
        "Universal prediction across all Drosophila neurons demonstrated"
    ] is False
)

audit["Deterministic interpretation appropriately rejected"] = (
    claim_boundaries[
        "Deterministic molecular identity from connectivity demonstrated"
    ] is False
)

audit["Independent causal interpretation appropriately rejected"] = (
    claim_boundaries[
        "Connectivity features shown to be independently causal"
    ] is False
)

# ============================================================
# 7. REQUIRED OUTPUT FILES
# ============================================================

required_outputs = [

    "data/processed/"
    "cell_type_level_connectivity_molecular_dataset.csv",

    "results/tables/"
    "checkpoint12_cell_type_effect_sizes.csv",

    "results/checkpoint12_within_cell_type_cv.csv",

    "results/checkpoint12_claim_audit.csv",

    "results/checkpoint12_biological_conclusion.txt",

    "results/tables/"
    "checkpoint12_predictive_vs_biological_feature_comparison.csv",

    "results/tables/"
    "checkpoint12_biological_interpretation_table.csv",

    "results/figures/"
    "checkpoint12_cell_type_connectivity_by_neurotransmitter.png"
]

for output_file in required_outputs:

    audit[
        f"Output exists: {output_file}"
    ] = os.path.exists(output_file)

# ============================================================
# 8. DISPLAY AUDIT RESULTS
# ============================================================

print("\nAUDIT RESULTS")
print("-" * 80)

for name, status in audit.items():

    symbol = "✓" if status else "✗"

    print(
        f"{symbol} {name}"
    )

# ============================================================
# 9. AUDIT SUMMARY
# ============================================================

passed = sum(audit.values())
total = len(audit)

print("\n" + "=" * 80)
print(
    f"AUDIT SCORE: {passed}/{total} checks passed"
)
print("=" * 80)

# ============================================================
# 10. FINAL SCIENTIFIC STATUS
# ============================================================

if passed == total:

    print("\n✓ CHECKPOINT 12 PASSED")

    print("\nFINAL SCIENTIFIC CONCLUSION")
    print("-" * 80)

    print(
        "The analysis supports the conclusion that connectivity "
        "architecture contains predictive information about "
        "neurotransmitter identity within the examined "
        "Drosophila olfactory neuron population."
    )

    print(
        "\nThe strongest predictive contributions are associated "
        "with features describing synaptic-weight distributions, "
        "including outgoing mean and median synaptic strength "
        "and incoming maximum synaptic strength."
    )

    print(
        "\nThese findings are interpreted as associative predictive "
        "evidence rather than mechanistic or causal evidence."
    )

    print(
        "\nThe analysis does not establish that connectivity "
        "determines neurotransmitter identity, that the model "
        "generalizes universally across Drosophila neurons, "
        "or that any individual connectivity feature is causal."
    )

    print(
        "\n✓ Checkpoint 12 provides a scientifically bounded "
        "biological interpretation of the predictive results."
    )

else:

    print("\n⚠ CHECKPOINT 12 NEEDS ATTENTION")

    failed_checks = [
        name
        for name, status in audit.items()
        if not status
    ]

    print("\nFailed checks:")

    for check in failed_checks:
        print("  ✗", check)
```

```python
# ============================================================
# CHECKPOINT 13.5 — FIGURE 2:
# TOP THREE PREDICTIVE CONNECTIVITY FEATURES
# ============================================================

print("=" * 80)
print("CHECKPOINT 13.5 — TOP THREE PREDICTIVE FEATURES")
print("=" * 80)

top_biological_features = (
    feature_interpretation
    .sort_values(
        "mean_permutation_importance",
        ascending=False
    )["feature"]
    .head(3)
    .tolist()
)

assert len(top_biological_features) == 3

assert set(
    top_biological_features
).issubset(
    set(ml_feature_columns)
)

feature_labels = {
    "outgoing_mean_synapses":
        "Outgoing mean synaptic strength",

    "outgoing_median_synapses":
        "Outgoing median synaptic strength",

    "incoming_max_synapses":
        "Incoming maximum synaptic strength"
}

for feature in top_biological_features:

    if feature not in feature_labels:
        feature_labels[feature] = (
            feature.replace("_", " ").title()
        )

print("\nTop three features:")
for rank, feature in enumerate(
    top_biological_features,
    start=1
):
    print(
        f"{rank}. {feature}"
    )

figure2_records = []

for rank, feature in enumerate(
    top_biological_features,
    start=1
):

    fig, ax = plt.subplots(
        figsize=(5.5, 4.5)
    )

    ach_values = (
        cell_type_level_dataset
        .loc[
            cell_type_level_dataset["neurotransmitter"]
            == "acetylcholine",
            feature
        ]
        .dropna()
    )

    gaba_values = (
        cell_type_level_dataset
        .loc[
            cell_type_level_dataset["neurotransmitter"]
            == "gaba",
            feature
        ]
        .dropna()
    )

    ax.boxplot(
        [
            ach_values,
            gaba_values
        ],
        tick_labels=["ACh", "GABA"],
        showfliers=False
    )

    rng = np.random.default_rng(
        42 + rank
    )

    for xpos, values in zip(
        [1, 2],
        [ach_values, gaba_values]
    ):

        jitter = rng.normal(
            0,
            0.035,
            size=len(values)
        )

        ax.scatter(
            np.full(
                len(values),
                xpos
            ) + jitter,
            values,
            alpha=0.45,
            s=18
        )

    ax.set_xlabel(
        "Neurotransmitter identity"
    )

    ax.set_ylabel(
        "Cell-type median connectivity value"
    )

    ax.set_title(
        feature_labels[feature]
    )

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()

    png_name = (
        f"figure2_{rank}_{feature}.png"
    )

    pdf_name = (
        f"figure2_{rank}_{feature}.pdf"
    )

    png_path = os.path.join(
        PUBLICATION_FIG_DIR,
        png_name
    )

    pdf_path = os.path.join(
        PUBLICATION_FIG_DIR,
        pdf_name
    )

    fig.savefig(
        png_path,
        dpi=300,
        bbox_inches="tight"
    )

    fig.savefig(
        pdf_path,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    figure2_records.append({
        "rank": rank,
        "feature": feature,
        "label": feature_labels[feature],
        "png": png_name,
        "pdf": pdf_name
    })

figure2_manifest = pd.DataFrame(
    figure2_records
)

figure2_manifest_path = os.path.join(
    PUBLICATION_TAB_DIR,
    "figure2_manifest.csv"
)

figure2_manifest.to_csv(
    figure2_manifest_path,
    index=False
)

display(figure2_manifest)

assert len(figure2_manifest) == 3

for filename in figure2_manifest["png"]:
    assert os.path.exists(
        os.path.join(
            PUBLICATION_FIG_DIR,
            filename
        )
    )

for filename in figure2_manifest["pdf"]:
    assert os.path.exists(
        os.path.join(
            PUBLICATION_FIG_DIR,
            filename
        )
    )

print("\n✓ Three Figure 2 feature-specific plots generated.")
print("✓ Feature selection inherited directly from CP11.")
print("✓ Cell-type-level values inherited directly from CP12.")

print("\n✓ CHECKPOINT 13.5 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 13.8 — PUBLICATION FIGURE 4
# GROUPED CROSS-VALIDATION PERFORMANCE
# ============================================================

print("=" * 80)
print("CHECKPOINT 13.8 — PUBLICATION FIGURE 4")
print("=" * 80)

cv_df = grouped_cv_results.copy()

print("\nGrouped CV result shape:", cv_df.shape)
print("Columns:")
print(list(cv_df.columns))

required_cv_columns = [
    "fold",
    "model",
    "balanced_accuracy"
]

for column in required_cv_columns:
    assert column in cv_df.columns

assert len(cv_df) == 20
assert cv_df["fold"].nunique() == 5
assert cv_df["model"].nunique() == 4

# ------------------------------------------------------------
# Calculate display summaries ONLY from existing CV results.
# No model fitting.
# ------------------------------------------------------------

cv_summary = (
    cv_df
    .groupby("model")
    ["balanced_accuracy"]
    .agg(
        mean="mean",
        sd="std"
    )
    .reset_index()
)

preferred_order = [
    "Dummy",
    "Logistic Regression",
    "Random Forest",
    "XGBoost"
]

if set(preferred_order).issubset(
    set(cv_summary["model"])
):

    cv_summary["model"] = pd.Categorical(
        cv_summary["model"],
        categories=preferred_order,
        ordered=True
    )

    cv_summary = (
        cv_summary
        .sort_values("model")
        .reset_index(drop=True)
    )

display(cv_summary)

# ------------------------------------------------------------
# Publication figure
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7.5, 5)
)

x = np.arange(
    len(cv_summary)
)

ax.bar(
    x,
    cv_summary["mean"],
    yerr=cv_summary["sd"],
    capsize=4
)

ax.set_xticks(x)

ax.set_xticklabels(
    cv_summary["model"],
    rotation=20,
    ha="right"
)

ax.set_ylabel(
    "Balanced accuracy"
)

ax.set_xlabel(
    "Model"
)

ax.set_title(
    "Cell-Type-Grouped Cross-Validation Performance"
)

ax.set_ylim(
    0,
    min(
        1.0,
        max(
            1.0,
            (
                cv_summary["mean"]
                + cv_summary["sd"]
            ).max() * 1.10
        )
    )
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

figure4_png = os.path.join(
    PUBLICATION_FIG_DIR,
    "figure4_grouped_cv_performance.png"
)

figure4_pdf = os.path.join(
    PUBLICATION_FIG_DIR,
    "figure4_grouped_cv_performance.pdf"
)

fig.savefig(
    figure4_png,
    dpi=300,
    bbox_inches="tight"
)

fig.savefig(
    figure4_pdf,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# ------------------------------------------------------------
# Save display summary
# ------------------------------------------------------------

figure4_summary_path = os.path.join(
    PUBLICATION_TAB_DIR,
    "figure4_grouped_cv_summary.csv"
)

cv_summary.to_csv(
    figure4_summary_path,
    index=False
)

assert os.path.exists(figure4_png)
assert os.path.getsize(figure4_png) > 0

assert os.path.exists(figure4_pdf)
assert os.path.getsize(figure4_pdf) > 0

assert os.path.exists(
    figure4_summary_path
)

print("\nSaved:")
print(" ", figure4_png)
print(" ", figure4_pdf)
print(" ", figure4_summary_path)

print("\n✓ CHECKPOINT 13.8 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 13.15 — PUBLICATION FIGURE FILE AUDIT
# ============================================================

print("=" * 80)
print("CHECKPOINT 13.15 — PUBLICATION FIGURE FILE AUDIT")
print("=" * 80)

# ------------------------------------------------------------
# Expected Figure 1
# ------------------------------------------------------------

figure_files = [
    "figure1_dataset_composition.png",
    "figure1_dataset_composition.pdf"
]

# ------------------------------------------------------------
# Figure 2
# Use the ACTUAL filenames generated by CP13.5.
# ------------------------------------------------------------

assert len(figure2_manifest) == 3

for filename in figure2_manifest["png"]:
    figure_files.append(filename)

for filename in figure2_manifest["pdf"]:
    figure_files.append(filename)

# ------------------------------------------------------------
# Figures 3–6
# ------------------------------------------------------------

figure_files.extend([
    "figure3_pca_connectivity_architecture.png",
    "figure3_pca_connectivity_architecture.pdf",

    "figure4_grouped_cv_performance.png",
    "figure4_grouped_cv_performance.pdf",

    "figure5_permutation_feature_importance.png",
    "figure5_permutation_feature_importance.pdf",

    "figure6_feature_family_contribution.png",
    "figure6_feature_family_contribution.pdf"
])

# ------------------------------------------------------------
# Audit every expected file
# ------------------------------------------------------------

figure_audit_rows = []

for filename in figure_files:

    filepath = os.path.join(
        PUBLICATION_FIG_DIR,
        filename
    )

    exists = os.path.exists(filepath)

    size_bytes = (
        os.path.getsize(filepath)
        if exists
        else 0
    )

    passed = (
        exists
        and size_bytes > 0
    )

    figure_audit_rows.append({
        "file": filename,
        "exists": exists,
        "size_bytes": size_bytes,
        "passed": passed
    })

figure_audit_df = pd.DataFrame(
    figure_audit_rows
)

display(figure_audit_df)

# ------------------------------------------------------------
# Failure report
# ------------------------------------------------------------

failed_figures = figure_audit_df[
    ~figure_audit_df["passed"]
]

if len(failed_figures) > 0:

    print(
        "\n✗ Missing or empty publication figure files:"
    )

    display(
        failed_figures
    )

    raise RuntimeError(
        "Publication figure audit failed."
    )

# ------------------------------------------------------------
# Final result
# ------------------------------------------------------------

print(
    "\n✓ Figure 1 verified."
)

print(
    "✓ Three Figure 2 feature-specific outputs verified."
)

print(
    "✓ Figure 3 verified."
)

print(
    "✓ Figure 4 verified."
)

print(
    "✓ Figure 5 verified."
)

print(
    "✓ Figure 6 verified."
)

print(
    f"\nTotal publication figure files verified: "
    f"{len(figure_audit_df)}"
)

print(
    "\n✓ All expected publication figure files "
    "exist and are non-empty."
)

print("\n✓ CHECKPOINT 13.15 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 14.1 — INTERACTIVE MODEL SETUP
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.1 — INTERACTIVE MODEL SETUP")
print("=" * 80)

# ------------------------------------------------------------
# Validate required objects from previous checkpoints
# ------------------------------------------------------------

required_gui_objects = [
    "final_ml_dataset",
    "ml_feature_columns",
    "publication_dataset",
    "grouped_cv_publication_table"
]

missing_gui_objects = [
    name
    for name in required_gui_objects
    if name not in globals()
]

if missing_gui_objects:

    print("\n✗ Missing required objects:")

    for name in missing_gui_objects:
        print("  -", name)

    raise RuntimeError(
        "Required CP5/CP9/CP13 objects are missing."
    )

# ------------------------------------------------------------
# Freeze GUI training data
# ------------------------------------------------------------

gui_dataset = final_ml_dataset.copy()

gui_features = gui_dataset[
    ml_feature_columns
].copy()

gui_labels = gui_dataset[
    "neurotransmitter"
].copy()

# ------------------------------------------------------------
# Integrity checks
# ------------------------------------------------------------

assert gui_dataset.shape[0] == 165
assert len(ml_feature_columns) == 16

assert gui_features.shape == (
    165,
    16
)

assert gui_features.isna().sum().sum() == 0
assert gui_labels.isna().sum() == 0

assert set(
    gui_labels.unique()
) == {
    "acetylcholine",
    "gaba"
}

print("\nGUI training dataset:")
print("  Neurons:", gui_dataset.shape[0])
print("  Features:", gui_features.shape[1])
print(
    "  ACh:",
    (gui_labels == "acetylcholine").sum()
)
print(
    "  GABA:",
    (gui_labels == "gaba").sum()
)

print("\n✓ GUI will use the same 16 connectivity features")
print("  used throughout the validated analysis.")

print("\n✓ CHECKPOINT 14.1 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 14.2 — FINAL DEMONSTRATION MODEL
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.2 — FINAL DEMONSTRATION MODEL")
print("=" * 80)

# ------------------------------------------------------------
# Final Logistic Regression model
#
# This model is trained on all 165 audited neurons ONLY for
# interactive demonstration/prediction.
#
# Its performance is NOT reported as independent validation.
# The validated performance remains the CP9 grouped-CV result.
# ------------------------------------------------------------

gui_scaler = StandardScaler()

X_gui_scaled = gui_scaler.fit_transform(
    gui_features
)

gui_model = LogisticRegression(
    class_weight="balanced",
    max_iter=2000,
    random_state=42
)

gui_model.fit(
    X_gui_scaled,
    gui_labels
)

# ------------------------------------------------------------
# Model integrity
# ------------------------------------------------------------

assert hasattr(
    gui_model,
    "predict"
)

assert hasattr(
    gui_model,
    "predict_proba"
)

assert len(
    gui_model.coef_[0]
) == 16

assert len(
    gui_model.classes_
) == 2

print("\nFinal demonstration model:")
print("  Model: Logistic Regression")
print("  Training observations:", len(gui_dataset))
print("  Features:", len(ml_feature_columns))
print("  Classes:", list(gui_model.classes_))

print(
    "\nImportant:"
)

print(
    "  The model above is a full-dataset demonstration model."
)

print(
    "  Its performance is NOT treated as a held-out test result."
)

print(
    "  Generalization is assessed using the grouped-CV results "
    "from CP9."
)

print("\n✓ CHECKPOINT 14.2 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 14.3 — INTERACTIVE PREDICTION FUNCTION
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.3 — INTERACTIVE PREDICTION FUNCTION")
print("=" * 80)

def predict_neurotransmitter(
    feature_values
):
    """
    Predict neurotransmitter identity from connectivity features.

    Parameters
    ----------
    feature_values : dict
        Dictionary mapping each connectivity feature name to a
        numeric value.

    Returns
    -------
    dict
        Prediction, class probabilities, and input feature values.
    """

    # --------------------------------------------------------
    # Validate input features
    # --------------------------------------------------------

    missing_features = [
        feature
        for feature in ml_feature_columns
        if feature not in feature_values
    ]

    if missing_features:

        raise ValueError(
            "Missing connectivity features: "
            + ", ".join(missing_features)
        )

    # --------------------------------------------------------
    # Construct feature vector in exact training order
    # --------------------------------------------------------

    input_df = pd.DataFrame(
        [
            {
                feature: float(
                    feature_values[feature]
                )
                for feature in ml_feature_columns
            }
        ]
    )

    # --------------------------------------------------------
    # Scale using the scaler fitted on the full training dataset
    # --------------------------------------------------------

    input_scaled = gui_scaler.transform(
        input_df
    )

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    prediction = gui_model.predict(
        input_scaled
    )[0]

    probabilities = gui_model.predict_proba(
        input_scaled
    )[0]

    probability_dict = {
        class_name: float(probability)
        for class_name, probability in zip(
            gui_model.classes_,
            probabilities
        )
    }

    # --------------------------------------------------------
    # Human-readable output
    # --------------------------------------------------------

    readable_prediction = {
        "acetylcholine":
            "Acetylcholine (ACh)",

        "gaba":
            "GABA"
    }

    result = {
        "predicted_neurotransmitter":
            readable_prediction.get(
                prediction,
                prediction
            ),

        "acetylcholine_probability":
            probability_dict.get(
                "acetylcholine",
                np.nan
            ),

        "gaba_probability":
            probability_dict.get(
                "gaba",
                np.nan
            ),

        "input_features":
            input_df.iloc[0].to_dict()
    }

    return result


print("\n✓ Prediction function created.")

print(
    "\nExpected input features:"
)

for feature in ml_feature_columns:
    print("  -", feature)

print("\n✓ CHECKPOINT 14.3 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 14.4 — PREDICTION ENGINE TEST
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.4 — PREDICTION ENGINE TEST")
print("=" * 80)

# Use the first audited neuron as a technical test input.
# This is NOT an independent performance evaluation.

test_row = gui_dataset.iloc[0]

test_input = {
    feature: float(
        test_row[feature]
    )
    for feature in ml_feature_columns
}

test_result = predict_neurotransmitter(
    test_input
)

print("\nTest neuron:")
print(
    "  bodyId:",
    test_row["bodyId"]
)

print(
    "  Actual neurotransmitter:",
    test_row["neurotransmitter"]
)

print("\nModel output:")
print(
    "  Predicted:",
    test_result[
        "predicted_neurotransmitter"
    ]
)

print(
    "  ACh probability:",
    f"{test_result['acetylcholine_probability']:.3f}"
)

print(
    "  GABA probability:",
    f"{test_result['gaba_probability']:.3f}"
)

# ------------------------------------------------------------
# Integrity checks
# ------------------------------------------------------------

assert (
    "predicted_neurotransmitter"
    in test_result
)

assert (
    0 <=
    test_result[
        "acetylcholine_probability"
    ] <= 1
)

assert (
    0 <=
    test_result[
        "gaba_probability"
    ] <= 1
)

assert np.isclose(
    test_result[
        "acetylcholine_probability"
    ]
    +
    test_result[
        "gaba_probability"
    ],
    1.0,
    atol=1e-6
)

print("\n✓ Prediction engine test passed.")
print("\n✓ CHECKPOINT 14.4 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 14.5 — INTERACTIVE NEUROTRANSMITTER PREDICTION GUI
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.5 — INTERACTIVE NEUROTRANSMITTER PREDICTION GUI")
print("=" * 80)

# ------------------------------------------------------------
# Install Gradio if necessary
# ------------------------------------------------------------

try:
    import gradio as gr
    print("\n✓ Gradio is available.")

except ImportError:

    print("\nInstalling Gradio...")

    !pip -q install gradio

    import gradio as gr

    print("✓ Gradio installed.")

# ------------------------------------------------------------
# Human-readable feature labels
# ------------------------------------------------------------

gui_feature_labels = {
    "outgoing_edges":
        "Outgoing edges",

    "outgoing_synapses":
        "Outgoing synapses",

    "outgoing_mean_synapses":
        "Outgoing mean synaptic strength",

    "outgoing_median_synapses":
        "Outgoing median synaptic strength",

    "outgoing_max_synapses":
        "Outgoing maximum synaptic strength",

    "outgoing_synapse_sd":
        "Outgoing synaptic strength SD",

    "incoming_edges":
        "Incoming edges",

    "incoming_synapses":
        "Incoming synapses",

    "incoming_mean_synapses":
        "Incoming mean synaptic strength",

    "incoming_median_synapses":
        "Incoming median synaptic strength",

    "incoming_max_synapses":
        "Incoming maximum synaptic strength",

    "incoming_synapse_sd":
        "Incoming synaptic strength SD",

    "total_edges":
        "Total edges",

    "total_synapses":
        "Total synapses",

    "outgoing_incoming_edge_ratio":
        "Outgoing/incoming edge ratio",

    "outgoing_incoming_synapse_ratio":
        "Outgoing/incoming synapse ratio"
}

# ------------------------------------------------------------
# GUI prediction wrapper
# ------------------------------------------------------------

def gui_predict(*values):

    feature_values = {
        feature: float(value)
        for feature, value in zip(
            ml_feature_columns,
            values
        )
    }

    result = predict_neurotransmitter(
        feature_values
    )

    prediction = result[
        "predicted_neurotransmitter"
    ]

    ach_probability = (
        result[
            "acetylcholine_probability"
        ]
        * 100
    )

    gaba_probability = (
        result[
            "gaba_probability"
        ]
        * 100
    )

    return (
        prediction,
        f"{ach_probability:.2f}%",
        f"{gaba_probability:.2f}%"
    )

# ------------------------------------------------------------
# Default values
#
# Use median connectivity values from the audited dataset.
# This provides a biologically realistic starting point rather
# than arbitrary numbers.
# ------------------------------------------------------------

gui_default_values = [
    float(
        gui_dataset[feature].median()
    )
    for feature in ml_feature_columns
]

# ------------------------------------------------------------
# Build interface
# ------------------------------------------------------------

input_components = []

for feature, default_value in zip(
    ml_feature_columns,
    gui_default_values
):

    component = gr.Number(
        label=gui_feature_labels.get(
            feature,
            feature.replace(
                "_",
                " "
            ).title()
        ),
        value=default_value,
        precision=4
    )

    input_components.append(
        component
    )

# ------------------------------------------------------------
# Interface description
# ------------------------------------------------------------

description = """
### Connectivity-Based Neurotransmitter Prediction

Enter the 16 connectivity features of an olfactory neuron to obtain
a model-based prediction of its neurotransmitter identity.

The model was trained on the final audited Drosophila olfactory
dataset using logistic regression with standardized connectivity
features.

**Important:** this interface is a computational demonstration.
The displayed probabilities come from a model trained on the full
dataset and should not be interpreted as independent validation
performance. Generalization should be assessed using the
cell-type-grouped cross-validation results reported in this study.
"""

# ------------------------------------------------------------
# Create Gradio application
# ------------------------------------------------------------

gui_app = gr.Interface(
    fn=gui_predict,
    inputs=input_components,

    outputs=[
        gr.Textbox(
            label="Predicted neurotransmitter"
        ),

        gr.Textbox(
            label="Acetylcholine probability"
        ),

        gr.Textbox(
            label="GABA probability"
        )
    ],

    title=(
        "Drosophila Olfactory "
        "Neurotransmitter Predictor"
    ),

    description=description,

    flagging_mode="never"
)

print("\n✓ GUI interface constructed.")
print("\n✓ 16 connectivity inputs configured.")
print("✓ Logistic Regression prediction backend connected.")
print("✓ ACh/GABA probability outputs configured.")

print("\n✓ CHECKPOINT 14.5 READY")
```

```python
# ============================================================
# CHECKPOINT 14.5 — LAUNCH GUI
# ============================================================

print("=" * 80)
print("LAUNCHING INTERACTIVE GUI")
print("=" * 80)

gui_app.launch(
    share=True,
    debug=False
)
```

```python
# ============================================================
# CHECKPOINT 14.6 — GUI/PREDICTION CONSISTENCY VALIDATION
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.6 — GUI/PREDICTION CONSISTENCY VALIDATION")
print("=" * 80)

# ------------------------------------------------------------
# Select known neurons from the audited dataset
# ------------------------------------------------------------

validation_indices = [
    0,
    20,
    50,
    100,
    140,
    164
]

gui_validation_rows = []

for idx in validation_indices:

    row = gui_dataset.iloc[idx]

    feature_values = {
        feature: float(
            row[feature]
        )
        for feature in ml_feature_columns
    }

    # Direct prediction
    direct_result = predict_neurotransmitter(
        feature_values
    )

    direct_prediction = (
        direct_result[
            "predicted_neurotransmitter"
        ]
    )

    direct_ach_probability = (
        direct_result[
            "acetylcholine_probability"
        ]
    )

    direct_gaba_probability = (
        direct_result[
            "gaba_probability"
        ]
    )

    # GUI wrapper prediction
    gui_output = gui_predict(
        *[
            feature_values[feature]
            for feature in ml_feature_columns
        ]
    )

    gui_prediction = gui_output[0]

    gui_ach_probability = (
        float(
            gui_output[1].replace(
                "%",
                ""
            )
        ) / 100
    )

    gui_gaba_probability = (
        float(
            gui_output[2].replace(
                "%",
                ""
            )
        ) / 100
    )

    # --------------------------------------------------------
    # Consistency checks
    # --------------------------------------------------------

    prediction_match = (
        direct_prediction
        == gui_prediction
    )

    ach_probability_match = np.isclose(
        direct_ach_probability,
        gui_ach_probability,
        atol=0.0001
    )

    gaba_probability_match = np.isclose(
        direct_gaba_probability,
        gui_gaba_probability,
        atol=0.0001
    )

    gui_validation_rows.append({

        "bodyId":
            row["bodyId"],

        "actual_neurotransmitter":
            row["neurotransmitter"],

        "direct_prediction":
            direct_prediction,

        "GUI_prediction":
            gui_prediction,

        "direct_ACh_probability":
            direct_ach_probability,

        "GUI_ACh_probability":
            gui_ach_probability,

        "direct_GABA_probability":
            direct_gaba_probability,

        "GUI_GABA_probability":
            gui_gaba_probability,

        "prediction_match":
            prediction_match,

        "ACh_probability_match":
            ach_probability_match,

        "GABA_probability_match":
            gaba_probability_match
    })

gui_validation_df = pd.DataFrame(
    gui_validation_rows
)

display(
    gui_validation_df
)

# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

assert (
    gui_validation_df[
        "prediction_match"
    ].all()
)

assert (
    gui_validation_df[
        "ACh_probability_match"
    ].all()
)

assert (
    gui_validation_df[
        "GABA_probability_match"
    ].all()
)

print("\n✓ GUI predictions match the underlying prediction function.")
print("✓ ACh probabilities match.")
print("✓ GABA probabilities match.")
print(
    f"✓ {len(gui_validation_df)} known neurons tested."
)

print("\n✓ CHECKPOINT 14.6 COMPLETE")
```

```python
# ============================================================
# CHECKPOINT 14.8 — FINAL DOCUMENTED INTERACTIVE GUI
# ============================================================

print("=" * 80)
print("CHECKPOINT 14.8 — FINAL DOCUMENTED INTERACTIVE GUI")
print("=" * 80)

# ------------------------------------------------------------
# Rebuild GUI description with scientific documentation
# ------------------------------------------------------------

final_gui_description = """
## What does this tool do?

This interactive tool predicts whether an examined Drosophila
olfactory neuron is more likely to be classified as **Acetylcholine
(ACh)** or **GABA** from its connectivity architecture.

The model uses the same **16 connectivity features** analysed in this
study.

### Dataset

- 165 olfactory neurons
- 65 canonical cell types
- 16 connectivity features
- 2 neurotransmitter classes: ACh and GABA

### Model

**Logistic Regression** with standardized connectivity features.

The interactive model was trained on the complete audited dataset
for demonstration purposes.

### Important scientific limitation

The probabilities shown here are **model probabilities**, not
experimentally measured neurotransmitter concentrations.

The model used for this demonstration was trained on the full
dataset. Therefore, its displayed predictions must **not** be
interpreted as independent test performance.

Model generalization was assessed separately using
**cell-type-grouped five-fold cross-validation**.

The analysis supports the conclusion that connectivity architecture
contains predictive information about neurotransmitter identity within
the examined olfactory population. It does **not** establish that
connectivity mechanistically determines neurotransmitter identity or
that the result universally generalizes to other neurons, brain
regions, or species.

---

### How to use

Enter the 16 connectivity measurements below and click **Predict**.

The output gives:

1. Predicted neurotransmitter
2. Acetylcholine probability
3. GABA probability

For demonstration purposes, the input fields initially contain
median values from the analysed dataset.
"""

# ------------------------------------------------------------
# Rebuild input components
# ------------------------------------------------------------

final_gui_inputs = []

for feature, default_value in zip(
    ml_feature_columns,
    gui_default_values
):

    final_gui_inputs.append(
        gr.Number(
            label=gui_feature_labels.get(
                feature,
                feature.replace(
                    "_",
                    " "
                ).title()
            ),
            value=default_value,
            precision=4
        )
    )

# ------------------------------------------------------------
# Final documented GUI
# ------------------------------------------------------------

final_gui_app = gr.Interface(

    fn=gui_predict,

    inputs=final_gui_inputs,

    outputs=[
        gr.Textbox(
            label="Predicted neurotransmitter"
        ),

        gr.Textbox(
            label="Acetylcholine probability"
        ),

        gr.Textbox(
            label="GABA probability"
        )
    ],

    title=(
        "Drosophila Olfactory "
        "Neurotransmitter Predictor"
    ),

    description=final_gui_description,

    flagging_mode="never"
)

# ------------------------------------------------------------
# Validate interface
# ------------------------------------------------------------

assert len(final_gui_inputs) == 16

assert (
    final_gui_app is not None
)

print("\n✓ Final GUI constructed.")
print("✓ 16 connectivity inputs configured.")
print("✓ Prediction backend connected.")
print("✓ Scientific documentation embedded.")
print("✓ Validation limitations disclosed.")

print("\n✓ CHECKPOINT 14.8 COMPLETE")
```

```python
# =============================================================================
# CHECKPOINT 15.5 — GROUPED CROSS-VALIDATION INTEGRITY AUDIT
# =============================================================================

print("=" * 80)
print("CHECKPOINT 15.5 — GROUPED CROSS-VALIDATION INTEGRITY AUDIT")
print("=" * 80)

cv_df = grouped_cv_results.copy()

print("Grouped CV result shape:", cv_df.shape)

print("\nModels:")
print(cv_df["model"].unique())

print("\nNumber of folds:")
print(cv_df.groupby("model").size())

expected_models = {
    "Dummy",
    "Logistic Regression",
    "Random Forest",
    "XGBoost",
}

actual_models = set(cv_df["model"].unique())

print("\nExpected models:", expected_models)
print("Actual models:", actual_models)

model_set_match = actual_models == expected_models
fold_count_match = all(
    cv_df.groupby("model").size() == 5
)

required_metrics = [
    "accuracy",
    "balanced_accuracy",
    "macro_F1",
    "ACh_recall",
    "GABA_recall",
]

missing_metrics = [
    m for m in required_metrics
    if m not in cv_df.columns
]

print("\nMissing required metrics:", missing_metrics)

metric_range_errors = {}

for metric in required_metrics:
    if metric in cv_df.columns:
        bad = cv_df[
            (cv_df[metric] < 0) |
            (cv_df[metric] > 1)
        ]

        metric_range_errors[metric] = len(bad)

print("\nMetric range violations:")
print(metric_range_errors)

cp15_5_pass = (
    cv_df.shape[0] == 20
    and model_set_match
    and fold_count_match
    and len(missing_metrics) == 0
    and all(v == 0 for v in metric_range_errors.values())
)

print("\nSTATUS:", "PASSED" if cp15_5_pass else "REVIEW")

print("=" * 80)
```

```python
# =============================================================================
# CHECKPOINT 15.8 — FEATURE INTERPRETATION CONSISTENCY AUDIT
# =============================================================================

print("=" * 80)
print("CHECKPOINT 15.8 — FEATURE INTERPRETATION CONSISTENCY AUDIT")
print("=" * 80)

fi = feature_interpretation.copy()

required_fi_columns = [
    "feature",
    "mean_coefficient",
    "sd_coefficient",
    "mean_absolute_coefficient",
    "mean_permutation_importance",
    "sd_permutation_importance",
    "median_permutation_importance",
    "permutation_rank",
    "coefficient_rank",
    "feature_family",
]

missing_fi_columns = [
    c for c in required_fi_columns
    if c not in fi.columns
]

print("Feature interpretation rows:", len(fi))
print("Missing required columns:", missing_fi_columns)

if len(missing_fi_columns) == 0:

    # Check exactly one interpretation row per feature
    feature_count_match = (
        len(fi) == len(ml_feature_columns)
        and set(fi["feature"]) == set(ml_feature_columns)
    )

    print(
        "Feature set matches the 16 ML features:",
        feature_count_match
    )

    # Rank consistency
    rank_values = fi["permutation_rank"].dropna()

    rank_integrity = (
        len(rank_values) == 16
        and sorted(rank_values.astype(int).tolist()) == list(range(1, 17))
    )

    print("Permutation ranks 1–16 complete:", rank_integrity)

    # Independent ranking from stored importance values
    calculated_rank_order = (
        fi.sort_values(
            "mean_permutation_importance",
            ascending=False
        )["feature"]
        .tolist()
    )

    stored_rank_order = (
        fi.sort_values(
            "permutation_rank",
            ascending=True
        )["feature"]
        .tolist()
    )

    rank_order_match = calculated_rank_order == stored_rank_order

    print(
        "Stored permutation ranks match importance ordering:",
        rank_order_match
    )

    # Top three
    observed_top_three = calculated_rank_order[:3]

    expected_top_three = [
        "outgoing_mean_synapses",
        "incoming_max_synapses",
        "outgoing_median_synapses",
    ]

    print("\nTop three features:")
    for i, feature in enumerate(observed_top_three, start=1):
        print(f"{i}. {feature}")

    top_three_match = observed_top_three == expected_top_three

else:
    feature_count_match = False
    rank_integrity = False
    rank_order_match = False
    top_three_match = False

cp15_8_pass = (
    len(missing_fi_columns) == 0
    and feature_count_match
    and rank_integrity
    and rank_order_match
    and top_three_match
)

print("\nSTATUS:", "PASSED" if cp15_8_pass else "REVIEW")

print("=" * 80)
```

```python
# =============================================================================
# CHECKPOINT 15.9 — BIOLOGICAL INTERPRETATION AUDIT
# =============================================================================

print("=" * 80)
print("CHECKPOINT 15.9 — BIOLOGICAL INTERPRETATION AUDIT")
print("=" * 80)

bio = biological_interpretation_table.copy()

required_bio_columns = [
    "feature",
    "mean_permutation_importance",
    "permutation_rank",
    "absolute_Cliffs_delta",
    "positive_importance_folds",
    "sign_consistency",
    "connectivity_dimension",
    "interpretation_category",
]

missing_bio_columns = [
    c for c in required_bio_columns
    if c not in bio.columns
]

print("Biological interpretation rows:", len(bio))
print("Missing required columns:", missing_bio_columns)

if len(missing_bio_columns) == 0:

    bio_feature_match = (
        set(bio["feature"]) == set(fi["feature"])
    )

    print(
        "Biological interpretation covers all 16 features:",
        bio_feature_match
    )

    # Check rank consistency
    bio_rank_order = (
        bio.sort_values(
            "permutation_rank",
            ascending=True
        )["feature"]
        .tolist()
    )

    fi_rank_order = (
        fi.sort_values(
            "permutation_rank",
            ascending=True
        )["feature"]
        .tolist()
    )

    rank_consistency = bio_rank_order == fi_rank_order

    print(
        "Biological interpretation ranks match feature interpretation:",
        rank_consistency
    )

    print("\nConnectivity dimensions:")
    display(bio["connectivity_dimension"].value_counts())

    print("\nInterpretation categories:")
    display(bio["interpretation_category"].value_counts())

else:
    bio_feature_match = False
    rank_consistency = False

print("\nScientific interpretation constraint:")
print(
    "The observed relationships are interpreted as associative "
    "predictive relationships and not as causal determinants of "
    "neurotransmitter identity."
)

cp15_9_pass = (
    len(bio) == 16
    and len(missing_bio_columns) == 0
    and bio_feature_match
    and rank_consistency
)

print("\nSTATUS:", "PASSED" if cp15_9_pass else "REVIEW")

print("=" * 80)
```

```python
# =============================================================================
# CHECKPOINT 15.10 — PCA EXPLORATORY ANALYSIS AUDIT
# =============================================================================

print("=" * 80)
print("CHECKPOINT 15.10 — PCA EXPLORATORY ANALYSIS AUDIT")
print("=" * 80)

# -------------------------------------------------------------------------
# Inspect the existing explained_variance object
# -------------------------------------------------------------------------

print("explained_variance object type:", type(explained_variance))

if isinstance(explained_variance, pd.DataFrame):

    print("\nColumns:")
    print(explained_variance.columns.tolist())

    # Find the numeric explained-variance column
    numeric_columns = explained_variance.select_dtypes(
        include=[np.number]
    ).columns.tolist()

    print("\nNumeric columns:", numeric_columns)

    if len(numeric_columns) == 1:
        ev = explained_variance[numeric_columns[0]].to_numpy(
            dtype=float
        )

    elif "explained_variance_ratio" in explained_variance.columns:
        ev = explained_variance[
            "explained_variance_ratio"
        ].to_numpy(dtype=float)

    elif "explained_variance" in explained_variance.columns:
        ev = explained_variance[
            "explained_variance"
        ].to_numpy(dtype=float)

    else:
        raise ValueError(
            "Could not identify the numeric explained-variance column."
        )

else:
    ev = np.asarray(explained_variance, dtype=float)

# -------------------------------------------------------------------------
# PCA audit
# -------------------------------------------------------------------------

print("\nNumber of PCA components:", len(ev))

for i, value in enumerate(ev[:5], start=1):
    print(f"PC{i}: {value * 100:.2f}%")

pc1 = ev[0] * 100
pc2 = ev[1] * 100
pc12 = pc1 + pc2

print(f"\nPC1 + PC2: {pc12:.2f}%")

# Established CP6 values
pc1_match = np.isclose(
    pc1,
    68.02,
    atol=0.10
)

pc2_match = np.isclose(
    pc2,
    14.48,
    atol=0.10
)

print(f"PC1 expected ≈ 68.02%: {pc1_match}")
print(f"PC2 expected ≈ 14.48%: {pc2_match}")

# Mathematical sanity check
variance_sum_valid = (
    np.all(ev >= 0)
    and np.all(ev <= 1)
    and np.isclose(ev.sum(), 1.0, atol=1e-6)
)

print(
    "Explained variance values mathematically valid:",
    variance_sum_valid
)

# -------------------------------------------------------------------------
# Scientific interpretation
# -------------------------------------------------------------------------

print("\nInterpretation constraint:")
print(
    "PCA is exploratory and unsupervised. It is not treated as "
    "evidence of predictive performance."
)

cp15_10_pass = (
    len(ev) > 1
    and variance_sum_valid
    and pc1_match
    and pc2_match
)

print("\nSTATUS:", "PASSED" if cp15_10_pass else "REVIEW")

print("=" * 80)
```

```python
# ============================================================
# CHECKPOINT 16
# FINAL GITHUB PACKAGING + VALIDATION + API PUSH
#
# Repository:
# https://github.com/utkarshitadhawan/wiring-to-molecular-identity
# ============================================================


import json
import shutil
import hashlib
import base64
import getpass
import requests


# ============================================================
# 16.1 — PROJECT ROOT
# ============================================================

project_root = Path.cwd()

print("=" * 70)
print("CHECKPOINT 16 — FINAL GITHUB PACKAGING")
print("=" * 70)
print(f"Project root: {project_root}")

assert project_root.exists()


# ============================================================
# 16.2 — SETTINGS
# ============================================================

GITHUB_OWNER = "utkarshitadhawan"
GITHUB_REPO = "wiring-to-molecular-identity"
GITHUB_BRANCH = "main"

GITHUB_API = "https://api.github.com"

SOURCE_DATASET = (
    project_root
    / "data"
    / "processed"
    / "final_olfactory_neurotransmitter_ml_dataset.csv"
)

RESULTS_DIR = project_root / "results"

PUBLICATION_FIG_DIR = (
    RESULTS_DIR / "figures" / "publication"
)

PUBLICATION_TAB_DIR = (
    RESULTS_DIR / "tables" / "publication"
)

CURRENT_NOTEBOOK = (
    project_root / "wiring_to_mol_id.ipynb"
)

PACKAGE_ROOT = (
    project_root
    / "github_package"
    / GITHUB_REPO
)

PACKAGE_DATA_DIR = PACKAGE_ROOT / "data"
PACKAGE_NOTEBOOK_DIR = PACKAGE_ROOT / "notebook"
PACKAGE_RESULTS_DIR = PACKAGE_ROOT / "results"
PACKAGE_FIGURES_DIR = PACKAGE_RESULTS_DIR / "figures"
PACKAGE_TABLES_DIR = PACKAGE_RESULTS_DIR / "tables"


assert SOURCE_DATASET.exists(), (
    f"Missing dataset:\n{SOURCE_DATASET}"
)

assert RESULTS_DIR.exists()

assert PUBLICATION_FIG_DIR.exists()

assert PUBLICATION_TAB_DIR.exists()

assert CURRENT_NOTEBOOK.exists()

print("✓ Project structure found")


# ============================================================
# 16.3 — NOTEBOOK VALIDATION
# ============================================================

print("\n[16.3] Validating final notebook...")

with open(
    CURRENT_NOTEBOOK,
    "r",
    encoding="utf-8"
) as f:
    notebook_json = json.load(f)

assert "cells" in notebook_json
assert notebook_json.get("nbformat") is not None

notebook_text = "\n".join(
    "".join(cell.get("source", []))
    for cell in notebook_json["cells"]
)

assert "CHECKPOINT 14.11" in notebook_text
assert "CHECKPOINT 15.16" in notebook_text

print("✓ Notebook JSON valid")
print(
    f"✓ Notebook cells: "
    f"{len(notebook_json['cells'])}"
)
print("✓ Checkpoint 14.11 present")
print("✓ Checkpoint 15.16 present")


# ============================================================
# 16.4 — DATASET VALIDATION
# ============================================================

print("\n[16.4] Validating final ML dataset...")

dataset = pd.read_csv(SOURCE_DATASET)

expected_features = [
    "outgoing_edges",
    "outgoing_synapses",
    "outgoing_mean_synapses",
    "outgoing_median_synapses",
    "outgoing_max_synapses",
    "outgoing_synapse_sd",
    "incoming_edges",
    "incoming_synapses",
    "incoming_mean_synapses",
    "incoming_median_synapses",
    "incoming_max_synapses",
    "incoming_synapse_sd",
    "total_edges",
    "total_synapses",
    "outgoing_incoming_edge_ratio",
    "outgoing_incoming_synapse_ratio",
]

required_columns = [
    "bodyId",
    "cell_type",
    "neurotransmitter",
] + expected_features

missing_columns = [
    c for c in required_columns
    if c not in dataset.columns
]

assert not missing_columns, (
    f"Missing columns: {missing_columns}"
)

assert dataset.shape == (165, 20)

assert dataset["bodyId"].nunique() == 165

assert dataset["cell_type"].nunique() == 65

assert set(
    dataset["neurotransmitter"].dropna().unique()
) == {
    "acetylcholine",
    "gaba",
}

assert not dataset[
    expected_features
].isna().any().any()

print(f"✓ Dataset shape: {dataset.shape}")
print("✓ 165 neurons")
print("✓ 65 canonical cell types")
print("✓ 16 connectivity features")
print("✓ Acetylcholine/GABA labels")
print("✓ No missing feature values")


# ============================================================
# 16.5 — PUBLICATION OUTPUT VALIDATION
# ============================================================

print("\n[16.5] Validating publication outputs...")

required_figures = [
    "figure1_dataset_composition.pdf",
    "figure1_dataset_composition.png",
    "figure2_1_outgoing_mean_synapses.pdf",
    "figure2_1_outgoing_mean_synapses.png",
    "figure2_2_outgoing_median_synapses.pdf",
    "figure2_2_outgoing_median_synapses.png",
    "figure2_3_incoming_max_synapses.pdf",
    "figure2_3_incoming_max_synapses.png",
    "figure3_pca_connectivity_architecture.pdf",
    "figure3_pca_connectivity_architecture.png",
    "figure4_grouped_cv_performance.pdf",
    "figure4_grouped_cv_performance.png",
    "figure5_permutation_feature_importance.pdf",
    "figure5_permutation_feature_importance.png",
    "figure6_feature_family_contribution.pdf",
    "figure6_feature_family_contribution.png",
]

required_tables = [
    "table1_dataset_composition.csv",
    "table2_grouped_cross_validation_model_comparison.csv",
    "table3_feature_level_predictive_interpretation.csv",
    "table4_biological_interpretation_of_connectivity_features.csv",
    "table5_feature_family_contribution.csv",
    "figure2_manifest.csv",
    "figure3_pca_coordinates.csv",
    "figure4_grouped_cv_summary.csv",
    "publication_manifest.csv",
    "scientific_claim_audit.csv",
]

missing_figures = [
    x for x in required_figures
    if not (PUBLICATION_FIG_DIR / x).exists()
]

missing_tables = [
    x for x in required_tables
    if not (PUBLICATION_TAB_DIR / x).exists()
]

assert not missing_figures, (
    f"Missing figures: {missing_figures}"
)

assert not missing_tables, (
    f"Missing tables: {missing_tables}"
)

print(
    f"✓ {len(required_figures)} publication figures"
)

print(
    f"✓ {len(required_tables)} publication tables"
)


# ============================================================
# 16.6 — TABLE 1 VALIDATION
# ============================================================

table1 = pd.read_csv(
    PUBLICATION_TAB_DIR
    / "table1_dataset_composition.csv"
)

assert set(
    table1["neurotransmitter"]
) == {
    "Acetylcholine",
    "GABA",
}

ach_count = int(
    table1.loc[
        table1["neurotransmitter"]
        == "Acetylcholine",
        "neuron_count"
    ].iloc[0]
)

gaba_count = int(
    table1.loc[
        table1["neurotransmitter"]
        == "GABA",
        "neuron_count"
    ].iloc[0]
)

assert ach_count == 142
assert gaba_count == 23

print("✓ Table 1 validated")


# ============================================================
# 16.7 — TABLE 2 VALIDATION
# ============================================================

table2 = pd.read_csv(
    PUBLICATION_TAB_DIR
    / "table2_grouped_cross_validation_model_comparison.csv"
)

assert len(table2) == 4

assert set(
    table2["model"]
) == {
    "LogisticRegression",
    "RandomForest",
    "XGBoost",
    "DummyClassifier",
}

print("✓ Table 2 validated")


# ============================================================
# 16.8 — REBUILD PACKAGE
# ============================================================

print("\n[16.8] Rebuilding clean GitHub package...")

if PACKAGE_ROOT.exists():
    shutil.rmtree(PACKAGE_ROOT)

PACKAGE_DATA_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PACKAGE_NOTEBOOK_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PACKAGE_FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PACKAGE_TABLES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("✓ Clean package created")


# ============================================================
# 16.9 — COPY DATASET
# ============================================================

shutil.copy2(
    SOURCE_DATASET,
    PACKAGE_DATA_DIR
    / "final_olfactory_neurotransmitter_ml_dataset.csv"
)

print("✓ Dataset copied")


# ============================================================
# 16.10 — COPY NOTEBOOK
# ============================================================

shutil.copy2(
    CURRENT_NOTEBOOK,
    PACKAGE_NOTEBOOK_DIR
    / "wiring_to_mol_id_final.ipynb"
)

print("✓ Notebook copied")


# ============================================================
# 16.11 — COPY PUBLICATION FIGURES
# ============================================================

for p in PUBLICATION_FIG_DIR.iterdir():

    if p.is_file():
        shutil.copy2(
            p,
            PACKAGE_FIGURES_DIR / p.name
        )

print(
    f"✓ Figures copied: "
    f"{len(list(PACKAGE_FIGURES_DIR.iterdir()))}"
)


# ============================================================
# 16.12 — COPY PUBLICATION TABLES
# ============================================================

for p in PUBLICATION_TAB_DIR.iterdir():

    if p.is_file():
        shutil.copy2(
            p,
            PACKAGE_TABLES_DIR / p.name
        )

print(
    f"✓ Tables copied: "
    f"{len(list(PACKAGE_TABLES_DIR.iterdir()))}"
)


# ============================================================
# 16.13 — COPY OTHER RESULTS
# ============================================================

obsolete_keywords = [
    "huggingface",
    "docker",
    "deployment",
    "hf_deploy",
    "space_app",
]

for source_path in RESULTS_DIR.rglob("*"):

    if not source_path.is_file():
        continue

    relative = source_path.relative_to(
        RESULTS_DIR
    )

    if "publication" in relative.parts:
        continue

    lower_name = source_path.name.lower()

    if any(
        key in lower_name
        for key in obsolete_keywords
    ):
        continue

    destination = (
        PACKAGE_RESULTS_DIR / relative
    )

    destination.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    shutil.copy2(
        source_path,
        destination
    )

print("✓ Other results copied")


# ============================================================
# 16.14 — ROOT METADATA
# ============================================================

for filename in [
    "README.md",
    "LICENSE",
    "requirements.txt",
]:

    source = project_root / filename

    assert source.exists(), (
        f"Missing {filename}"
    )

    shutil.copy2(
        source,
        PACKAGE_ROOT / filename
    )

print("✓ README copied")
print("✓ LICENSE copied")
print("✓ requirements.txt copied")


# ============================================================
# 16.15 — CITATION.CFF
# ============================================================

citation_text = """cff-version: 1.2.0
title: "Wiring to Molecular Identity: A Connectivity-Based Machine Learning Proof-of-Concept for Neurotransmitter Prediction in the Drosophila Olfactory System"
message: "If this repository, its code, methodology, analyses, or derived results materially contribute to a scholarly publication, please cite this work."
type: software
authors:
  - family-names: "Dhawan"
    given-names: "Utkarshita"
    affiliation: "TERI School of Advanced Studies"
version: "1.0.0"
date-released: "2026-09-14"
license: "MIT"
repository-code: "https://github.com/utkarshitadhawan/wiring-to-molecular-identity"
keywords:
  - Drosophila
  - olfactory system
  - connectomics
  - neurotransmitter
  - molecular identity
  - machine learning
  - computational neuroscience
  - neural connectivity
"""

with open(
    PACKAGE_ROOT / "CITATION.cff",
    "w",
    encoding="utf-8"
) as f:
    f.write(citation_text)

print("✓ CITATION.cff created")


# ============================================================
# 16.16 — GITIGNORE
# ============================================================

gitignore_text = """# Python
__pycache__/
*.py[cod]

# Jupyter
.ipynb_checkpoints/

# Environments
.venv/
venv/
env/

# OS
.DS_Store
Thumbs.db

# Secrets
.env
.env.*
*.pem
*.key

# IDE
.vscode/
.idea/
"""

with open(
    PACKAGE_ROOT / ".gitignore",
    "w",
    encoding="utf-8"
) as f:
    f.write(gitignore_text)

print("✓ .gitignore created")


# ============================================================
# 16.17 — NO PERMANENT GUI DEPLOYMENT
# ============================================================

gui_dir = PACKAGE_ROOT / "gui"

if gui_dir.exists():
    shutil.rmtree(gui_dir)

print("✓ No permanent deployment directory")


# ============================================================
# 16.18 — SECURITY AUDIT
# ============================================================

print("\n[16.18] Security audit...")

secret_hits = []

secret_patterns = [
    re.compile(
        r"ghp_[A-Za-z0-9]{20,}"
    ),
    re.compile(
        r"github_pat_[A-Za-z0-9_]{20,}"
    ),
    re.compile(
        r"gho_[A-Za-z0-9]{20,}"
    ),
    re.compile(
        r"-----BEGIN [A-Z ]*PRIVATE KEY-----"
    ),
    re.compile(
        r"""(?i)
        (?:password|passwd|secret|api[_-]?key|access[_-]?token)
        \s*=\s*
        ['"][^'"]{20,}['"]
        """
    ),
]

credential_filename_patterns = [
    re.compile(
        r"(?i)(^|[/\\])\.env($|\.|[/\\])"
    ),
    re.compile(
        r"(?i)(^|[/\\])(credentials?|secrets?)(\.|$)"
    ),
]

for p in PACKAGE_ROOT.rglob("*"):

    if not p.is_file():
        continue

    relative_name = str(
        p.relative_to(PACKAGE_ROOT)
    )

    filename_problem = False

    for pattern in credential_filename_patterns:

        if pattern.search(relative_name):
            secret_hits.append(relative_name)
            filename_problem = True
            break

    if filename_problem:
        continue

    if p.suffix.lower() not in {
        ".ipynb",
        ".py",
        ".md",
        ".txt",
        ".csv",
        ".json",
        ".yaml",
        ".yml",
        ".toml",
        ".ini",
        ".cfg",
        ".cff",
        ".gitignore",
    }:
        continue

    try:
        text = p.read_text(
            encoding="utf-8",
            errors="ignore"
        )
    except Exception:
        continue

    for pattern in secret_patterns:

        if pattern.search(text):
            secret_hits.append(relative_name)
            break

secret_hits = list(
    dict.fromkeys(secret_hits)
)

assert not secret_hits, (
    "Potential real credential/token detected: "
    f"{secret_hits}"
)

print("✓ No GitHub PATs detected")
print("✓ No OAuth tokens detected")
print("✓ No private keys detected")
print("✓ No obvious hard-coded credentials")


# ============================================================
# 16.19 — SHA256 MANIFEST
# ============================================================

print("\n[16.19] Creating SHA256 manifest...")

manifest_rows = []

for p in sorted(
    [
        x for x in PACKAGE_ROOT.rglob("*")
        if x.is_file()
        and x.name != "repository_sha256_manifest.csv"
    ],
    key=lambda x: str(
        x.relative_to(PACKAGE_ROOT)
    )
):

    relative_name = str(
        p.relative_to(PACKAGE_ROOT)
    ).replace("\\", "/")

    digest = hashlib.sha256()

    with open(p, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            digest.update(chunk)

    manifest_rows.append({
        "path": relative_name,
        "size_bytes": p.stat().st_size,
        "sha256": digest.hexdigest(),
    })

manifest_df = pd.DataFrame(
    manifest_rows
)

manifest_path = (
    PACKAGE_ROOT
    / "repository_sha256_manifest.csv"
)

manifest_df.to_csv(
    manifest_path,
    index=False
)

print(
    f"✓ SHA256 manifest: "
    f"{len(manifest_df)} files"
)


# ============================================================
# 16.20 — PACKAGE AUDIT
# ============================================================

print("\n[16.20] Final package audit...")

required_package_files = [
    "README.md",
    "LICENSE",
    "CITATION.cff",
    "requirements.txt",
    ".gitignore",
    "data/final_olfactory_neurotransmitter_ml_dataset.csv",
    "notebook/wiring_to_mol_id_final.ipynb",
    "repository_sha256_manifest.csv",
]

for relative_path in required_package_files:

    assert (
        PACKAGE_ROOT / relative_path
    ).exists(), (
        f"Missing: {relative_path}"
    )

packaged_dataset = pd.read_csv(
    PACKAGE_ROOT
    / "data"
    / "final_olfactory_neurotransmitter_ml_dataset.csv"
)

assert packaged_dataset.shape == (165, 20)

assert packaged_dataset[
    "bodyId"
].nunique() == 165

assert packaged_dataset[
    "cell_type"
].nunique() == 65

assert set(
    packaged_dataset[
        "neurotransmitter"
    ].unique()
) == {
    "acetylcholine",
    "gaba",
}

print("✓ Required package files present")
print("✓ Packaged dataset verified")
print("✓ 165 neurons")
print("✓ 65 cell types")
print("✓ 16 connectivity features")
print("✓ Two neurotransmitter classes")


# ============================================================
# 16.21 — PACKAGE INVENTORY
# ============================================================

package_files = sorted(
    [
        p for p in PACKAGE_ROOT.rglob("*")
        if p.is_file()
    ]
)

print(
    f"✓ Total package files: "
    f"{len(package_files)}"
)

print(
    f"✓ Package size: "
    f"{sum(p.stat().st_size for p in package_files) / 1024**2:.2f} MB"
)


# ============================================================
# 16.22 — GITHUB TOKEN
# ============================================================

print("\n" + "=" * 70)
print("GITHUB UPLOAD")
print("=" * 70)

github_token = getpass.getpass(
    "\nPaste your GitHub fine-grained PAT: "
)

assert github_token.strip(), (
    "GitHub PAT was not supplied."
)

headers = {
    "Authorization": f"Bearer {github_token}",
    "Accept": "application/vnd.github+json",
    "X-GitHub-Api-Version": "2022-11-28",
}

repo_api_url = (
    f"{GITHUB_API}/repos/"
    f"{GITHUB_OWNER}/{GITHUB_REPO}"
)


# ============================================================
# 16.23 — VERIFY REPOSITORY
# ============================================================

repo_response = requests.get(
    repo_api_url,
    headers=headers,
    timeout=30,
)

assert repo_response.status_code == 200, (
    "Cannot access GitHub repository.\n"
    f"HTTP {repo_response.status_code}\n"
    f"{repo_response.text[:2000]}"
)

print(
    f"✓ Repository accessible: "
    f"{GITHUB_OWNER}/{GITHUB_REPO}"
)


# ============================================================
# 16.24 — DETERMINE REPOSITORY STATE
# ============================================================

print("\n[16.24] Checking repository state...")

ref_url = (
    f"{repo_api_url}/git/ref/heads/{GITHUB_BRANCH}"
)

ref_response = requests.get(
    ref_url,
    headers=headers,
    timeout=30,
)

existing_commit_sha = None
existing_tree_sha = None

if ref_response.status_code == 200:

    ref_data = ref_response.json()

    existing_commit_sha = (
        ref_data["object"]["sha"]
    )

    commit_response = requests.get(
        f"{repo_api_url}/git/commits/"
        f"{existing_commit_sha}",
        headers=headers,
        timeout=30,
    )

    assert commit_response.status_code == 200

    existing_tree_sha = (
        commit_response.json()["tree"]["sha"]
    )

    print(
        f"✓ Existing main branch: "
        f"{existing_commit_sha[:12]}"
    )

elif ref_response.status_code == 409:

    if (
        "git repository is empty"
        in ref_response.text.lower()
    ):

        print(
            "✓ Repository is completely empty"
        )

        print(
            "✓ Bootstrapping repository..."
        )

    else:

        raise RuntimeError(
            "Unexpected HTTP 409:\n"
            f"{ref_response.text}"
        )

elif ref_response.status_code == 404:

    print(
        "✓ main branch does not exist"
    )

else:

    raise RuntimeError(
        "Unexpected GitHub response:\n"
        f"HTTP {ref_response.status_code}\n"
        f"{ref_response.text[:2000]}"
    )


# ============================================================
# 16.25 — HANDLE EMPTY REPOSITORY
#
# GitHub Git Data API cannot create blobs in a repository
# containing zero commits.
#
# Therefore create the FIRST COMMIT through the Contents API.
# ============================================================

if existing_commit_sha is None:

    print(
        "\n[16.25] Bootstrapping empty repository..."
    )

    bootstrap_path = "README.md"

    bootstrap_bytes = (
        (PACKAGE_ROOT / bootstrap_path)
        .read_bytes()
    )

    bootstrap_content = base64.b64encode(
        bootstrap_bytes
    ).decode("ascii")

    bootstrap_url = (
        f"{repo_api_url}/contents/"
        f"{bootstrap_path}"
    )

    bootstrap_response = requests.put(
        bootstrap_url,
        headers=headers,
        json={
            "message": "Initialize repository",
            "content": bootstrap_content,
            "branch": GITHUB_BRANCH,
        },
        timeout=120,
    )

    # If branch did not exist, GitHub creates it.
    # If another process initialized it between checks,
    # retrieve the resulting branch.
    assert bootstrap_response.status_code in {
        201,
        200,
    }, (
        "Failed to bootstrap GitHub repository.\n"
        f"HTTP {bootstrap_response.status_code}\n"
        f"{bootstrap_response.text[:2000]}"
    )

    print(
        "✓ Repository bootstrapped with README.md"
    )

    # Retrieve the newly created branch.
    ref_response = requests.get(
        ref_url,
        headers=headers,
        timeout=30,
    )

    assert ref_response.status_code == 200, (
        "Bootstrap succeeded but main branch "
        "could not be retrieved.\n"
        f"HTTP {ref_response.status_code}\n"
        f"{ref_response.text[:2000]}"
    )

    existing_commit_sha = (
        ref_response.json()["object"]["sha"]
    )

    commit_response = requests.get(
        f"{repo_api_url}/git/commits/"
        f"{existing_commit_sha}",
        headers=headers,
        timeout=30,
    )

    assert commit_response.status_code == 200

    existing_tree_sha = (
        commit_response.json()["tree"]["sha"]
    )

    print(
        f"✓ Bootstrap commit: "
        f"{existing_commit_sha[:12]}"
    )

    print(
        f"✓ Bootstrap tree: "
        f"{existing_tree_sha[:12]}"
    )


# ============================================================
# 16.26 — CREATE BLOBS
# ============================================================

print(
    "\n[16.26] Uploading repository files..."
)

tree_entries = []

for i, p in enumerate(
    package_files,
    start=1
):

    relative_path = str(
        p.relative_to(PACKAGE_ROOT)
    ).replace("\\", "/")

    file_bytes = p.read_bytes()

    blob_response = requests.post(
        f"{repo_api_url}/git/blobs",
        headers=headers,
        json={
            "content": base64.b64encode(
                file_bytes
            ).decode("ascii"),
            "encoding": "base64",
        },
        timeout=180,
    )

    assert blob_response.status_code == 201, (
        f"Failed to upload blob:\n"
        f"{relative_path}\n"
        f"HTTP {blob_response.status_code}\n"
        f"{blob_response.text[:2000]}"
    )

    blob_sha = (
        blob_response.json()["sha"]
    )

    tree_entries.append({
        "path": relative_path,
        "mode": "100644",
        "type": "blob",
        "sha": blob_sha,
    })

    print(
        f"[{i:>3}/{len(package_files)}] "
        f"{relative_path}"
    )

print(
    f"✓ {len(tree_entries)} blobs created"
)


# ============================================================
# 16.27 — CREATE COMPLETE TREE
# ============================================================

print("\n[16.27] Creating complete Git tree...")

tree_response = requests.post(
    f"{repo_api_url}/git/trees",
    headers=headers,
    json={
        "base_tree": existing_tree_sha,
        "tree": tree_entries,
    },
    timeout=180,
)

assert tree_response.status_code == 201, (
    "Failed to create Git tree.\n"
    f"HTTP {tree_response.status_code}\n"
    f"{tree_response.text[:3000]}"
)

new_tree_sha = (
    tree_response.json()["sha"]
)

print(
    f"✓ Tree created: {new_tree_sha}"
)


# ============================================================
# 16.28 — CREATE FINAL COMMIT
# ============================================================

print("\n[16.28] Creating final repository commit...")

commit_response = requests.post(
    f"{repo_api_url}/git/commits",
    headers=headers,
    json={
        "message": (
            "Initial release: "
            "connectivity-based neurotransmitter prediction"
        ),
        "tree": new_tree_sha,
        "parents": [existing_commit_sha],
    },
    timeout=120,
)

assert commit_response.status_code == 201, (
    "Failed to create final commit.\n"
    f"HTTP {commit_response.status_code}\n"
    f"{commit_response.text[:3000]}"
)

new_commit_sha = (
    commit_response.json()["sha"]
)

print(
    f"✓ Final commit created: "
    f"{new_commit_sha}"
)


# ============================================================
# 16.29 — UPDATE MAIN
# ============================================================

print("\n[16.29] Updating main branch...")

update_ref_response = requests.patch(
    ref_url,
    headers=headers,
    json={
        "sha": new_commit_sha,
        "force": True,
    },
    timeout=60,
)

assert update_ref_response.status_code == 200, (
    "Failed to update main branch.\n"
    f"HTTP {update_ref_response.status_code}\n"
    f"{update_ref_response.text[:3000]}"
)

print(
    "✓ main branch updated"
)


# ============================================================
# 16.30 — VERIFY REMOTE COMMIT
# ============================================================

print("\n[16.30] Verifying remote repository...")

verify_response = requests.get(
    ref_url,
    headers=headers,
    timeout=30,
)

assert verify_response.status_code == 200

remote_commit_sha = (
    verify_response.json()["object"]["sha"]
)

assert remote_commit_sha == new_commit_sha, (
    "Remote commit does not match local upload."
)

print(
    f"✓ Remote commit verified: "
    f"{remote_commit_sha}"
)


# ============================================================
# 16.31 — VERIFY IMPORTANT FILES
# ============================================================

print(
    "\n[16.31] Verifying important remote files..."
)

remote_files = [
    "README.md",
    "LICENSE",
    "CITATION.cff",
    "requirements.txt",
    "data/final_olfactory_neurotransmitter_ml_dataset.csv",
    "notebook/wiring_to_mol_id_final.ipynb",
    "results/figures/figure3_pca_connectivity_architecture.pdf",
    "results/figures/figure4_grouped_cv_performance.pdf",
    "results/figures/figure5_permutation_feature_importance.pdf",
    "results/figures/figure6_feature_family_contribution.pdf",
    "results/tables/table1_dataset_composition.csv",
    "results/tables/table2_grouped_cross_validation_model_comparison.csv",
]

for relative_path in remote_files:

    url = (
        f"{repo_api_url}/contents/"
        f"{relative_path}"
        f"?ref={GITHUB_BRANCH}"
    )

    response = requests.get(
        url,
        headers=headers,
        timeout=60,
    )

    assert response.status_code == 200, (
        f"Remote file missing:\n"
        f"{relative_path}\n"
        f"HTTP {response.status_code}"
    )

    print(
        f"✓ {relative_path}"
    )


# ============================================================
# 16.32 — TOKEN CLEANUP
# ============================================================

github_token = None
headers = None

print(
    "\n✓ GitHub token cleared from memory"
)


# ============================================================
# 16.33 — FINAL SUCCESS
# ============================================================

print("\n")
print("=" * 70)
print("CHECKPOINT 16 COMPLETE")
print("=" * 70)

print(
    "✓ Final package built"
)

print(
    "✓ Final dataset validated"
)

print(
    "✓ Final notebook validated"
)

print(
    "✓ Publication figures validated"
)

print(
    "✓ Publication tables validated"
)

print(
    "✓ Security audit passed"
)

print(
    "✓ SHA256 manifest created"
)

print(
    "✓ No permanent Hugging Face deployment included"
)

print(
    "✓ GitHub upload completed without Git installation"
)

print(
    "✓ main branch verified"
)

print(
    f"✓ Commit: {new_commit_sha}"
)

print(
    "\nGitHub:"
)

print(
    f"https://github.com/"
    f"{GITHUB_OWNER}/"
    f"{GITHUB_REPO}"
)

print(
    "\nNEXT: Create GitHub v1.0.0 release, "
    "then archive on Zenodo."
)

print("=" * 70)
```
